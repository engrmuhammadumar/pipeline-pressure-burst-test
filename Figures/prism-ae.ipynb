```python
"""
PRISM-AE — Stage 1: pressure-confounding and reliability-value audit
====================================================================

This notebook-style Python file is the first implementation stage for:

    PRISM-AE: Pressure-Residualized, sensor-Invariant Stochastic
    Monitoring using Acoustic Emission.

Open this file in VS Code/Jupyter and run one ``# %%`` cell at a time.

Scientific goals of Stage 1
---------------------------
1. Load the existing B/C/D/E feature archives without touching the 445+ GB WFS files.
2. require externally verified rupture times for publishable RUL labels;
3. reconstruct the nominal pressure program, or ingest actual pressure logs;
4. create causal, bandwidth-aware, sensor-invariant AE features;
5. fit pressure-response models using training pipes only;
6. evaluate age-only, pressure-only, raw-AE, and PRISM residual-AE models under
   strict leave-one-pipe-out (LOPO) validation;
7. quantify whether AE adds predictive value beyond age and pressure;
8. produce an auditable Stage-1 result package for deciding whether the full
   stochastic reliability model is justified.

Important
---------
The uncertainty intervals in Stage 1 are conservative group-jackknife pilot
intervals. They are not the final Bayesian aleatory/epistemic decomposition.
That is Stage 2, after Stage 1 establishes that AE provides incremental value.
"""


# %% CELL 1 — Environment, imports, and project configuration
# Run once. If an import fails, uncomment and run the installation line first.

# %pip install -U numpy pandas scipy scikit-learn matplotlib seaborn

from __future__ import annotations

import json
import math
import os
import warnings
from dataclasses import dataclass, field
from pathlib import Path
from typing import Dict, Iterable, List, Mapping, Optional, Sequence, Tuple

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.base import clone
from sklearn.compose import TransformedTargetRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler


SEED = 20260829
rng = np.random.default_rng(SEED)
np.random.seed(SEED)

mpl.rcParams.update(
    {
        "figure.dpi": 120,
        "savefig.dpi": 300,
        "font.family": "DejaVu Sans",
        "font.size": 11,
        "axes.labelsize": 11,
        "axes.titlesize": 12,
        "axes.linewidth": 1.0,
        "axes.grid": True,
        "grid.alpha": 0.20,
        "legend.frameon": False,
        "savefig.bbox": "tight",
        "savefig.facecolor": "white",
    }
)


@dataclass
class Config:
    # Existing feature files: B.npz, C.npz, D.npz, E.npz
    feature_dir: Path = Path(r"D:\Pipeline RUL Data\data\features")

    # A new result directory; Stage 1 never overwrites the original feature files.
    output_dir: Path = Path(r"F:\Pipeline Prressure Work\Code\results_prism_stage1")

    pipes: Tuple[str, ...] = ("B", "C", "D", "E")
    channels: Tuple[int, ...] = (5, 6, 7, 8)

    # These must match the six bands stored in E[:, channel, band].
    bands_khz: Tuple[Tuple[float, float], ...] = (
        (20.0, 60.0),
        (60.0, 100.0),
        (100.0, 140.0),
        (140.0, 190.0),
        (190.0, 250.0),
        (250.0, 400.0),
    )
    lf_band_indices: Tuple[int, ...] = (1, 2, 3)   # 60–190 kHz
    hf_band_indices: Tuple[int, ...] = (4, 5)      # 190–400 kHz

    # Robust temporal aggregation. Ten seconds gives about 400–700 bins per pipe.
    bin_seconds: float = 10.0
    initial_baseline_minutes: float = 10.0
    activity_mad_multiplier: float = 6.0

    # Post-hoc evaluation fractions. These are allowed for benchmarking, but they
    # are NOT model inputs because the true life fraction is unknown online.
    observation_fractions: Tuple[float, ...] = tuple(np.arange(0.10, 0.901, 0.05))
    rolling_windows_minutes: Tuple[float, ...] = (1.0, 5.0, 10.0)

    # Fill these from the experimental log/video. They are minutes measured from
    # the first feature timestamp. Do not use an AE-derived endpoint in the paper.
    rupture_min_override: Dict[str, Optional[float]] = field(
        default_factory=lambda: {"B": None, "C": None, "D": None, "E": None}
    )

    # Offset between feature time zero and start of the published pressure program.
    # None is deliberate: the diagnostic plots show that assuming zero can be wrong.
    pressure_start_min_override: Dict[str, Optional[float]] = field(
        default_factory=lambda: {"B": None, "C": None, "D": None, "E": None}
    )

    # Exploratory controls. Leave both False for publication-grade analysis.
    # If enabled, the manifest and plot labels explicitly identify proxy use.
    allow_ae_proxy_endpoint: bool = False
    ae_proxy_endpoint_method: str = "existing_all_rms_gt3"
    allow_ae_proxy_pressure_start: bool = False
    pressure_start_search_min: Tuple[float, float] = (5.0, 35.0)
    pressure_start_smoothing_sec: float = 30.0
    pressure_start_contrast_min: float = 2.0

    # Ridge penalties searched by inner leave-one-training-pipe-out validation.
    ridge_alphas: Tuple[float, ...] = tuple(np.logspace(-4, 4, 33))

    # Group-jackknife interval target. With only three outer-training pipes this
    # remains exploratory and deliberately conservative.
    interval_level: float = 0.90

    random_state: int = SEED


CFG = Config()

CFG.feature_dir = Path(r"D:\Pipeline RUL Data\data\features")
CFG.output_dir = Path(
    r"F:\Pipeline Prressure Work\Code\results_prism_stage1"
)

# Exploratory development run
CFG.allow_ae_proxy_endpoint = True
CFG.ae_proxy_endpoint_method = "existing_all_rms_gt3"

# Estimate pressure-program start offsets from the early AE transitions
CFG.allow_ae_proxy_pressure_start = True

CFG.output_dir.mkdir(parents=True, exist_ok=True)
CFG.output_dir.mkdir(parents=True, exist_ok=True)

PIPE_COLORS = {"B": "#4477AA", "C": "#EE6677", "D": "#228833", "E": "#CCBB44"}
METHOD_COLORS = {
    "Age only": "#777777",
    "Pressure only": "#AA4499",
    "Pressure + raw AE": "#44AA99",
    "PRISM-AE residual": "#332288",
}

print("Feature directory:", CFG.feature_dir)
print("Output directory :", CFG.output_dir)
print("Pipes            :", CFG.pipes)


# %% CELL 2 — Data structures, numerical utilities, and schema-safe loading

EPS = 1e-12


@dataclass
class PipeFeatures:
    pipe: str
    t_sec: np.ndarray          # [N]
    rms: np.ndarray            # [N, 4]
    band_energy: np.ndarray    # [N, 4, 6]
    source_path: Path

    @property
    def t_min(self) -> np.ndarray:
        return self.t_sec / 60.0

    @property
    def duration_min(self) -> float:
        return float((self.t_sec[-1] - self.t_sec[0]) / 60.0)


def robust_mad(x: np.ndarray, axis=None) -> np.ndarray:
    """Median absolute deviation, scaled to be Gaussian-consistent."""
    x = np.asarray(x, dtype=float)
    med = np.nanmedian(x, axis=axis, keepdims=True)
    mad = 1.4826 * np.nanmedian(np.abs(x - med), axis=axis)
    return np.maximum(mad, EPS)


def safe_log(x: np.ndarray) -> np.ndarray:
    return np.log(np.maximum(np.asarray(x, dtype=float), EPS))


def normalized_entropy(weights: np.ndarray, axis: int = -1) -> np.ndarray:
    """Entropy in [0, 1]; 1 means equal sensor contribution."""
    w = np.asarray(weights, dtype=float)
    w = np.clip(w, 0.0, None)
    w = w / (w.sum(axis=axis, keepdims=True) + EPS)
    h = -(w * np.log(w + EPS)).sum(axis=axis)
    n = w.shape[axis]
    return h / np.log(max(n, 2))


def _select_active_channels(a: np.ndarray, expected_channels: int, name: str) -> np.ndarray:
    """Support feature archives containing either four active or all eight channels."""
    if a.ndim < 2:
        raise ValueError(f"{name} must have a channel axis; got {a.shape}")
    if a.shape[1] == expected_channels:
        return a
    if a.shape[1] == 8 and expected_channels == 4:
        warnings.warn(f"{name} contains 8 channels; selecting channels 5–8 (columns 4:8).")
        return a[:, 4:8, ...]
    raise ValueError(
        f"{name} has {a.shape[1]} channels; expected {expected_channels} or 8."
    )


def load_pipe_npz(path: Path, pipe: str, cfg: Config = CFG) -> PipeFeatures:
    with np.load(path, allow_pickle=False) as z:
        missing = {"t", "RMS", "E"} - set(z.files)
        if missing:
            raise KeyError(f"{path.name} is missing keys: {sorted(missing)}; found {z.files}")
        t = np.asarray(z["t"], dtype=np.float64).reshape(-1)
        rms = np.asarray(z["RMS"], dtype=np.float64)
        energy = np.asarray(z["E"], dtype=np.float64)

    rms = _select_active_channels(rms, len(cfg.channels), "RMS")
    energy = _select_active_channels(energy, len(cfg.channels), "E")

    if rms.ndim != 2:
        raise ValueError(f"RMS must be [time, channel]; got {rms.shape}")
    if energy.ndim != 3:
        raise ValueError(f"E must be [time, channel, band]; got {energy.shape}")
    if not (len(t) == rms.shape[0] == energy.shape[0]):
        raise ValueError(f"Time mismatch in {path.name}: t={len(t)}, RMS={rms.shape}, E={energy.shape}")
    if energy.shape[2] != len(cfg.bands_khz):
        raise ValueError(
            f"E has {energy.shape[2]} bands but config defines {len(cfg.bands_khz)} bands."
        )

    finite_t = np.isfinite(t)
    t, rms, energy = t[finite_t], rms[finite_t], energy[finite_t]
    order = np.argsort(t, kind="stable")
    t, rms, energy = t[order], rms[order], energy[order]

    # Remove duplicate timestamps deterministically; retain the first observation.
    keep = np.r_[True, np.diff(t) > 0]
    t, rms, energy = t[keep], rms[keep], energy[keep]

    if len(t) < 20:
        raise ValueError(f"{path.name} contains too few usable observations: {len(t)}")
    if np.nanmin(rms) < 0 or np.nanmin(energy) < 0:
        warnings.warn(f"{path.name} has negative RMS/energy values; verify feature generation.")

    return PipeFeatures(pipe=pipe, t_sec=t, rms=rms, band_energy=energy, source_path=path)


def load_all_pipes(cfg: Config = CFG) -> Dict[str, PipeFeatures]:
    out: Dict[str, PipeFeatures] = {}
    rows = []
    for pipe in cfg.pipes:
        path = cfg.feature_dir / f"{pipe}.npz"
        if not path.exists():
            raise FileNotFoundError(f"Missing feature file: {path}")
        d = load_pipe_npz(path, pipe, cfg)
        out[pipe] = d
        rows.append(
            {
                "pipe": pipe,
                "rows": len(d.t_sec),
                "duration_min": d.duration_min,
                "median_dt_sec": float(np.nanmedian(np.diff(d.t_sec))),
                "rms_shape": str(d.rms.shape),
                "energy_shape": str(d.band_energy.shape),
            }
        )
    report = pd.DataFrame(rows)
    print(report.to_string(index=False))
    report.to_csv(cfg.output_dir / "01_input_schema_report.csv", index=False)
    return out


# Run this line on the real data.
PIPEDATA = load_all_pipes(CFG)


# %% CELL 3 — Endpoint diagnostics and publication-safe label resolution

def ae_proxy_endpoint_minutes(d: PipeFeatures, cfg: Config = CFG) -> Dict[str, float]:
    """
    Produce several AE-derived endpoint candidates for manual inspection only.

    None of these is acceptable as physical rupture ground truth without external
    verification. Multiple candidates are printed precisely to expose sensitivity.
    """
    t_min = d.t_min
    log_rms = np.log1p(np.maximum(d.rms, 0.0))
    baseline = t_min <= (t_min[0] + cfg.initial_baseline_minutes)
    if baseline.sum() < 5:
        baseline = np.arange(len(t_min)) < max(5, len(t_min) // 20)

    med = np.nanmedian(log_rms[baseline], axis=0)
    mad = robust_mad(log_rms[baseline], axis=0)
    active = log_rms > (med + cfg.activity_mad_multiplier * mad)

    candidates: Dict[str, float] = {}
    for n_required in (1, 2, 3, 4):
        idx = np.where(active.sum(axis=1) >= n_required)[0]
        candidates[f"last_{n_required}plus_active"] = float(t_min[idx[-1]]) if len(idx) else np.nan

    # Existing heuristic, included only so its sensitivity can be compared.
    idx_all_gt3 = np.where((d.rms > 3.0).all(axis=1))[0]
    candidates["existing_all_rms_gt3"] = (
        float(t_min[idx_all_gt3[-1]]) if len(idx_all_gt3) else np.nan
    )
    candidates["recording_end"] = float(t_min[-1])
    return candidates


def plot_endpoint_diagnostics(data: Mapping[str, PipeFeatures], cfg: Config = CFG) -> pd.DataFrame:
    rows = []
    fig, axes = plt.subplots(len(cfg.pipes), 1, figsize=(12, 10), sharex=False)
    if len(cfg.pipes) == 1:
        axes = [axes]

    for ax, pipe in zip(axes, cfg.pipes):
        d = data[pipe]
        cand = ae_proxy_endpoint_minutes(d, cfg)
        rows.append({"pipe": pipe, **cand})

        y = np.nanmedian(np.log1p(d.rms), axis=1)
        ax.plot(d.t_min, y, color=PIPE_COLORS[pipe], lw=0.8, label="Median log(1+RMS)")
        for name, color, ls in (
            ("last_2plus_active", "#EE7733", "--"),
            ("existing_all_rms_gt3", "#CC3311", ":"),
            ("recording_end", "#333333", "-."),
        ):
            x = cand[name]
            if np.isfinite(x):
                ax.axvline(x, color=color, ls=ls, lw=1.4, label=name)
        ax.set_ylabel(f"Pipe {pipe}\nlog RMS")
        ax.legend(ncol=4, fontsize=8, loc="upper left")

    axes[-1].set_xlabel("Feature time (min)")
    fig.suptitle("Endpoint candidates — diagnostic only, not rupture ground truth", y=1.005)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "02_endpoint_diagnostics.png")
    plt.show()

    table = pd.DataFrame(rows)
    table.to_csv(cfg.output_dir / "02_endpoint_candidates_NOT_ground_truth.csv", index=False)
    print(table.round(3).to_string(index=False))
    return table


ENDPOINT_CANDIDATES = plot_endpoint_diagnostics(PIPEDATA, CFG)


def resolve_rupture_times(
    data: Mapping[str, PipeFeatures], cfg: Config = CFG
) -> Dict[str, float]:
    valid_proxy_methods = {
        "last_1plus_active",
        "last_2plus_active",
        "last_3plus_active",
        "last_4plus_active",
        "existing_all_rms_gt3",
    }
    if cfg.ae_proxy_endpoint_method not in valid_proxy_methods:
        raise ValueError(
            f"Unknown ae_proxy_endpoint_method={cfg.ae_proxy_endpoint_method!r}. "
            f"Choose one of {sorted(valid_proxy_methods)}."
        )

    rupture: Dict[str, float] = {}
    missing = []
    for pipe in cfg.pipes:
        v = cfg.rupture_min_override.get(pipe)
        if v is None or not np.isfinite(v):
            missing.append(pipe)
        else:
            rupture[pipe] = float(v)

    if missing and not cfg.allow_ae_proxy_endpoint:
        suggestions = {
            p: {
                "shutdown_onset": ae_proxy_endpoint_minutes(data[p], cfg)["existing_all_rms_gt3"],
                "last_2plus_active": ae_proxy_endpoint_minutes(data[p], cfg)["last_2plus_active"],
            }
            for p in missing
        }
        raise RuntimeError(
            "Verified rupture times are missing for "
            f"{missing}. Fill CFG.rupture_min_override and rerun from CELL 1. "
            "Alternatively, for development only, set allow_ae_proxy_endpoint=True. "
            "AE-derived inspection candidates (NOT ground truth) are: "
            f"{suggestions}"
        )

    for pipe in missing:
        rupture[pipe] = ae_proxy_endpoint_minutes(data[pipe], cfg)[
            cfg.ae_proxy_endpoint_method
        ]
        warnings.warn(
            f"Pipe {pipe}: using AE-derived endpoint method "
            f"{cfg.ae_proxy_endpoint_method!r}. Results are exploratory and must not be "
            "presented as physical RUL validation."
        )

    for pipe, value in rupture.items():
        if not (data[pipe].t_min[0] < value <= data[pipe].t_min[-1] + 1e-6):
            raise ValueError(
                f"Pipe {pipe}: rupture_min={value:.3f} lies outside recording range "
                f"[{data[pipe].t_min[0]:.3f}, {data[pipe].t_min[-1]:.3f}]."
            )
    return rupture


# IMPORTANT: by default this line stops until physical rupture times are filled.
RUPTURE_MIN = resolve_rupture_times(PIPEDATA, CFG)
print("Resolved rupture times (min):", RUPTURE_MIN)
print(
    "Endpoint provenance:",
    (
        f"EXPLORATORY AE proxy ({CFG.ae_proxy_endpoint_method})"
        if CFG.allow_ae_proxy_endpoint
        else "external/physical overrides"
    ),
)


# %% CELL 4 — Nominal pressure program and optional actual-pressure ingestion

def nominal_pressure_pd(t_min: np.ndarray, start_min: float = 0.0) -> np.ndarray:
    """
    Reconstruct the pressure/design-pressure ratio described in the experiment.

    Sequence after pressure-program start:
        0–10 min: 0.50 Pd
       10–20 min: 0.65 Pd
       20–30 min: 0.80 Pd
       30–60 min: 1.00 Pd
       60–65 min: 2.00 Pd
       65–70 min: 4.00 Pd
       70–75 min: 6.00 Pd
       >=75 min : 7.00 Pd, then +1.00 Pd every five minutes.
    """
    tau = np.asarray(t_min, dtype=float) - float(start_min)
    p = np.zeros_like(tau)
    p[(tau >= 0) & (tau < 10)] = 0.50
    p[(tau >= 10) & (tau < 20)] = 0.65
    p[(tau >= 20) & (tau < 30)] = 0.80
    p[(tau >= 30) & (tau < 60)] = 1.00
    p[(tau >= 60) & (tau < 65)] = 2.00
    p[(tau >= 65) & (tau < 70)] = 4.00
    p[(tau >= 70) & (tau < 75)] = 6.00
    late = tau >= 75
    p[late] = 7.00 + np.floor((tau[late] - 75.0) / 5.0)
    return p


def time_in_current_hold(t_min: np.ndarray, start_min: float = 0.0) -> np.ndarray:
    tau = np.asarray(t_min, dtype=float) - float(start_min)
    out = np.zeros_like(tau)
    early_edges = np.array([0.0, 10.0, 20.0, 30.0, 60.0, 65.0, 70.0, 75.0])
    for i, value in enumerate(tau):
        if value < 0:
            out[i] = 0.0
        elif value >= 75.0:
            out[i] = (value - 75.0) % 5.0
        else:
            out[i] = value - early_edges[early_edges <= value].max()
    return out


def load_actual_pressure_log(pipe: str, cfg: Config = CFG) -> Optional[pd.DataFrame]:
    """
    Optional CSV convention:
      <feature_dir>/pressure/B_pressure.csv
      columns: time_min, pressure_pd

    If unavailable, the nominal program is used and explicitly flagged.
    """
    path = cfg.feature_dir / "pressure" / f"{pipe}_pressure.csv"
    if not path.exists():
        return None
    p = pd.read_csv(path)
    required = {"time_min", "pressure_pd"}
    if not required.issubset(p.columns):
        raise ValueError(f"{path} must contain columns {sorted(required)}")
    p = p[["time_min", "pressure_pd"]].dropna().sort_values("time_min")
    if p.empty:
        raise ValueError(f"Pressure log is empty: {path}")
    return p


def estimate_pressure_start_from_ae(
    d: PipeFeatures, cfg: Config = CFG
) -> Dict[str, object]:
    """
    Estimate an inspection candidate for the pressure-program start.

    This is not a measured pressure timestamp. The estimator searches the early
    record for the first near-maximum sustained positive level shift in the
    sensor-median log-RMS. It is useful for finding acquisition/loading offsets,
    but its result remains an AE-derived proxy and must be sensitivity-tested.
    """
    t = np.asarray(d.t_min, dtype=float)
    y = np.nanmedian(np.log1p(np.maximum(d.rms, 0.0)), axis=1)
    finite = np.isfinite(t) & np.isfinite(y)
    t, y = t[finite], y[finite]
    if len(t) < 50:
        raise ValueError(f"Pipe {d.pipe}: too few finite samples for alignment diagnostic")

    # Work on a one-second grid so the calculation is fast and comparable among pipes.
    grid_dt_min = 1.0 / 60.0
    grid = np.arange(t[0], t[-1] + 0.5 * grid_dt_min, grid_dt_min)
    yg = np.interp(grid, t, y)

    smooth_n = max(5, int(round(cfg.pressure_start_smoothing_sec)))
    contrast_n = max(30, int(round(cfg.pressure_start_contrast_min * 60.0)))
    smooth = (
        pd.Series(yg)
        .rolling(smooth_n, center=True, min_periods=max(3, smooth_n // 3))
        .median()
    )
    before = smooth.rolling(contrast_n, min_periods=max(10, contrast_n // 2)).median()
    after = (
        smooth.iloc[::-1]
        .rolling(contrast_n, min_periods=max(10, contrast_n // 2))
        .median()
        .iloc[::-1]
    )
    contrast = (after - before).to_numpy(float)

    lo, hi = cfg.pressure_start_search_min
    search = (grid >= lo) & (grid <= min(hi, grid[-1])) & np.isfinite(contrast)
    if not search.any():
        raise ValueError(
            f"Pipe {d.pipe}: no valid samples in pressure-start search window {lo}–{hi} min"
        )

    peak = float(np.nanmax(contrast[search]))
    # Prefer the first sustained transition when several transitions are almost tied.
    near_peak = search & (contrast >= 0.85 * peak)
    idx = int(np.flatnonzero(near_peak)[0])
    candidate = float(grid[idx])

    search_values = contrast[search]
    score_scale = float(robust_mad(search_values))
    score = peak / (score_scale + EPS)
    return {
        "pipe": d.pipe,
        "candidate_min": candidate,
        "contrast_log_rms": peak,
        "robust_score": score,
        "grid_min": grid,
        "raw_log_rms": yg,
        "smooth_log_rms": smooth.to_numpy(float),
        "contrast": contrast,
    }


def plot_pressure_start_diagnostics(
    data: Mapping[str, PipeFeatures], cfg: Config = CFG
) -> pd.DataFrame:
    """Plot and export AE-derived pressure-start inspection candidates."""
    details = {p: estimate_pressure_start_from_ae(data[p], cfg) for p in cfg.pipes}
    rows = []
    fig, axes = plt.subplots(len(cfg.pipes), 1, figsize=(12.5, 10.5), sharex=False)
    if len(cfg.pipes) == 1:
        axes = [axes]

    for ax, pipe in zip(axes, cfg.pipes):
        r = details[pipe]
        candidate = float(r["candidate_min"])
        rows.append(
            {
                "pipe": pipe,
                "ae_candidate_min": candidate,
                "contrast_log_rms": float(r["contrast_log_rms"]),
                "robust_score": float(r["robust_score"]),
                "manual_override_min": cfg.pressure_start_min_override.get(pipe),
                "actual_pressure_log": load_actual_pressure_log(pipe, cfg) is not None,
            }
        )
        ax.plot(
            r["grid_min"], r["raw_log_rms"], color=PIPE_COLORS[pipe],
            lw=0.35, alpha=0.35, label="Median log(1+RMS), 1-s grid"
        )
        ax.plot(
            r["grid_min"], r["smooth_log_rms"], color=PIPE_COLORS[pipe],
            lw=1.2, label="30-s rolling median"
        )
        ax.axvline(
            candidate, color="#CC3311", ls="--", lw=1.5,
            label=f"AE alignment candidate = {candidate:.2f} min"
        )
        override = cfg.pressure_start_min_override.get(pipe)
        if override is not None and np.isfinite(override):
            ax.axvline(
                float(override), color="#0077BB", ls=":", lw=1.8,
                label=f"configured start = {float(override):.2f} min"
            )
        ax.set_xlim(0, min(45.0, data[pipe].duration_min))
        ax.set_ylabel(f"Pipe {pipe}\nlog RMS")
        ax.legend(fontsize=8, ncol=3, loc="best")

    axes[-1].set_xlabel("Feature time (min)")
    fig.suptitle(
        "Pressure-program start alignment — AE candidates for inspection, not measurements",
        y=1.005,
    )
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "03_pressure_start_alignment_diagnostics.png")
    plt.show()

    table = pd.DataFrame(rows)
    table.to_csv(
        cfg.output_dir / "03_pressure_start_candidates_NOT_measurements.csv", index=False
    )
    print(table.round(3).to_string(index=False))
    return table


PRESSURE_START_CANDIDATES = plot_pressure_start_diagnostics(PIPEDATA, CFG)


def resolve_pressure_starts(
    data: Mapping[str, PipeFeatures], candidates: pd.DataFrame, cfg: Config = CFG
) -> Dict[str, Optional[float]]:
    """Resolve nominal-program offsets while keeping proxy provenance explicit."""
    candidate_map = candidates.set_index("pipe")["ae_candidate_min"].to_dict()
    resolved: Dict[str, Optional[float]] = {}
    missing = []

    for pipe in cfg.pipes:
        if load_actual_pressure_log(pipe, cfg) is not None:
            resolved[pipe] = None  # not used when measured pressure is available
            continue
        value = cfg.pressure_start_min_override.get(pipe)
        if value is not None and np.isfinite(value):
            resolved[pipe] = float(value)
        elif cfg.allow_ae_proxy_pressure_start:
            resolved[pipe] = float(candidate_map[pipe])
            warnings.warn(
                f"Pipe {pipe}: nominal pressure program aligned with an AE-derived start "
                "candidate. This is exploratory, not measured pressure alignment."
            )
        else:
            missing.append(pipe)

    if missing:
        suggestions = {p: float(candidate_map[p]) for p in missing}
        raise RuntimeError(
            "Pressure-program start offsets are unresolved for "
            f"{missing}. Supply pressure CSVs or fill CFG.pressure_start_min_override. "
            "For development only, set allow_ae_proxy_pressure_start=True. "
            f"AE-derived alignment candidates (NOT measurements) are: {suggestions}"
        )

    return resolved


PRESSURE_START_MIN = resolve_pressure_starts(
    PIPEDATA, PRESSURE_START_CANDIDATES, CFG
)
# Downstream functions read this mapping. Keeping it in CFG also serializes it.
CFG.pressure_start_min_override.update(PRESSURE_START_MIN)
print("Resolved pressure-program starts (min):", PRESSURE_START_MIN)


def attach_pressure_columns(df: pd.DataFrame, pipe: str, cfg: Config = CFG) -> pd.DataFrame:
    out = df.copy()
    actual = load_actual_pressure_log(pipe, cfg)
    if actual is None:
        start = cfg.pressure_start_min_override[pipe]
        if start is None or not np.isfinite(start):
            raise RuntimeError(f"Pipe {pipe}: unresolved nominal pressure-program start")
        out["pressure_pd"] = nominal_pressure_pd(out["time_min"].to_numpy(), start)
        out["time_in_hold_min"] = time_in_current_hold(out["time_min"].to_numpy(), start)
        out["pressure_source"] = "nominal_program"
    else:
        out["pressure_pd"] = np.interp(
            out["time_min"], actual["time_min"], actual["pressure_pd"],
            left=actual["pressure_pd"].iloc[0], right=actual["pressure_pd"].iloc[-1]
        )
        # Approximate time since most recent pressure change.
        p = out["pressure_pd"].to_numpy()
        t = out["time_min"].to_numpy()
        change = np.r_[True, np.abs(np.diff(p)) > 1e-6]
        last_change = np.maximum.accumulate(np.where(change, np.arange(len(t)), 0))
        out["time_in_hold_min"] = t - t[last_change]
        out["pressure_source"] = "actual_log"

    p = out["pressure_pd"].to_numpy(dtype=float)
    t = out["time_min"].to_numpy(dtype=float)
    dt = np.r_[0.0, np.maximum(np.diff(t), 0.0)]
    out["delta_pressure_pd"] = np.r_[0.0, np.diff(p)]
    out["abs_delta_pressure_pd"] = np.abs(out["delta_pressure_pd"])
    out["dose_1"] = np.cumsum(np.maximum(p - 1.0, 0.0) * dt)
    out["dose_2"] = np.cumsum(np.maximum(p - 1.0, 0.0) ** 2 * dt)
    out["dose_4"] = np.cumsum(np.maximum(p - 1.0, 0.0) ** 4 * dt)
    return out


# %% CELL 5 — Causal, bandwidth-aware, sensor-invariant time-bin features

def sensor_invariant_statistics(values: np.ndarray, prefix: str) -> Dict[str, np.ndarray]:
    """values: [time, sensor]. Return fixed-order sensor-set statistics."""
    return {
        f"{prefix}_med": np.nanmedian(values, axis=1),
        f"{prefix}_mean": np.nanmean(values, axis=1),
        f"{prefix}_std": np.nanstd(values, axis=1),
        f"{prefix}_iqr": np.nanpercentile(values, 75, axis=1)
        - np.nanpercentile(values, 25, axis=1),
        f"{prefix}_max": np.nanmax(values, axis=1),
    }


def build_pipe_time_table(d: PipeFeatures, rupture_min: float, cfg: Config = CFG) -> pd.DataFrame:
    """Aggregate irregular probes into robust fixed-width bins, then pool sensors invariantly."""
    t0 = float(d.t_sec[0])
    bin_id = np.floor((d.t_sec - t0) / cfg.bin_seconds).astype(np.int64)

    raw = {"bin_id": bin_id, "t_sec": d.t_sec}
    for j, ch in enumerate(cfg.channels):
        raw[f"rms_ch{ch}"] = d.rms[:, j]
        for b in range(len(cfg.bands_khz)):
            raw[f"e_ch{ch}_b{b}"] = d.band_energy[:, j, b]

    # Median is intentionally used to resist isolated acquisition spikes.
    binned = pd.DataFrame(raw).groupby("bin_id", sort=True).median(numeric_only=True)
    binned = binned.reset_index(drop=False)
    binned["time_min"] = (binned["t_sec"] - t0) / 60.0
    binned["pipe"] = d.pipe
    binned["rupture_min"] = float(rupture_min)
    binned["true_rul_min"] = float(rupture_min) - binned["time_min"]
    binned = binned[binned["time_min"] <= float(rupture_min)].copy()

    rms = np.column_stack([binned[f"rms_ch{ch}"].to_numpy(float) for ch in cfg.channels])
    energy = np.empty((len(binned), len(cfg.channels), len(cfg.bands_khz)), dtype=float)
    for j, ch in enumerate(cfg.channels):
        for b in range(len(cfg.bands_khz)):
            energy[:, j, b] = binned[f"e_ch{ch}_b{b}"].to_numpy(float)

    widths = np.array([hi - lo for lo, hi in cfg.bands_khz], dtype=float)
    centres = np.array([(lo + hi) / 2.0 for lo, hi in cfg.bands_khz], dtype=float)
    band_density = energy / widths[None, None, :]

    lf = np.asarray(cfg.lf_band_indices, dtype=int)
    hf = np.asarray(cfg.hf_band_indices, dtype=int)
    selected = np.r_[lf, hf]

    # Mean spectral density in each physical range prevents wider bands from
    # receiving an automatic advantage under spectrally white noise.
    lf_density = energy[:, :, lf].sum(axis=2) / widths[lf].sum()
    hf_density = energy[:, :, hf].sum(axis=2) / widths[hf].sum()
    total_selected = energy[:, :, selected].sum(axis=2)
    spectral_centroid = (
        energy[:, :, selected] * centres[selected][None, None, :]
    ).sum(axis=2) / (total_selected + EPS)

    log_rms = np.log1p(np.maximum(rms, 0.0))
    log_power = np.log1p(np.maximum(total_selected, 0.0))
    hf_logratio = np.log((hf_density + EPS) / (lf_density + EPS))

    # Activity threshold is estimated only from the initial, causally available interval.
    baseline = binned["time_min"].to_numpy() <= cfg.initial_baseline_minutes
    if baseline.sum() < 5:
        baseline[: max(5, len(binned) // 20)] = True
    activity_threshold = np.nanmedian(log_rms[baseline], axis=0) + (
        cfg.activity_mad_multiplier * robust_mad(log_rms[baseline], axis=0)
    )
    active = log_rms > activity_threshold[None, :]

    derived: Dict[str, np.ndarray] = {}
    derived.update(sensor_invariant_statistics(log_rms, "log_rms"))
    derived.update(sensor_invariant_statistics(log_power, "log_power"))
    derived.update(sensor_invariant_statistics(hf_logratio, "hf_logratio"))
    derived.update(sensor_invariant_statistics(spectral_centroid, "spec_centroid_khz"))

    shares = total_selected / (total_selected.sum(axis=1, keepdims=True) + EPS)
    derived["sensor_share_entropy"] = normalized_entropy(shares, axis=1)
    derived["sensor_share_dominance"] = np.nanmax(shares, axis=1)
    derived["active_sensor_count"] = active.sum(axis=1).astype(float)
    derived["active_sensor_fraction"] = active.mean(axis=1)

    # Energy-weighted spectral ratio suppresses unstable ratios from inactive sensors.
    pooled_lf = energy[:, :, lf].sum(axis=(1, 2)) / widths[lf].sum()
    pooled_hf = energy[:, :, hf].sum(axis=(1, 2)) / widths[hf].sum()
    derived["pooled_hf_logratio"] = np.log((pooled_hf + EPS) / (pooled_lf + EPS))

    for name, values in derived.items():
        binned[name] = values

    binned = attach_pressure_columns(binned, d.pipe, cfg)
    return binned.reset_index(drop=True)


TIME_TABLES = {
    p: build_pipe_time_table(PIPEDATA[p], RUPTURE_MIN[p], CFG) for p in CFG.pipes
}
TIME_ALL = pd.concat(TIME_TABLES.values(), ignore_index=True)

try:
    TIME_ALL.to_parquet(CFG.output_dir / "03_prism_time_bins.parquet", index=False)
except (ImportError, ModuleNotFoundError):
    warnings.warn(
        "Parquet support is unavailable (install pyarrow for Parquet). "
        "Writing the Stage-1 time table as compressed CSV instead."
    )
    TIME_ALL.to_csv(CFG.output_dir / "03_prism_time_bins.csv.gz", index=False, compression="gzip")
print(
    TIME_ALL.groupby("pipe")
    .agg(rows=("time_min", "size"), end_min=("time_min", "max"), p_max=("pressure_pd", "max"))
    .round(3)
)


# %% CELL 6 — Corrected descriptive figures: pressure, AE, and sensor redistribution

def plot_pressure_ae_overview(time_tables: Mapping[str, pd.DataFrame], cfg: Config = CFG) -> None:
    fig, axes = plt.subplots(len(cfg.pipes), 1, figsize=(13, 11), sharex=False)
    endpoint_label = (
        f"AE proxy endpoint: {cfg.ae_proxy_endpoint_method}"
        if cfg.allow_ae_proxy_endpoint
        else "Externally verified rupture"
    )
    for ax, pipe in zip(axes, cfg.pipes):
        d = time_tables[pipe]
        ax.plot(d["time_min"], d["log_rms_med"], color="#CC3311", lw=1.0, label="Median log RMS")
        ax.plot(
            d["time_min"], d["log_power_med"], color="#0077BB", lw=1.0,
            label="Median log 60–400 kHz energy"
        )
        ax.axvline(
            RUPTURE_MIN[pipe], color="black", ls="--", lw=1.5,
            label=endpoint_label
        )
        ax.set_ylabel(f"Pipe {pipe}\nAE feature")

        ap = ax.twinx()
        ap.step(
            d["time_min"], d["pressure_pd"], where="post", color="#888888",
            alpha=0.45, lw=1.0, label="Pressure / Pd"
        )
        ap.set_ylabel("P/Pd", color="#666666")

        if pipe == cfg.pipes[0]:
            lines1, labels1 = ax.get_legend_handles_labels()
            lines2, labels2 = ap.get_legend_handles_labels()
            ax.legend(lines1 + lines2, labels1 + labels2, ncol=4, loc="upper left")

    axes[-1].set_xlabel("Time from feature start (min)")
    fig.suptitle("Pressure-aligned acoustic-emission histories", y=1.005)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "04_pressure_ae_overview.png")
    plt.show()


def plot_sensor_fingerprint(data: Mapping[str, PipeFeatures], cfg: Config = CFG) -> None:
    # Use an early-to-middle interval for descriptive comparison only.
    matrix = np.zeros((len(cfg.pipes), len(cfg.channels)), dtype=float)
    for i, pipe in enumerate(cfg.pipes):
        d = data[pipe]
        t = d.t_min
        rupture = RUPTURE_MIN[pipe]
        mask = (t >= 0.30 * rupture) & (t <= 0.70 * rupture)
        selected = np.r_[cfg.lf_band_indices, cfg.hf_band_indices]
        v = np.nanmedian(d.band_energy[mask][:, :, selected].sum(axis=2), axis=0)
        matrix[i] = v / (v.sum() + EPS)

    fig, ax = plt.subplots(figsize=(7.2, 4.2))
    im = ax.imshow(matrix, vmin=0, vmax=max(0.5, float(matrix.max())), cmap="viridis", aspect="auto")
    ax.set_xticks(range(len(cfg.channels)), [f"Channel {c}" for c in cfg.channels])
    ax.set_yticks(range(len(cfg.pipes)), [f"Pipe {p}" for p in cfg.pipes])
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            ax.text(j, i, f"{matrix[i, j]:.2f}", ha="center", va="center", color="white" if matrix[i,j] > .25 else "black")
    ax.set_title("Sensor-energy redistribution across unseen pipes")
    fig.colorbar(im, ax=ax, label="Energy share")
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "05_sensor_fingerprint_heatmap.png")
    plt.show()


plot_pressure_ae_overview(TIME_TABLES, CFG)
plot_sensor_fingerprint(PIPEDATA, CFG)


# %% CELL 7 — Train-only pressure-response models and pressure-residualized AE

PRESSURE_FEATURES_TIME = [
    "pressure_pd",
    "delta_pressure_pd",
    "abs_delta_pressure_pd",
    "time_in_hold_min",
    "dose_1",
    "dose_2",
    "dose_4",
]

# Keep this set small and interpretable; Stage 1 tests incremental information.
AE_FEATURES_TIME = [
    "log_rms_med",
    "log_rms_iqr",
    "log_power_med",
    "log_power_iqr",
    "hf_logratio_med",
    "hf_logratio_iqr",
    "spec_centroid_khz_med",
    "spec_centroid_khz_iqr",
    "sensor_share_entropy",
    "sensor_share_dominance",
    "active_sensor_fraction",
    "pooled_hf_logratio",
]


def balanced_pipe_weights(df: pd.DataFrame) -> np.ndarray:
    counts = df["pipe"].value_counts()
    w = df["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    return w / np.mean(w)


def fit_pressure_response_models(
    train_time: pd.DataFrame,
    ae_features: Sequence[str] = AE_FEATURES_TIME,
    pressure_features: Sequence[str] = PRESSURE_FEATURES_TIME,
    cfg: Config = CFG,
) -> Dict[str, HistGradientBoostingRegressor]:
    X = train_time[list(pressure_features)].replace([np.inf, -np.inf], np.nan)
    X = X.fillna(X.median(numeric_only=True))
    w = balanced_pipe_weights(train_time)

    models = {}
    for feature in ae_features:
        y = train_time[feature].to_numpy(float)
        finite = np.isfinite(y)
        model = HistGradientBoostingRegressor(
            loss="squared_error",
            learning_rate=0.04,
            max_iter=250,
            max_leaf_nodes=8,
            min_samples_leaf=max(10, int(0.01 * finite.sum())),
            l2_regularization=2.0,
            random_state=cfg.random_state,
        )
        model.fit(X.loc[finite], y[finite], sample_weight=w[finite])
        models[feature] = model
    return models


def apply_pressure_residualization(
    time_df: pd.DataFrame,
    models: Mapping[str, HistGradientBoostingRegressor],
    pressure_features: Sequence[str] = PRESSURE_FEATURES_TIME,
) -> pd.DataFrame:
    out = time_df.copy()
    X = out[list(pressure_features)].replace([np.inf, -np.inf], np.nan)
    X = X.fillna(X.median(numeric_only=True))
    for feature, model in models.items():
        expected = model.predict(X)
        out[f"expected_{feature}"] = expected
        out[f"resid_{feature}"] = out[feature].to_numpy(float) - expected
    return out


def make_outer_fold_time_tables(
    test_pipe: str, time_tables: Mapping[str, pd.DataFrame], cfg: Config = CFG
) -> Tuple[Dict[str, pd.DataFrame], Dict[str, HistGradientBoostingRegressor]]:
    train_pipes = [p for p in cfg.pipes if p != test_pipe]
    train_time = pd.concat([time_tables[p] for p in train_pipes], ignore_index=True)
    load_models = fit_pressure_response_models(train_time, cfg=cfg)
    residualized = {p: apply_pressure_residualization(time_tables[p], load_models) for p in cfg.pipes}
    return residualized, load_models


# Visual diagnostic for one fold: test pipe B must not influence its pressure-response model.
EXAMPLE_RESIDUALIZED, _ = make_outer_fold_time_tables("B", TIME_TABLES, CFG)
fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
for pipe in CFG.pipes:
    d = EXAMPLE_RESIDUALIZED[pipe]
    style = "-" if pipe == "B" else "--"
    alpha = 1.0 if pipe == "B" else 0.55
    axes[0].plot(d["time_min"], d["log_power_med"], style, alpha=alpha, color=PIPE_COLORS[pipe], label=pipe)
    axes[1].plot(d["time_min"], d["resid_log_power_med"], style, alpha=alpha, color=PIPE_COLORS[pipe], label=pipe)
axes[0].set_ylabel("Observed log power")
axes[1].set_ylabel("Pressure-residualized log power")
axes[1].set_xlabel("Time (min)")
axes[0].legend(ncol=4, title="Fold: B held out")
fig.tight_layout()
fig.savefig(CFG.output_dir / "06_example_pressure_residualization_B_held_out.png")
plt.show()


# %% CELL 8 — Strictly causal checkpoint summaries

def linear_slope_per_minute(t: np.ndarray, y: np.ndarray) -> float:
    finite = np.isfinite(t) & np.isfinite(y)
    if finite.sum() < 3 or np.ptp(t[finite]) <= 0:
        return 0.0
    return float(stats.linregress(t[finite], y[finite]).slope)


def last_finite(series: pd.Series) -> float:
    x = series.to_numpy(float)
    x = x[np.isfinite(x)]
    return float(x[-1]) if len(x) else np.nan


def summarize_history_at(
    d: pd.DataFrame,
    observation_min: float,
    raw_features: Sequence[str] = AE_FEATURES_TIME,
    cfg: Config = CFG,
) -> Dict[str, float]:
    """Construct features using only rows at or before observation_min."""
    hist = d[d["time_min"] <= observation_min].copy()
    if hist.empty:
        raise ValueError(f"No history exists at t={observation_min:.3f} min")
    now = hist.iloc[-1]

    row: Dict[str, float] = {
        "age_min": float(now["time_min"]),
        "pressure_pd": float(now["pressure_pd"]),
        "delta_pressure_pd": float(now["delta_pressure_pd"]),
        "abs_delta_pressure_pd": float(now["abs_delta_pressure_pd"]),
        "time_in_hold_min": float(now["time_in_hold_min"]),
        "dose_1": float(now["dose_1"]),
        "dose_2": float(now["dose_2"]),
        "dose_4": float(now["dose_4"]),
    }

    residual_features = [f"resid_{f}" for f in raw_features]
    for feature in list(raw_features) + residual_features:
        if feature not in hist.columns:
            continue
        row[f"{feature}__current"] = float(now[feature])

        # Expanding causal summaries.
        values = hist[feature].to_numpy(float)
        times = hist["time_min"].to_numpy(float)
        row[f"{feature}__exp_med"] = float(np.nanmedian(values))
        row[f"{feature}__exp_iqr"] = float(np.nanpercentile(values, 75) - np.nanpercentile(values, 25))
        positive = np.maximum(values, 0.0)
        row[f"{feature}__positive_dose"] = float(np.trapezoid(positive, times)) if len(times) > 1 else 0.0

        for window in cfg.rolling_windows_minutes:
            recent = hist[hist["time_min"] >= observation_min - window]
            rv = recent[feature].to_numpy(float)
            rt = recent["time_min"].to_numpy(float)
            tag = str(window).replace(".", "p")
            row[f"{feature}__med_{tag}m"] = float(np.nanmedian(rv))
            row[f"{feature}__iqr_{tag}m"] = float(
                np.nanpercentile(rv, 75) - np.nanpercentile(rv, 25)
            )
            row[f"{feature}__slope_{tag}m"] = linear_slope_per_minute(rt, rv)
    return row


def make_checkpoint_table_for_fold(
    test_pipe: str,
    residualized_tables: Mapping[str, pd.DataFrame],
    rupture_min: Mapping[str, float],
    cfg: Config = CFG,
) -> pd.DataFrame:
    rows = []
    for pipe in cfg.pipes:
        life = float(rupture_min[pipe])
        d = residualized_tables[pipe]
        for fraction in cfg.observation_fractions:
            observation = float(fraction * life)
            row = summarize_history_at(d, observation, cfg=cfg)
            row.update(
                {
                    "outer_test_pipe": test_pipe,
                    "pipe": pipe,
                    "observation_fraction": float(fraction),
                    "life_min": life,
                    "true_rul_min": max(life - row["age_min"], 0.0),
                }
            )
            rows.append(row)
    return pd.DataFrame(rows)


# Feature families are selected by names, making the audit explicit and reproducible.
def checkpoint_feature_families(cp: pd.DataFrame) -> Dict[str, List[str]]:
    pressure = [
        "age_min", "pressure_pd", "delta_pressure_pd", "abs_delta_pressure_pd",
        "time_in_hold_min", "dose_1", "dose_2", "dose_4",
    ]
    raw_ae = [c for c in cp.columns if any(c.startswith(f"{f}__") for f in AE_FEATURES_TIME)]
    resid_ae = [c for c in cp.columns if c.startswith("resid_")]
    return {
        "Pressure only": pressure,
        "Pressure + raw AE": pressure + raw_ae,
        "PRISM-AE residual": pressure + resid_ae,
    }


# %% CELL 9 — Group-aware ridge selection, pilot intervals, and LOPO evaluation

def make_ridge(alpha: float) -> Pipeline:
    return Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
            ("scale", StandardScaler()),
            ("ridge", Ridge(alpha=float(alpha))),
        ]
    )


def fit_with_balanced_weights(model: Pipeline, X: pd.DataFrame, y: np.ndarray, groups: np.ndarray) -> Pipeline:
    group_counts = pd.Series(groups).value_counts()
    w = np.array([1.0 / group_counts[g] for g in groups], dtype=float)
    w /= w.mean()
    model.fit(X, y, ridge__sample_weight=w)
    return model


def select_alpha_group_cv(
    train: pd.DataFrame,
    features: Sequence[str],
    cfg: Config = CFG,
) -> Tuple[float, pd.DataFrame]:
    groups = train["pipe"].to_numpy()
    unique_groups = np.unique(groups)
    records = []

    for alpha in cfg.ridge_alphas:
        fold_scores = []
        for held in unique_groups:
            tr = train[train["pipe"] != held]
            va = train[train["pipe"] == held]
            model = make_ridge(alpha)
            fit_with_balanced_weights(
                model,
                tr[list(features)],
                tr["life_min"].to_numpy(float),
                tr["pipe"].to_numpy(),
            )
            pred_life = model.predict(va[list(features)])
            # RUL error equals total-life error after subtracting the known current age.
            score = mean_absolute_error(va["life_min"], pred_life)
            fold_scores.append(float(score))
        records.append({"alpha": alpha, "group_balanced_mae": float(np.mean(fold_scores))})

    cv = pd.DataFrame(records).sort_values(["group_balanced_mae", "alpha"])
    return float(cv.iloc[0]["alpha"]), cv


def group_jackknife_halfwidth(
    train: pd.DataFrame,
    features: Sequence[str],
    alpha: float,
    interval_level: float,
) -> Tuple[float, pd.DataFrame]:
    """
    Leave one training pipe out, collect pipe-level high-quantile errors, and use
    the maximum across training pipes. This is conservative but not a final UQ model.
    """
    rows = []
    for held in sorted(train["pipe"].unique()):
        tr = train[train["pipe"] != held]
        va = train[train["pipe"] == held]
        model = make_ridge(alpha)
        fit_with_balanced_weights(
            model, tr[list(features)], tr["life_min"].to_numpy(float), tr["pipe"].to_numpy()
        )
        pred = model.predict(va[list(features)])
        errors = np.abs(pred - va["life_min"].to_numpy(float))
        rows.append(
            {
                "held_pipe": held,
                "mae": float(np.mean(errors)),
                "q_error": float(np.quantile(errors, interval_level)),
                "max_error": float(np.max(errors)),
            }
        )
    detail = pd.DataFrame(rows)
    halfwidth = float(detail["q_error"].max())
    return halfwidth, detail


def predict_age_only(train: pd.DataFrame, test: pd.DataFrame, interval_level: float) -> pd.DataFrame:
    lives = train.groupby("pipe")["life_min"].first().to_numpy(float)
    pred_life = float(np.mean(lives))
    # With only three reference lives, use their full range as an honest pilot interval.
    lo_life = float(np.min(lives))
    hi_life = float(np.max(lives))

    out = test[["pipe", "observation_fraction", "age_min", "life_min", "true_rul_min"]].copy()
    out["method"] = "Age only"
    out["pred_life_min"] = np.maximum(pred_life, out["age_min"])
    out["pred_rul_min"] = np.maximum(out["pred_life_min"] - out["age_min"], 0.0)
    out["lower_rul_min"] = np.maximum(lo_life - out["age_min"], 0.0)
    out["upper_rul_min"] = np.maximum(hi_life - out["age_min"], 0.0)
    out["selected_alpha"] = np.nan
    out["pilot_halfwidth_min"] = (hi_life - lo_life) / 2.0
    return out


def fit_predict_ridge_family(
    train: pd.DataFrame,
    test: pd.DataFrame,
    features: Sequence[str],
    method_name: str,
    cfg: Config = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    alpha, cv = select_alpha_group_cv(train, features, cfg)
    halfwidth, jackknife = group_jackknife_halfwidth(
        train, features, alpha, cfg.interval_level
    )
    model = make_ridge(alpha)
    fit_with_balanced_weights(
        model, train[list(features)], train["life_min"].to_numpy(float), train["pipe"].to_numpy()
    )
    pred_life_raw = model.predict(test[list(features)])
    pred_life = np.maximum(pred_life_raw, test["age_min"].to_numpy(float))

    out = test[["pipe", "observation_fraction", "age_min", "life_min", "true_rul_min"]].copy()
    out["method"] = method_name
    out["pred_life_min"] = pred_life
    out["pred_rul_min"] = np.maximum(pred_life - out["age_min"], 0.0)
    out["lower_rul_min"] = np.maximum(pred_life - halfwidth - out["age_min"], 0.0)
    out["upper_rul_min"] = np.maximum(pred_life + halfwidth - out["age_min"], 0.0)
    out["selected_alpha"] = alpha
    out["pilot_halfwidth_min"] = halfwidth
    cv = cv.assign(method=method_name)
    jackknife = jackknife.assign(method=method_name, selected_alpha=alpha)
    return out, cv, jackknife


def run_strict_lopo(cfg: Config = CFG) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    predictions = []
    cv_records = []
    interval_records = []

    for test_pipe in cfg.pipes:
        print(f"\n=== Outer fold: hold out pipe {test_pipe} ===")
        residualized, _ = make_outer_fold_time_tables(test_pipe, TIME_TABLES, cfg)
        cp = make_checkpoint_table_for_fold(test_pipe, residualized, RUPTURE_MIN, cfg)
        families = checkpoint_feature_families(cp)

        train = cp[cp["pipe"] != test_pipe].reset_index(drop=True)
        test = cp[cp["pipe"] == test_pipe].reset_index(drop=True)

        age = predict_age_only(train, test, cfg.interval_level)
        age["outer_test_pipe"] = test_pipe
        predictions.append(age)

        for method_name, features in families.items():
            pred, cv, jackknife = fit_predict_ridge_family(
                train, test, features, method_name, cfg
            )
            pred["outer_test_pipe"] = test_pipe
            cv["outer_test_pipe"] = test_pipe
            jackknife["outer_test_pipe"] = test_pipe
            predictions.append(pred)
            cv_records.append(cv)
            interval_records.append(jackknife)
            print(
                f"{method_name:24s} alpha={pred['selected_alpha'].iloc[0]:.4g} "
                f"pilot ±{pred['pilot_halfwidth_min'].iloc[0]:.2f} min"
            )

    pred_all = pd.concat(predictions, ignore_index=True)
    cv_all = pd.concat(cv_records, ignore_index=True)
    interval_all = pd.concat(interval_records, ignore_index=True)
    return pred_all, cv_all, interval_all


PREDICTIONS, RIDGE_CV, PILOT_INTERVALS = run_strict_lopo(CFG)
PREDICTIONS.to_csv(CFG.output_dir / "07_lopo_predictions.csv", index=False)
RIDGE_CV.to_csv(CFG.output_dir / "07_ridge_inner_group_cv.csv", index=False)
PILOT_INTERVALS.to_csv(CFG.output_dir / "07_group_jackknife_intervals.csv", index=False)


# %% CELL 10 — RESS-oriented metrics: accuracy, calibration, early-warning value

def phm_weighted_absolute_error(df: pd.DataFrame) -> float:
    # Reproduces the paper's late-life weighting using observation fraction.
    weights = df["observation_fraction"].to_numpy(float)
    errors = np.abs(df["pred_rul_min"] - df["true_rul_min"]).to_numpy(float)
    return float(np.sum(weights * errors) / (np.sum(weights) + EPS))


def evaluate_prediction_group(df: pd.DataFrame, alarm_horizon_min: float = 10.0) -> Dict[str, float]:
    y = df["true_rul_min"].to_numpy(float)
    p = df["pred_rul_min"].to_numpy(float)
    covered = (y >= df["lower_rul_min"].to_numpy(float)) & (
        y <= df["upper_rul_min"].to_numpy(float)
    )

    # Pilot event probability from a Gaussian approximation whose scale is tied
    # to the group-jackknife half-width. This is diagnostic, not final Stage-2 UQ.
    half = float(np.nanmedian(df["pilot_halfwidth_min"]))
    z90 = stats.norm.ppf(0.5 + CFG.interval_level / 2.0)
    sigma = max(half / max(z90, EPS), 0.5)
    fail_prob = stats.norm.cdf((alarm_horizon_min - p) / sigma)
    event = (y <= alarm_horizon_min).astype(float)
    brier = float(np.mean((fail_prob - event) ** 2))

    return {
        "MAE_min": float(mean_absolute_error(y, p)),
        "RMSE_min": float(np.sqrt(mean_squared_error(y, p))),
        "PHM_weighted_MAE": phm_weighted_absolute_error(df),
        "coverage": float(np.mean(covered)),
        "mean_interval_width_min": float(np.mean(df["upper_rul_min"] - df["lower_rul_min"])),
        f"Brier_fail_within_{alarm_horizon_min:g}min": brier,
    }


metric_rows = []
for (method, pipe), g in PREDICTIONS.groupby(["method", "pipe"], sort=False):
    metric_rows.append({"method": method, "pipe": pipe, **evaluate_prediction_group(g)})

METRICS_BY_PIPE = pd.DataFrame(metric_rows)
METRICS_SUMMARY = (
    METRICS_BY_PIPE.groupby("method", sort=False)
    .agg(
        MAE_mean=("MAE_min", "mean"),
        MAE_std=("MAE_min", "std"),
        RMSE_mean=("RMSE_min", "mean"),
        PHM_mean=("PHM_weighted_MAE", "mean"),
        coverage_mean=("coverage", "mean"),
        width_mean=("mean_interval_width_min", "mean"),
        Brier_10min_mean=("Brier_fail_within_10min", "mean"),
    )
    .reset_index()
)

METRICS_BY_PIPE.to_csv(CFG.output_dir / "08_metrics_by_pipe.csv", index=False)
METRICS_SUMMARY.to_csv(CFG.output_dir / "08_metrics_summary.csv", index=False)

print("\nGroup-balanced LOPO summary (each pipe receives equal weight):")
print(METRICS_SUMMARY.round(4).to_string(index=False))


def incremental_value_table(metrics_by_pipe: pd.DataFrame) -> pd.DataFrame:
    wide = metrics_by_pipe.pivot(index="pipe", columns="method", values="MAE_min")
    for method in ("Pressure + raw AE", "PRISM-AE residual"):
        wide[f"MAE_gain_{method}"] = wide["Pressure only"] - wide[method]
        wide[f"relative_gain_{method}"] = (
            wide["Pressure only"] - wide[method]
        ) / (wide["Pressure only"] + EPS)
    return wide.reset_index()


INCREMENTAL_VALUE = incremental_value_table(METRICS_BY_PIPE)
INCREMENTAL_VALUE.to_csv(CFG.output_dir / "08_incremental_value_over_pressure.csv", index=False)
print("\nPositive gain means AE improves over the pressure-only model:")
print(INCREMENTAL_VALUE.round(4).to_string(index=False))


# %% CELL 11 — Publication-quality Stage-1 figures

METHOD_ORDER = ["Age only", "Pressure only", "Pressure + raw AE", "PRISM-AE residual"]


def plot_lopo_rul_trajectories(pred: pd.DataFrame, cfg: Config = CFG) -> None:
    fig, axes = plt.subplots(2, 2, figsize=(13, 9), sharex=False, sharey=False)
    for ax, pipe in zip(axes.flat, cfg.pipes):
        gp = pred[pred["pipe"] == pipe]
        truth = gp[gp["method"] == METHOD_ORDER[0]].sort_values("age_min")
        ax.plot(truth["age_min"], truth["true_rul_min"], color="black", lw=2.5, label="True RUL")

        for method in METHOD_ORDER:
            g = gp[gp["method"] == method].sort_values("age_min")
            ax.plot(
                g["age_min"], g["pred_rul_min"], marker="o", ms=3.2, lw=1.4,
                color=METHOD_COLORS[method], label=method
            )
            if method == "PRISM-AE residual":
                ax.fill_between(
                    g["age_min"].to_numpy(),
                    g["lower_rul_min"].to_numpy(),
                    g["upper_rul_min"].to_numpy(),
                    color=METHOD_COLORS[method], alpha=0.14, label=f"{int(100*cfg.interval_level)}% pilot interval"
                )
        ax.set_title(f"Held-out pipe {pipe}")
        ax.set_xlabel("Observed age (min)")
        ax.set_ylabel("Remaining useful life (min)")
        ax.set_ylim(bottom=0)

    handles, labels = axes.flat[0].get_legend_handles_labels()
    by_label = dict(zip(labels, handles))
    fig.legend(by_label.values(), by_label.keys(), loc="upper center", ncol=3, bbox_to_anchor=(0.5, 1.02))
    fig.suptitle("Strict leave-one-pipe-out RUL trajectories", y=1.07)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "09_lopo_rul_trajectories.png")
    plt.show()


def plot_metric_summary(summary: pd.DataFrame, cfg: Config = CFG) -> None:
    s = summary.set_index("method").loc[METHOD_ORDER].reset_index()
    x = np.arange(len(s))
    colors = [METHOD_COLORS[m] for m in s["method"]]

    fig, axes = plt.subplots(1, 3, figsize=(15, 4.6))
    axes[0].bar(x, s["MAE_mean"], yerr=s["MAE_std"], color=colors, edgecolor="black", capsize=4)
    axes[0].set_ylabel("Pipe-balanced MAE (min)")
    axes[0].set_title("Accuracy")

    axes[1].bar(x, s["coverage_mean"], color=colors, edgecolor="black")
    axes[1].axhline(cfg.interval_level, color="black", ls="--", lw=1.3, label="Nominal")
    axes[1].set_ylim(0, 1.05)
    axes[1].set_ylabel("Empirical coverage")
    axes[1].set_title("Pilot interval calibration")
    axes[1].legend()

    axes[2].bar(x, s["Brier_10min_mean"], color=colors, edgecolor="black")
    axes[2].set_ylabel("Brier score")
    axes[2].set_title("Rupture within 10 min")

    for ax in axes:
        ax.set_xticks(x, [m.replace("Pressure + ", "P + ") for m in s["method"]], rotation=25, ha="right")
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "10_stage1_metric_summary.png")
    plt.show()


def plot_incremental_ae_value(metrics_by_pipe: pd.DataFrame, cfg: Config = CFG) -> None:
    wide = metrics_by_pipe.pivot(index="pipe", columns="method", values="MAE_min").loc[list(cfg.pipes)]
    gain_raw = wide["Pressure only"] - wide["Pressure + raw AE"]
    gain_resid = wide["Pressure only"] - wide["PRISM-AE residual"]
    x = np.arange(len(cfg.pipes))

    fig, ax = plt.subplots(figsize=(8.5, 4.8))
    ax.bar(x - 0.2, gain_raw, 0.4, color=METHOD_COLORS["Pressure + raw AE"], edgecolor="black", label="Raw AE")
    ax.bar(x + 0.2, gain_resid, 0.4, color=METHOD_COLORS["PRISM-AE residual"], edgecolor="black", label="Pressure-residualized AE")
    ax.axhline(0, color="black", lw=1.4)
    ax.set_xticks(x, [f"Pipe {p}" for p in cfg.pipes])
    ax.set_ylabel("MAE reduction versus pressure only (min)")
    ax.set_title("Does acoustic emission add reliability information?")
    ax.legend()
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "11_incremental_ae_value.png")
    plt.show()


plot_lopo_rul_trajectories(PREDICTIONS, CFG)
plot_metric_summary(METRICS_SUMMARY, CFG)
plot_incremental_ae_value(METRICS_BY_PIPE, CFG)


# %% CELL 12 — Decision gate and reproducibility manifest

def stage1_decision_gate(metrics: pd.DataFrame, cfg: Config = CFG) -> Dict[str, object]:
    m = metrics.set_index("method")
    pressure_mae = float(m.loc["Pressure only", "MAE_mean"])
    raw_mae = float(m.loc["Pressure + raw AE", "MAE_mean"])
    prism_mae = float(m.loc["PRISM-AE residual", "MAE_mean"])
    relative_gain = (pressure_mae - prism_mae) / (pressure_mae + EPS)
    wins = int(
        (
            METRICS_BY_PIPE.pivot(index="pipe", columns="method", values="MAE_min")["PRISM-AE residual"]
            < METRICS_BY_PIPE.pivot(index="pipe", columns="method", values="MAE_min")["Pressure only"]
        ).sum()
    )

    if relative_gain >= 0.15 and wins >= 3:
        verdict = "GO"
        next_step = "Proceed to Stage 2 Bayesian stochastic damage, calibrated survival, and risk decision model."
    elif relative_gain > 0 and wins >= 2:
        verdict = "CONDITIONAL GO"
        next_step = "Inspect pipe-specific failures, simplify features, and verify pressure/rupture alignment before Stage 2."
    else:
        verdict = "PIVOT"
        next_step = "Do not claim RUL value yet; focus on pressure confounding, damage detectability, or collect additional tests."

    exploratory_inputs = bool(
        cfg.allow_ae_proxy_endpoint or cfg.allow_ae_proxy_pressure_start
    )
    if exploratory_inputs and verdict in {"GO", "CONDITIONAL GO"}:
        verdict = f"EXPLORATORY {verdict}"
        next_step = (
            "The signal is promising, but verify rupture timestamps and pressure alignment "
            "externally before publication-grade Stage 2 validation."
        )

    return {
        "verdict": verdict,
        "pressure_MAE_min": pressure_mae,
        "raw_AE_MAE_min": raw_mae,
        "PRISM_MAE_min": prism_mae,
        "PRISM_relative_gain": relative_gain,
        "pipes_won_out_of_4": wins,
        "next_step": next_step,
        "labels_are_physical": not cfg.allow_ae_proxy_endpoint,
        "pressure_alignment_is_external": not cfg.allow_ae_proxy_pressure_start,
        "exploratory_inputs": exploratory_inputs,
    }


DECISION = stage1_decision_gate(METRICS_SUMMARY, CFG)
print(json.dumps(DECISION, indent=2))

manifest = {
    "project": "PRISM-AE Stage 1",
    "seed": CFG.random_state,
    "pipes": list(CFG.pipes),
    "channels": list(CFG.channels),
    "bands_khz": [list(x) for x in CFG.bands_khz],
    "bin_seconds": CFG.bin_seconds,
    "observation_fractions": list(CFG.observation_fractions),
    "rupture_min": RUPTURE_MIN,
    "endpoint_source": (
        f"exploratory_AE_proxy:{CFG.ae_proxy_endpoint_method}"
        if CFG.allow_ae_proxy_endpoint
        else "external_physical_override"
    ),
    "pressure_start_min": CFG.pressure_start_min_override,
    "pressure_alignment_source": (
        "exploratory_AE_proxy_or_actual_log"
        if CFG.allow_ae_proxy_pressure_start
        else "external_override_or_actual_log"
    ),
    "actual_pressure_logs_found": {
        p: load_actual_pressure_log(p, CFG) is not None for p in CFG.pipes
    },
    "decision_gate": DECISION,
}

with open(CFG.output_dir / "12_reproducibility_manifest.json", "w", encoding="utf-8") as fh:
    json.dump(manifest, fh, indent=2)

print("\nStage 1 complete. Key files:")
for path in sorted(CFG.output_dir.glob("*")):
    print(" -", path.name)


# %% CELL 13 — What to send back before Stage 2

print(
    """
SEND THESE RESULTS FOR THE NEXT IMPLEMENTATION STAGE
----------------------------------------------------
1. The printed input schema report from CELL 2.
2. Verified rupture times and whether actual pressure CSVs were available.
3. 08_metrics_summary.csv
4. 08_metrics_by_pipe.csv
5. 08_incremental_value_over_pressure.csv
6. 04_pressure_ae_overview.png
7. 06_example_pressure_residualization_B_held_out.png
8. 09_lopo_rul_trajectories.png
9. 11_incremental_ae_value.png
10. The JSON decision printed by CELL 12.

If the decision is GO or CONDITIONAL GO, Stage 2 will implement:
- a hierarchical Bayesian stochastic damage state;
- separate aleatory and epistemic uncertainty;
- dynamic survival and failure-within-horizon probabilities;
- cost-sensitive alarm/pressure-hold decisions;
- sensor dropout/gain/noise simulations;
- posterior calibration and RESS figures.
"""
)
```

```python
"""
PRISM-Hazard — Stage 2
======================

Pressure-conditioned, sensor-invariant, calibrated near-term failure-risk
forecasting for pressurized pipelines.

This is the scientifically motivated pivot after Stage 1 showed that point-RUL
regression was dominated by elapsed time/pressure, while acoustic emission (AE)
showed promising failure-within-10-minute Brier performance.

Run as a VS Code/Jupyter ``# %%`` notebook. The implementation is deliberately
strict about held-out-pipe separation and about the exploratory provenance of
AE-derived endpoint and pressure-alignment labels.
"""


# %% CELL 1 — Configuration and reproducible environment

from __future__ import annotations

import json
import math
import os
import warnings
from dataclasses import asdict, dataclass, field
from pathlib import Path
from typing import Dict, Iterable, List, Mapping, Optional, Sequence, Tuple

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.special import expit, logit
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    log_loss,
    roc_auc_score,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler


SEED = 20260829
EPS = 1e-12
np.random.seed(SEED)

mpl.rcParams.update(
    {
        "figure.dpi": 120,
        "savefig.dpi": 300,
        "font.family": "DejaVu Sans",
        "font.size": 10.5,
        "axes.labelsize": 11,
        "axes.titlesize": 12,
        "axes.grid": True,
        "grid.alpha": 0.22,
        "legend.frameon": False,
        "savefig.bbox": "tight",
        "savefig.facecolor": "white",
    }
)


@dataclass
class HazardConfig:
    # Input produced by PRISM-AE Stage 1.
    stage1_dir: Path = Path(
        r"F:\Pipeline Prressure Work\Code\results_prism_stage1"
    )
    output_dir: Path = Path(
        r"F:\Pipeline Prressure Work\Code\results_prism_stage2_hazard"
    )

    pipes: Tuple[str, ...] = ("B", "C", "D", "E")
    channels: Tuple[int, ...] = (5, 6, 7, 8)
    bands_khz: Tuple[Tuple[float, float], ...] = (
        (20.0, 60.0),
        (60.0, 100.0),
        (100.0, 140.0),
        (140.0, 190.0),
        (190.0, 250.0),
        (250.0, 400.0),
    )
    lf_band_indices: Tuple[int, ...] = (1, 2, 3)
    hf_band_indices: Tuple[int, ...] = (4, 5)

    # Dynamic risk horizons. The 10-minute horizon is the primary endpoint;
    # the others test whether conclusions depend on one arbitrary horizon.
    horizons_min: Tuple[float, ...] = (2.0, 5.0, 10.0, 15.0)
    primary_horizon_min: float = 10.0

    # Half-minute sampling limits serial pseudo-replication while preserving
    # sufficient positive samples for the shortest horizon.
    sample_minutes: float = 0.5
    analysis_start_min: float = 10.0
    healthy_margin_min: float = 20.0
    baseline_minutes: float = 10.0
    rolling_windows_min: Tuple[float, ...] = (1.0, 5.0, 10.0)

    # Simple models are intentional: there are only four independent pipes.
    logistic_c_grid: Tuple[float, ...] = tuple(np.logspace(-3, 3, 7))
    pressure_model_max_iter: int = 180

    # Sequential-alarm design. Policy selection uses cross-fitted predictions
    # from the training pipes only.
    alarm_threshold_grid: Tuple[float, ...] = tuple(np.linspace(0.05, 0.95, 37))
    alarm_persistence_grid: Tuple[int, ...] = (1, 2, 3, 4, 5)
    target_false_alarm_events_per_hour: float = 0.5

    # Endpoint sensitivity cases derived by Stage 1. These are diagnostic
    # proxies, not physical rupture times.
    endpoint_scenarios: Mapping[str, str] = field(
        default_factory=lambda: {
            "shutdown_onset": "existing_all_rms_gt3",
            "last_all_active": "last_4plus_active",
            "last_two_active": "last_2plus_active",
        }
    )

    # Feature-domain stress experiments; not substitutes for waveform-level tests.
    stress_repeats: int = 8
    gain_log_sigma: float = 0.45
    noise_floor_fraction: float = 0.20

    # These flags are written to every decision artifact.
    labels_are_physical: bool = False
    pressure_alignment_is_external: bool = False
    random_state: int = SEED


CFG = HazardConfig()
CFG.output_dir.mkdir(parents=True, exist_ok=True)

PIPE_COLORS = {"B": "#4477AA", "C": "#EE6677", "D": "#228833", "E": "#CCBB44"}
METHOD_COLORS = {
    "Age only": "#777777",
    "Pressure only": "#AA4499",
    "Pressure + raw AE": "#44AA99",
    "PRISM-Hazard": "#332288",
}
METHOD_ORDER = ["Age only", "Pressure only", "Pressure + raw AE", "PRISM-Hazard"]

print("Stage-1 directory:", CFG.stage1_dir)
print("Stage-2 directory:", CFG.output_dir)
print("Risk horizons (min):", CFG.horizons_min)
print("Primary scientific endpoint: failure within", CFG.primary_horizon_min, "min")


# %% CELL 2 — Load Stage-1 outputs and verify scientific provenance

def read_stage1_time_table(cfg: HazardConfig = CFG) -> pd.DataFrame:
    parquet = cfg.stage1_dir / "03_prism_time_bins.parquet"
    csv_gz = cfg.stage1_dir / "03_prism_time_bins.csv.gz"
    if parquet.exists():
        df = pd.read_parquet(parquet)
        source = parquet
    elif csv_gz.exists():
        df = pd.read_csv(csv_gz)
        source = csv_gz
    else:
        raise FileNotFoundError(
            "Could not find 03_prism_time_bins.parquet or .csv.gz in "
            f"{cfg.stage1_dir}. Run PRISM-AE Stage 1 first."
        )

    required = {
        "pipe", "time_min", "true_rul_min", "rupture_min", "pressure_pd",
        "time_in_hold_min", "dose_1", "dose_2", "dose_4",
        "log_rms_med", "log_power_med", "pooled_hf_logratio",
    }
    missing = sorted(required - set(df.columns))
    if missing:
        raise ValueError(f"Stage-1 table is missing columns: {missing}")

    df = df[df["pipe"].isin(cfg.pipes)].copy()
    df["pipe"] = df["pipe"].astype(str)
    df = df.sort_values(["pipe", "time_min"]).reset_index(drop=True)
    if df.empty:
        raise ValueError("Stage-1 time table contains no requested pipes")
    print(f"Loaded {len(df):,} rows from {source}")
    return df


def read_endpoint_candidates(cfg: HazardConfig = CFG) -> pd.DataFrame:
    path = cfg.stage1_dir / "02_endpoint_candidates_NOT_ground_truth.csv"
    if not path.exists():
        raise FileNotFoundError(f"Missing endpoint-candidate file: {path}")
    table = pd.read_csv(path)
    if "pipe" not in table:
        raise ValueError(f"{path} must contain a pipe column")
    table["pipe"] = table["pipe"].astype(str)
    return table


def read_stage1_manifest(cfg: HazardConfig = CFG) -> Dict[str, object]:
    path = cfg.stage1_dir / "12_reproducibility_manifest.json"
    if not path.exists():
        warnings.warn("Stage-1 manifest not found; provenance will be marked unresolved")
        return {}
    with open(path, "r", encoding="utf-8") as fh:
        return json.load(fh)


TIME_RAW = read_stage1_time_table(CFG)
ENDPOINT_CANDIDATES = read_endpoint_candidates(CFG)
STAGE1_MANIFEST = read_stage1_manifest(CFG)

CFG.labels_are_physical = bool(
    STAGE1_MANIFEST.get("decision_gate", {}).get("labels_are_physical", False)
)
CFG.pressure_alignment_is_external = bool(
    STAGE1_MANIFEST.get("decision_gate", {}).get("pressure_alignment_is_external", False)
)

schema_report = (
    TIME_RAW.groupby("pipe")
    .agg(
        rows=("time_min", "size"),
        start_min=("time_min", "min"),
        available_end_min=("time_min", "max"),
        stage1_endpoint_min=("rupture_min", "max"),
        max_pressure_pd=("pressure_pd", "max"),
    )
    .reset_index()
)
schema_report.to_csv(CFG.output_dir / "01_stage2_input_audit.csv", index=False)
print(schema_report.round(3).to_string(index=False))
print("\nLabel provenance:", "PHYSICAL" if CFG.labels_are_physical else "EXPLORATORY PROXY")
print(
    "Pressure alignment:",
    "EXTERNAL" if CFG.pressure_alignment_is_external else "EXPLORATORY/AE-ALIGNED",
)


# %% CELL 3 — Endpoint scenarios and common-support risk labels

def endpoint_map_for_scenario(
    scenario: str,
    candidates: pd.DataFrame = ENDPOINT_CANDIDATES,
    cfg: HazardConfig = CFG,
) -> Dict[str, float]:
    if scenario not in cfg.endpoint_scenarios:
        raise KeyError(f"Unknown endpoint scenario: {scenario}")
    column = cfg.endpoint_scenarios[scenario]
    if column not in candidates.columns:
        raise KeyError(f"Endpoint candidate column not found: {column}")
    indexed = candidates.set_index("pipe")[column]
    out = {p: float(indexed.loc[p]) for p in cfg.pipes}
    if not all(np.isfinite(list(out.values()))):
        raise ValueError(f"Non-finite endpoint in scenario {scenario}: {out}")
    return out


def attach_scenario_labels(
    table: pd.DataFrame,
    scenario: str,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    endpoints = endpoint_map_for_scenario(scenario, cfg=cfg)
    pieces, audit = [], []
    for pipe in cfg.pipes:
        d = table[table["pipe"] == pipe].copy()
        endpoint = endpoints[pipe]
        available_end = float(d["time_min"].max())
        # Rows after a proposed endpoint are invalid; later proposed endpoints may
        # extend past Stage-1 support and are explicitly audited below.
        d = d[d["time_min"] <= endpoint].copy()
        d["endpoint_scenario"] = scenario
        d["endpoint_min"] = endpoint
        d["true_rul_min"] = np.maximum(endpoint - d["time_min"], 0.0)
        for horizon in cfg.horizons_min:
            tag = str(horizon).replace(".", "p")
            d[f"fail_within_{tag}m"] = (d["true_rul_min"] <= horizon).astype(int)
        pieces.append(d)
        audit.append(
            {
                "scenario": scenario,
                "pipe": pipe,
                "endpoint_min": endpoint,
                "available_end_min": available_end,
                "unsupported_tail_min": max(endpoint - available_end, 0.0),
                "rows": len(d),
            }
        )
    return pd.concat(pieces, ignore_index=True), pd.DataFrame(audit)


def sample_timeline(table: pd.DataFrame, cfg: HazardConfig = CFG) -> pd.DataFrame:
    d = table[table["time_min"] >= cfg.analysis_start_min].copy()
    d["sample_bin"] = np.floor(d["time_min"] / cfg.sample_minutes).astype(int)
    d = (
        d.sort_values(["pipe", "time_min"])
        .groupby(["pipe", "sample_bin"], sort=False, as_index=False)
        .tail(1)
        .drop(columns="sample_bin")
        .reset_index(drop=True)
    )
    return d


PRIMARY_SCENARIO = "shutdown_onset"
PRIMARY_FULL, PRIMARY_SUPPORT_AUDIT = attach_scenario_labels(TIME_RAW, PRIMARY_SCENARIO, CFG)
PRIMARY = sample_timeline(PRIMARY_FULL, CFG)

support_tables = []
for scenario in CFG.endpoint_scenarios:
    _, a = attach_scenario_labels(TIME_RAW, scenario, CFG)
    support_tables.append(a)
SUPPORT_AUDIT = pd.concat(support_tables, ignore_index=True)
SUPPORT_AUDIT.to_csv(CFG.output_dir / "02_endpoint_scenario_support_audit.csv", index=False)

print(SUPPORT_AUDIT.round(3).to_string(index=False))
print("\nPrimary sampled rows per pipe:")
print(PRIMARY.groupby("pipe").size().to_string())


# %% CELL 4 — Causal AE dynamics and feature families

PRESSURE_FEATURES = [
    "pressure_pd",
    "delta_pressure_pd",
    "abs_delta_pressure_pd",
    "time_in_hold_min",
    "dose_1",
    "dose_2",
    "dose_4",
]

AE_FEATURES = [
    "log_rms_med",
    "log_rms_iqr",
    "log_power_med",
    "log_power_iqr",
    "hf_logratio_med",
    "hf_logratio_iqr",
    "spec_centroid_khz_med",
    "spec_centroid_khz_iqr",
    "sensor_share_entropy",
    "sensor_share_dominance",
    "active_sensor_fraction",
    "pooled_hf_logratio",
]


def robust_mad(x: np.ndarray, axis=None) -> np.ndarray:
    x = np.asarray(x, dtype=float)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", category=RuntimeWarning)
        med = np.nanmedian(x, axis=axis, keepdims=True)
        mad = np.nanmedian(np.abs(x - med), axis=axis)
    return 1.4826 * mad + EPS


def add_causal_dynamics(
    table: pd.DataFrame,
    features: Sequence[str],
    prefix: str,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, List[str]]:
    """Create only backward-looking changes, rolling dispersion, and baseline z scores."""
    pieces = []
    created: List[str] = []
    for pipe, src in table.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        dt = float(np.nanmedian(np.diff(d["time_min"]))) if len(d) > 1 else cfg.sample_minutes
        dt = max(dt, 1e-6)
        baseline_mask = d["time_min"] <= (
            float(d["time_min"].min()) + cfg.baseline_minutes
        )
        if baseline_mask.sum() < 5:
            baseline_mask.iloc[: min(10, len(d))] = True

        for feature in features:
            if feature not in d:
                continue
            x = d[feature].astype(float)
            base = x[baseline_mask].to_numpy(float)
            med = float(np.nanmedian(base))
            scale = float(robust_mad(base))

            current = f"{prefix}{feature}__current"
            basez = f"{prefix}{feature}__baseline_z"
            d[current] = x
            d[basez] = (x - med) / scale
            for name in (current, basez):
                if name not in created:
                    created.append(name)

            for window in cfg.rolling_windows_min:
                n = max(1, int(round(window / dt)))
                tag = str(window).replace(".", "p")
                delta = f"{prefix}{feature}__delta_{tag}m"
                spread = f"{prefix}{feature}__std_{tag}m"
                d[delta] = x - x.shift(n)
                d[spread] = x.rolling(n, min_periods=max(2, n // 3)).std()
                for name in (delta, spread):
                    if name not in created:
                        created.append(name)
        pieces.append(d)
    return pd.concat(pieces, ignore_index=True), created


def balanced_pipe_weights(df: pd.DataFrame) -> np.ndarray:
    counts = df["pipe"].value_counts()
    w = df["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    return w / np.mean(w)


def pressure_feature_family() -> List[str]:
    return ["time_min"] + PRESSURE_FEATURES


# %% CELL 5 — Training-pipe-only healthy-envelope residualization

class HealthyPressureResidualizer:
    """
    Learn the expected AE response to load from non-terminal training-pipe data.

    Unlike the Stage-1 exploratory residualizer, the healthy envelope excludes
    the final ``healthy_margin_min`` of each training pipe. This prevents the
    pressure model from explaining away the damage signal it is meant to expose.
    """

    def __init__(self, cfg: HazardConfig = CFG):
        self.cfg = cfg
        self.models: Dict[str, HistGradientBoostingRegressor] = {}
        self.pressure_fill: Optional[pd.Series] = None

    def fit(self, train: pd.DataFrame) -> "HealthyPressureResidualizer":
        healthy = train[train["true_rul_min"] > self.cfg.healthy_margin_min].copy()
        if healthy.groupby("pipe").size().min() < 10:
            warnings.warn(
                "Healthy-envelope filter left too few rows; using all training rows."
            )
            healthy = train.copy()

        X = healthy[PRESSURE_FEATURES].replace([np.inf, -np.inf], np.nan)
        self.pressure_fill = X.median(numeric_only=True)
        X = X.fillna(self.pressure_fill)
        weights = balanced_pipe_weights(healthy)

        self.models = {}
        for feature in AE_FEATURES:
            y = healthy[feature].to_numpy(float)
            finite = np.isfinite(y)
            model = HistGradientBoostingRegressor(
                loss="absolute_error",
                learning_rate=0.045,
                max_iter=self.cfg.pressure_model_max_iter,
                max_leaf_nodes=7,
                min_samples_leaf=max(10, int(0.02 * finite.sum())),
                l2_regularization=3.0,
                random_state=self.cfg.random_state,
            )
            model.fit(X.loc[finite], y[finite], sample_weight=weights[finite])
            self.models[feature] = model
        return self

    def transform(self, table: pd.DataFrame) -> pd.DataFrame:
        if self.pressure_fill is None or not self.models:
            raise RuntimeError("Residualizer has not been fitted")
        out = table.copy()
        X = out[PRESSURE_FEATURES].replace([np.inf, -np.inf], np.nan)
        X = X.fillna(self.pressure_fill)
        for feature, model in self.models.items():
            expected = model.predict(X)
            out[f"expected_{feature}"] = expected
            out[f"resid_{feature}"] = out[feature].to_numpy(float) - expected
        return out


def prepare_feature_pair(
    train: pd.DataFrame,
    other: pd.DataFrame,
    method: str,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame, List[str], Optional[HealthyPressureResidualizer]]:
    pressure = pressure_feature_family()

    if method == "Age only":
        return train.copy(), other.copy(), ["time_min"], None
    if method == "Pressure only":
        return train.copy(), other.copy(), pressure, None

    if method == "Pressure + raw AE":
        tr, ae_cols = add_causal_dynamics(train, AE_FEATURES, "raw_", cfg)
        ot, _ = add_causal_dynamics(other, AE_FEATURES, "raw_", cfg)
        return tr, ot, pressure + ae_cols, None

    if method == "PRISM-Hazard":
        residualizer = HealthyPressureResidualizer(cfg).fit(train)
        tr0 = residualizer.transform(train)
        ot0 = residualizer.transform(other)
        residual_features = [f"resid_{f}" for f in AE_FEATURES]
        tr, resid_cols = add_causal_dynamics(tr0, residual_features, "", cfg)
        ot, _ = add_causal_dynamics(ot0, residual_features, "", cfg)
        return tr, ot, pressure + resid_cols, residualizer

    raise KeyError(f"Unknown method: {method}")


# %% CELL 6 — Nested held-out-pipe probability models and cross-fitting

def horizon_label_column(horizon_min: float) -> str:
    return f"fail_within_{str(horizon_min).replace('.', 'p')}m"


def make_probability_model(c_value: float, seed: int = SEED) -> Pipeline:
    return Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
            ("scale", StandardScaler()),
            (
                "logistic",
                LogisticRegression(
                    C=float(c_value),
                    solver="lbfgs",
                    max_iter=4000,
                    random_state=seed,
                ),
            ),
        ]
    )


def fit_probability_model(
    model: Pipeline,
    X: pd.DataFrame,
    y: np.ndarray,
    groups: np.ndarray,
) -> Pipeline:
    frame = pd.DataFrame({"pipe": groups})
    counts = frame["pipe"].value_counts()
    weights = frame["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    weights /= weights.mean()
    model.fit(X, y, logistic__sample_weight=weights)
    return model


def group_balanced_brier(
    y: np.ndarray, probability: np.ndarray, groups: np.ndarray
) -> float:
    scores = []
    for pipe in np.unique(groups):
        m = groups == pipe
        scores.append(float(np.mean((probability[m] - y[m]) ** 2)))
    return float(np.mean(scores))


def fit_platt_calibrator(
    scores: np.ndarray,
    y: np.ndarray,
    groups: np.ndarray,
) -> Optional[LogisticRegression]:
    finite = np.isfinite(scores) & np.isfinite(y)
    scores, y, groups = scores[finite], y[finite].astype(int), groups[finite]
    if len(np.unique(y)) < 2:
        return None
    frame = pd.DataFrame({"pipe": groups})
    counts = frame["pipe"].value_counts()
    weights = frame["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    weights /= weights.mean()
    calibrator = LogisticRegression(
        C=100.0, solver="lbfgs", max_iter=3000, random_state=SEED
    )
    calibrator.fit(scores.reshape(-1, 1), y, sample_weight=weights)
    return calibrator


def apply_calibrator(
    calibrator: Optional[LogisticRegression], scores: np.ndarray
) -> np.ndarray:
    if calibrator is None:
        return expit(scores)
    return calibrator.predict_proba(scores.reshape(-1, 1))[:, 1]


def prepare_inner_splits(
    outer_train: pd.DataFrame,
    method: str,
    cfg: HazardConfig = CFG,
) -> List[Dict[str, object]]:
    prepared = []
    for held_pipe in sorted(outer_train["pipe"].unique()):
        tr0 = outer_train[outer_train["pipe"] != held_pipe].copy()
        va0 = outer_train[outer_train["pipe"] == held_pipe].copy()
        tr, va, features, _ = prepare_feature_pair(tr0, va0, method, cfg)
        prepared.append(
            {
                "held_pipe": held_pipe,
                "train": tr,
                "valid": va,
                "features": features,
            }
        )
    return prepared


def crossfit_for_c(
    prepared: Sequence[Mapping[str, object]],
    c_value: float,
    label_col: str,
) -> pd.DataFrame:
    rows = []
    for split in prepared:
        tr = split["train"]
        va = split["valid"]
        features = split["features"]
        model = make_probability_model(c_value)
        fit_probability_model(
            model,
            tr[features],
            tr[label_col].to_numpy(int),
            tr["pipe"].to_numpy(),
        )
        score = model.decision_function(va[features])
        block = va[
            ["pipe", "time_min", "true_rul_min", "endpoint_min", label_col]
        ].copy()
        block = block.rename(columns={label_col: "event"})
        block["raw_score"] = score
        block["raw_probability"] = expit(score)
        rows.append(block)
    return pd.concat(rows, ignore_index=True)


FEATURE_PREPARATION_CACHE: Dict[Tuple[object, ...], Dict[str, object]] = {}


def scenario_table_signature(table: pd.DataFrame) -> Tuple[object, ...]:
    endpoints = (
        table.groupby("pipe")["endpoint_min"].max().reindex(CFG.pipes).round(6).tolist()
    )
    return tuple(endpoints) + (len(table),)


def cross_calibrated_oof(oof: pd.DataFrame) -> Tuple[pd.DataFrame, Optional[LogisticRegression]]:
    """Cross-calibrate each training pipe without using that pipe in its calibrator."""
    out = oof.copy()
    out["probability"] = np.nan
    for held in out["pipe"].unique():
        tr = out[out["pipe"] != held]
        va = out[out["pipe"] == held]
        calibrator = fit_platt_calibrator(
            tr["raw_score"].to_numpy(float),
            tr["event"].to_numpy(int),
            tr["pipe"].to_numpy(),
        )
        out.loc[va.index, "probability"] = apply_calibrator(
            calibrator, va["raw_score"].to_numpy(float)
        )

    final_calibrator = fit_platt_calibrator(
        out["raw_score"].to_numpy(float),
        out["event"].to_numpy(int),
        out["pipe"].to_numpy(),
    )
    return out, final_calibrator


def fit_outer_fold(
    scenario_table: pd.DataFrame,
    test_pipe: str,
    method: str,
    horizon_min: float,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame, Dict[str, object], pd.DataFrame]:
    label_col = horizon_label_column(horizon_min)
    outer_train = scenario_table[scenario_table["pipe"] != test_pipe].copy()
    outer_test = scenario_table[scenario_table["pipe"] == test_pipe].copy()
    cache_key = (scenario_table_signature(scenario_table), test_pipe, method)
    if cache_key not in FEATURE_PREPARATION_CACHE:
        prepared = prepare_inner_splits(outer_train, method, cfg)
        train_ready, test_ready, features, residualizer = prepare_feature_pair(
            outer_train, outer_test, method, cfg
        )
        FEATURE_PREPARATION_CACHE[cache_key] = {
            "prepared": prepared,
            "train_ready": train_ready,
            "test_ready": test_ready,
            "features": features,
            "residualizer": residualizer,
        }
    cached = FEATURE_PREPARATION_CACHE[cache_key]
    prepared = cached["prepared"]

    tuning_rows = []
    oof_by_c: Dict[float, pd.DataFrame] = {}
    for c_value in cfg.logistic_c_grid:
        oof = crossfit_for_c(prepared, c_value, label_col)
        score = group_balanced_brier(
            oof["event"].to_numpy(int),
            oof["raw_probability"].to_numpy(float),
            oof["pipe"].to_numpy(),
        )
        tuning_rows.append({"C": c_value, "group_Brier": score})
        oof_by_c[float(c_value)] = oof

    tuning = pd.DataFrame(tuning_rows).sort_values(["group_Brier", "C"])
    best_c = float(tuning.iloc[0]["C"])
    oof, final_calibrator = cross_calibrated_oof(oof_by_c[best_c])

    train_ready = cached["train_ready"]
    test_ready = cached["test_ready"]
    features = cached["features"]
    residualizer = cached["residualizer"]
    model = make_probability_model(best_c)
    fit_probability_model(
        model,
        train_ready[features],
        train_ready[label_col].to_numpy(int),
        train_ready["pipe"].to_numpy(),
    )
    test_score = model.decision_function(test_ready[features])
    test_probability = apply_calibrator(final_calibrator, test_score)

    pred = test_ready[
        ["pipe", "time_min", "true_rul_min", "endpoint_min", label_col]
    ].copy()
    pred = pred.rename(columns={label_col: "event"})
    pred["probability"] = np.clip(test_probability, 1e-6, 1 - 1e-6)
    pred["raw_score"] = test_score
    pred["method"] = method
    pred["horizon_min"] = horizon_min
    pred["outer_test_pipe"] = test_pipe
    pred["selected_C"] = best_c

    oof["method"] = method
    oof["horizon_min"] = horizon_min
    oof["outer_test_pipe"] = test_pipe
    oof["selected_C"] = best_c

    artifact = {
        "method": method,
        "horizon_min": horizon_min,
        "test_pipe": test_pipe,
        "selected_C": best_c,
        "features": features,
        "model": model,
        "calibrator": final_calibrator,
        "residualizer": residualizer,
    }
    tuning["method"] = method
    tuning["horizon_min"] = horizon_min
    tuning["outer_test_pipe"] = test_pipe
    return pred, oof, artifact, tuning


# %% CELL 7 — Execute strict LOPO dynamic-risk evaluation

PRIMARY_PREDICTIONS: List[pd.DataFrame] = []
PRIMARY_OOF: List[pd.DataFrame] = []
PRIMARY_TUNING: List[pd.DataFrame] = []
PRIMARY_ARTIFACTS: Dict[Tuple[float, str, str], Dict[str, object]] = {}

for horizon in CFG.horizons_min:
    for test_pipe in CFG.pipes:
        print(f"\n=== Horizon {horizon:g} min | held-out pipe {test_pipe} ===")
        for method in METHOD_ORDER:
            pred, oof, artifact, tuning = fit_outer_fold(
                PRIMARY, test_pipe, method, horizon, CFG
            )
            PRIMARY_PREDICTIONS.append(pred)
            PRIMARY_OOF.append(oof)
            PRIMARY_TUNING.append(tuning)
            PRIMARY_ARTIFACTS[(float(horizon), method, test_pipe)] = artifact
            brier = float(np.mean((pred["probability"] - pred["event"]) ** 2))
            print(f"{method:24s} C={artifact['selected_C']:<8g} Brier={brier:.5f}")

PREDICTIONS = pd.concat(PRIMARY_PREDICTIONS, ignore_index=True)
OOF_PREDICTIONS = pd.concat(PRIMARY_OOF, ignore_index=True)
TUNING = pd.concat(PRIMARY_TUNING, ignore_index=True)

PREDICTIONS.to_csv(CFG.output_dir / "03_lopo_hazard_predictions.csv", index=False)
OOF_PREDICTIONS.to_csv(CFG.output_dir / "03_crossfitted_training_predictions.csv", index=False)
TUNING.to_csv(CFG.output_dir / "03_nested_group_tuning.csv", index=False)


# %% CELL 8 — Discrimination, calibration, and incremental-risk metrics

def expected_calibration_error(
    y: np.ndarray, probability: np.ndarray, n_bins: int = 10
) -> float:
    y = np.asarray(y, dtype=float)
    p = np.asarray(probability, dtype=float)
    edges = np.linspace(0.0, 1.0, n_bins + 1)
    value = 0.0
    for lo, hi in zip(edges[:-1], edges[1:]):
        mask = (p >= lo) & (p < hi if hi < 1.0 else p <= hi)
        if mask.any():
            value += mask.mean() * abs(float(y[mask].mean() - p[mask].mean()))
    return float(value)


def safe_auc(y: np.ndarray, p: np.ndarray, kind: str) -> float:
    if len(np.unique(y)) < 2:
        return np.nan
    if kind == "roc":
        return float(roc_auc_score(y, p))
    return float(average_precision_score(y, p))


def calibration_intercept_slope(y: np.ndarray, p: np.ndarray) -> Tuple[float, float]:
    if len(np.unique(y)) < 2:
        return np.nan, np.nan
    z = logit(np.clip(p, 1e-5, 1 - 1e-5)).reshape(-1, 1)
    model = LogisticRegression(C=1e4, solver="lbfgs", max_iter=3000)
    model.fit(z, y)
    return float(model.intercept_[0]), float(model.coef_[0, 0])


def metric_rows(predictions: pd.DataFrame) -> pd.DataFrame:
    rows = []
    keys = ["horizon_min", "method", "pipe"]
    for (horizon, method, pipe), d in predictions.groupby(keys, sort=False):
        y = d["event"].to_numpy(int)
        p = np.clip(d["probability"].to_numpy(float), 1e-6, 1 - 1e-6)
        intercept, slope = calibration_intercept_slope(y, p)
        rows.append(
            {
                "horizon_min": float(horizon),
                "method": method,
                "pipe": pipe,
                "n": len(d),
                "event_prevalence": float(y.mean()),
                "Brier": float(np.mean((p - y) ** 2)),
                "log_loss": float(log_loss(y, p, labels=[0, 1])),
                "ROC_AUC": safe_auc(y, p, "roc"),
                "PR_AUC": safe_auc(y, p, "pr"),
                "ECE": expected_calibration_error(y, p),
                "calibration_intercept": intercept,
                "calibration_slope": slope,
            }
        )
    return pd.DataFrame(rows)


METRICS_BY_PIPE = metric_rows(PREDICTIONS)
METRICS_SUMMARY = (
    METRICS_BY_PIPE.groupby(["horizon_min", "method"], as_index=False)
    .agg(
        Brier_mean=("Brier", "mean"),
        Brier_std=("Brier", "std"),
        log_loss_mean=("log_loss", "mean"),
        ROC_AUC_mean=("ROC_AUC", "mean"),
        PR_AUC_mean=("PR_AUC", "mean"),
        ECE_mean=("ECE", "mean"),
        calibration_slope_mean=("calibration_slope", "mean"),
    )
)

pressure_brier = METRICS_BY_PIPE[
    METRICS_BY_PIPE["method"] == "Pressure only"
][["horizon_min", "pipe", "Brier"]].rename(columns={"Brier": "pressure_Brier"})
INCREMENTAL = METRICS_BY_PIPE.merge(pressure_brier, on=["horizon_min", "pipe"])
INCREMENTAL["Brier_skill_vs_pressure"] = 1.0 - (
    INCREMENTAL["Brier"] / (INCREMENTAL["pressure_Brier"] + EPS)
)

METRICS_BY_PIPE.to_csv(CFG.output_dir / "04_hazard_metrics_by_pipe.csv", index=False)
METRICS_SUMMARY.to_csv(CFG.output_dir / "04_hazard_metrics_summary.csv", index=False)
INCREMENTAL.to_csv(CFG.output_dir / "04_incremental_hazard_value.csv", index=False)

print("\nGroup-balanced hazard summary:")
print(METRICS_SUMMARY.round(4).to_string(index=False))
print("\nPositive Brier skill means improvement over pressure-only:")
print(
    INCREMENTAL[
        ["horizon_min", "method", "pipe", "Brier_skill_vs_pressure"]
    ].round(4).to_string(index=False)
)


# %% CELL 9 — Publication figures for horizon value and calibration

def plot_brier_by_horizon(summary: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    fig, ax = plt.subplots(figsize=(8.2, 4.8))
    for method in METHOD_ORDER:
        d = summary[summary["method"] == method].sort_values("horizon_min")
        ax.plot(
            d["horizon_min"], d["Brier_mean"], marker="o", lw=1.8,
            color=METHOD_COLORS[method], label=method
        )
        ax.fill_between(
            d["horizon_min"].to_numpy(float),
            np.maximum(d["Brier_mean"] - d["Brier_std"], 0).to_numpy(float),
            (d["Brier_mean"] + d["Brier_std"]).to_numpy(float),
            color=METHOD_COLORS[method], alpha=0.10,
        )
    ax.set_xlabel("Failure horizon (min)")
    ax.set_ylabel("Group-balanced Brier score ↓")
    ax.set_title("Does AE improve dynamic risk beyond the pressure clock?")
    ax.legend(ncol=2)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "05_brier_by_failure_horizon.png")
    plt.show()


def plot_primary_calibration(predictions: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    d0 = predictions[predictions["horizon_min"] == cfg.primary_horizon_min]
    fig, ax = plt.subplots(figsize=(6.4, 5.5))
    ax.plot([0, 1], [0, 1], "--", color="black", lw=1.2, label="Ideal")
    edges = np.linspace(0, 1, 9)
    for method in METHOD_ORDER:
        d = d0[d0["method"] == method]
        points = []
        for lo, hi in zip(edges[:-1], edges[1:]):
            m = (d["probability"] >= lo) & (
                d["probability"] < hi if hi < 1 else d["probability"] <= hi
            )
            if m.sum() >= 4:
                # Equal pipe weight inside each plotted bin.
                by_pipe = d[m].groupby("pipe").agg(
                    predicted=("probability", "mean"), observed=("event", "mean")
                )
                points.append((by_pipe["predicted"].mean(), by_pipe["observed"].mean()))
        if points:
            x, y = np.asarray(points).T
            ax.plot(x, y, "o-", lw=1.5, color=METHOD_COLORS[method], label=method)
    ax.set_xlim(-0.02, 1.02)
    ax.set_ylim(-0.02, 1.02)
    ax.set_xlabel("Predicted failure probability")
    ax.set_ylabel("Observed event frequency")
    ax.set_title(f"Held-out-pipe calibration: failure within {cfg.primary_horizon_min:g} min")
    ax.legend(fontsize=8)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "06_primary_horizon_calibration.png")
    plt.show()


def plot_pipe_brier_skill(incremental: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    d = incremental[
        (incremental["horizon_min"] == cfg.primary_horizon_min)
        & incremental["method"].isin(["Pressure + raw AE", "PRISM-Hazard"])
    ]
    fig, ax = plt.subplots(figsize=(7.8, 4.6))
    x = np.arange(len(cfg.pipes))
    width = 0.36
    for j, method in enumerate(["Pressure + raw AE", "PRISM-Hazard"]):
        values = d[d["method"] == method].set_index("pipe").reindex(cfg.pipes)
        ax.bar(
            x + (j - 0.5) * width,
            values["Brier_skill_vs_pressure"],
            width,
            color=METHOD_COLORS[method],
            edgecolor="black",
            linewidth=0.6,
            label=method,
        )
    ax.axhline(0, color="black", lw=1)
    ax.set_xticks(x, [f"Pipe {p}" for p in cfg.pipes])
    ax.set_ylabel("Brier skill versus pressure-only ↑")
    ax.set_title(f"Incremental AE value at {cfg.primary_horizon_min:g}-minute horizon")
    ax.legend()
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "07_primary_horizon_incremental_value.png")
    plt.show()


plot_brier_by_horizon(METRICS_SUMMARY, CFG)
plot_primary_calibration(PREDICTIONS, CFG)
plot_pipe_brier_skill(INCREMENTAL, CFG)


# %% CELL 10 — Training-only sequential alarm policy

def persistent_alarm(probability: np.ndarray, threshold: float, persistence: int) -> np.ndarray:
    above = np.asarray(probability, dtype=float) >= float(threshold)
    if persistence <= 1:
        return above
    run = pd.Series(above.astype(int)).rolling(persistence, min_periods=persistence).sum()
    return (run.to_numpy(float) >= persistence)


def evaluate_alarm_policy(
    frame: pd.DataFrame,
    threshold: float,
    persistence: int,
    horizon_min: float,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, Dict[str, float]]:
    pipe_rows = []
    for pipe, src in frame.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        alarm = persistent_alarm(d["probability"].to_numpy(float), threshold, persistence)
        starts = alarm & ~np.r_[False, alarm[:-1]]
        preterminal = d["true_rul_min"].to_numpy(float) > horizon_min
        terminal = ~preterminal
        false_events = int(np.sum(starts & preterminal))
        exposure_h = max(float(np.sum(preterminal) * cfg.sample_minutes / 60.0), EPS)
        timely = alarm & terminal
        detected = bool(timely.any())
        lead = float(np.max(d.loc[timely, "true_rul_min"])) if detected else 0.0
        pipe_rows.append(
            {
                "pipe": pipe,
                "detected": int(detected),
                "lead_time_min": lead,
                "false_alarm_events": false_events,
                "preterminal_exposure_h": exposure_h,
                "false_alarm_events_per_h": false_events / exposure_h,
            }
        )
    detail = pd.DataFrame(pipe_rows)
    aggregate = {
        "detection_rate": float(detail["detected"].mean()),
        "mean_lead_time_min": float(detail.loc[detail["detected"] == 1, "lead_time_min"].mean())
        if detail["detected"].any() else 0.0,
        "false_alarm_events_per_h": float(
            detail["false_alarm_events"].sum()
            / max(detail["preterminal_exposure_h"].sum(), EPS)
        ),
    }
    return detail, aggregate


def select_alarm_policy(
    training_oof: pd.DataFrame,
    horizon_min: float,
    cfg: HazardConfig = CFG,
) -> Tuple[float, int, pd.DataFrame]:
    rows = []
    for threshold in cfg.alarm_threshold_grid:
        for persistence in cfg.alarm_persistence_grid:
            _, result = evaluate_alarm_policy(
                training_oof, threshold, persistence, horizon_min, cfg
            )
            rows.append(
                {
                    "threshold": threshold,
                    "persistence_bins": persistence,
                    **result,
                }
            )
    search = pd.DataFrame(rows)
    feasible = search[
        search["false_alarm_events_per_h"] <= cfg.target_false_alarm_events_per_hour
    ]
    if not feasible.empty:
        selected = feasible.sort_values(
            ["detection_rate", "mean_lead_time_min", "false_alarm_events_per_h", "threshold"],
            ascending=[False, False, True, False],
        ).iloc[0]
    else:
        selected = search.sort_values(
            ["false_alarm_events_per_h", "detection_rate", "mean_lead_time_min"],
            ascending=[True, False, False],
        ).iloc[0]
    return float(selected["threshold"]), int(selected["persistence_bins"]), search


ALARM_POLICIES = []
ALARM_BY_PIPE = []
ALARM_SEARCH = []
horizon = CFG.primary_horizon_min

for test_pipe in CFG.pipes:
    for method in METHOD_ORDER:
        train_oof = OOF_PREDICTIONS[
            (OOF_PREDICTIONS["outer_test_pipe"] == test_pipe)
            & (OOF_PREDICTIONS["horizon_min"] == horizon)
            & (OOF_PREDICTIONS["method"] == method)
        ].copy()
        test = PREDICTIONS[
            (PREDICTIONS["outer_test_pipe"] == test_pipe)
            & (PREDICTIONS["horizon_min"] == horizon)
            & (PREDICTIONS["method"] == method)
        ].copy()
        threshold, persistence, search = select_alarm_policy(train_oof, horizon, CFG)
        detail, result = evaluate_alarm_policy(test, threshold, persistence, horizon, CFG)
        ALARM_POLICIES.append(
            {
                "outer_test_pipe": test_pipe,
                "method": method,
                "horizon_min": horizon,
                "threshold": threshold,
                "persistence_bins": persistence,
                **result,
            }
        )
        detail["outer_test_pipe"] = test_pipe
        detail["method"] = method
        detail["horizon_min"] = horizon
        detail["threshold"] = threshold
        detail["persistence_bins"] = persistence
        ALARM_BY_PIPE.append(detail)
        search["outer_test_pipe"] = test_pipe
        search["method"] = method
        search["horizon_min"] = horizon
        ALARM_SEARCH.append(search)

ALARM_POLICIES = pd.DataFrame(ALARM_POLICIES)
ALARM_BY_PIPE = pd.concat(ALARM_BY_PIPE, ignore_index=True)
ALARM_SEARCH = pd.concat(ALARM_SEARCH, ignore_index=True)
ALARM_POLICIES.to_csv(CFG.output_dir / "08_alarm_policy_lopo.csv", index=False)
ALARM_BY_PIPE.to_csv(CFG.output_dir / "08_alarm_results_by_pipe.csv", index=False)
ALARM_SEARCH.to_csv(CFG.output_dir / "08_alarm_policy_training_search.csv", index=False)

print("\nHeld-out-pipe sequential alarm results:")
print(ALARM_POLICIES.round(4).to_string(index=False))


def plot_primary_risk_trajectories(
    predictions: pd.DataFrame,
    policies: pd.DataFrame,
    cfg: HazardConfig = CFG,
) -> None:
    fig, axes = plt.subplots(len(cfg.pipes), 1, figsize=(12.5, 10.5), sharex=False)
    for ax, pipe in zip(axes, cfg.pipes):
        for method in ["Pressure only", "Pressure + raw AE", "PRISM-Hazard"]:
            d = predictions[
                (predictions["pipe"] == pipe)
                & (predictions["horizon_min"] == cfg.primary_horizon_min)
                & (predictions["method"] == method)
            ].sort_values("time_min")
            ax.plot(
                d["time_min"], d["probability"], lw=1.25,
                color=METHOD_COLORS[method], label=method
            )
        policy = policies[
            (policies["outer_test_pipe"] == pipe)
            & (policies["method"] == "PRISM-Hazard")
        ].iloc[0]
        endpoint = float(d["endpoint_min"].iloc[0])
        ax.axvspan(
            endpoint - cfg.primary_horizon_min, endpoint,
            color="#CC3311", alpha=0.10, label="True failure horizon"
        )
        ax.axhline(
            policy["threshold"], color="#CC3311", ls="--", lw=1.0,
            label=f"PRISM alarm threshold ({int(policy['persistence_bins'])} bins)"
        )
        ax.set_ylim(-0.02, 1.02)
        ax.set_ylabel(f"Pipe {pipe}\nRisk")
        if pipe == cfg.pipes[0]:
            ax.legend(ncol=3, fontsize=8, loc="upper left")
    axes[-1].set_xlabel("Time from feature start (min)")
    fig.suptitle(
        f"Strict LOPO failure-within-{cfg.primary_horizon_min:g}-minute risk",
        y=1.005,
    )
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "09_lopo_dynamic_risk_trajectories.png")
    plt.show()


plot_primary_risk_trajectories(PREDICTIONS, ALARM_POLICIES, CFG)


# %% CELL 11 — Endpoint-label sensitivity under common observed support

SENSITIVITY_ROWS = []
SENSITIVITY_METHODS = ["Pressure only", "Pressure + raw AE", "PRISM-Hazard"]

for scenario in CFG.endpoint_scenarios:
    scenario_full, _ = attach_scenario_labels(TIME_RAW, scenario, CFG)
    scenario_table = sample_timeline(scenario_full, CFG)
    print(f"\n=== Endpoint sensitivity: {scenario} ===")
    for test_pipe in CFG.pipes:
        for method in SENSITIVITY_METHODS:
            pred, _, _, _ = fit_outer_fold(
                scenario_table,
                test_pipe,
                method,
                CFG.primary_horizon_min,
                CFG,
            )
            y = pred["event"].to_numpy(int)
            p = pred["probability"].to_numpy(float)
            SENSITIVITY_ROWS.append(
                {
                    "scenario": scenario,
                    "pipe": test_pipe,
                    "method": method,
                    "horizon_min": CFG.primary_horizon_min,
                    "Brier": float(np.mean((p - y) ** 2)),
                    "PR_AUC": safe_auc(y, p, "pr"),
                    "event_prevalence": float(y.mean()),
                }
            )

ENDPOINT_SENSITIVITY = pd.DataFrame(SENSITIVITY_ROWS)
pressure_reference = ENDPOINT_SENSITIVITY[
    ENDPOINT_SENSITIVITY["method"] == "Pressure only"
][["scenario", "pipe", "Brier"]].rename(columns={"Brier": "pressure_Brier"})
ENDPOINT_SENSITIVITY = ENDPOINT_SENSITIVITY.merge(
    pressure_reference, on=["scenario", "pipe"], how="left"
)
ENDPOINT_SENSITIVITY["Brier_skill_vs_pressure"] = 1.0 - (
    ENDPOINT_SENSITIVITY["Brier"] / (ENDPOINT_SENSITIVITY["pressure_Brier"] + EPS)
)
ENDPOINT_SENSITIVITY.to_csv(
    CFG.output_dir / "10_endpoint_sensitivity_metrics.csv", index=False
)

ENDPOINT_SENSITIVITY_SUMMARY = (
    ENDPOINT_SENSITIVITY.groupby(["scenario", "method"], as_index=False)
    .agg(
        Brier_mean=("Brier", "mean"),
        PR_AUC_mean=("PR_AUC", "mean"),
        Brier_skill_mean=("Brier_skill_vs_pressure", "mean"),
        pipes_with_positive_skill=("Brier_skill_vs_pressure", lambda x: int((x > 0).sum())),
    )
)
ENDPOINT_SENSITIVITY_SUMMARY.to_csv(
    CFG.output_dir / "10_endpoint_sensitivity_summary.csv", index=False
)
print(ENDPOINT_SENSITIVITY_SUMMARY.round(4).to_string(index=False))


def plot_endpoint_sensitivity(
    sensitivity: pd.DataFrame, cfg: HazardConfig = CFG
) -> None:
    methods = ["Pressure + raw AE", "PRISM-Hazard"]
    scenarios = list(cfg.endpoint_scenarios)
    matrix = np.full((len(scenarios), len(methods)), np.nan)
    for i, scenario in enumerate(scenarios):
        for j, method in enumerate(methods):
            d = sensitivity[
                (sensitivity["scenario"] == scenario)
                & (sensitivity["method"] == method)
            ]
            matrix[i, j] = d["Brier_skill_vs_pressure"].mean()

    bound = max(0.25, float(np.nanmax(np.abs(matrix))))
    fig, ax = plt.subplots(figsize=(7.4, 4.2))
    im = ax.imshow(matrix, cmap="RdBu_r", vmin=-bound, vmax=bound, aspect="auto")
    ax.set_xticks(range(len(methods)), methods)
    ax.set_yticks(range(len(scenarios)), scenarios)
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            ax.text(j, i, f"{matrix[i, j]:+.2f}", ha="center", va="center")
    ax.set_title("Endpoint-robust Brier skill versus pressure-only")
    fig.colorbar(im, ax=ax, label="Brier skill ↑")
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "10_endpoint_sensitivity_heatmap.png")
    plt.show()


plot_endpoint_sensitivity(ENDPOINT_SENSITIVITY, CFG)


# %% CELL 12 — Sensor dropout, gain drift, and energy-floor stress experiments

def rowwise_sensor_statistics(values: np.ndarray, prefix: str) -> Dict[str, np.ndarray]:
    return {
        f"{prefix}_med": np.nanmedian(values, axis=1),
        f"{prefix}_mean": np.nanmean(values, axis=1),
        f"{prefix}_std": np.nanstd(values, axis=1),
        f"{prefix}_iqr": np.nanpercentile(values, 75, axis=1)
        - np.nanpercentile(values, 25, axis=1),
        f"{prefix}_max": np.nanmax(values, axis=1),
    }


def normalized_sensor_entropy(shares: np.ndarray) -> np.ndarray:
    p = np.asarray(shares, dtype=float)
    valid = np.isfinite(p) & (p > 0)
    entropy = -np.nansum(np.where(valid, p * np.log(p + EPS), 0.0), axis=1)
    count = valid.sum(axis=1)
    denom = np.log(np.maximum(count, 2))
    return entropy / denom


def recompute_ae_aggregates(
    table: pd.DataFrame,
    rms: np.ndarray,
    energy: np.ndarray,
    cfg: HazardConfig = CFG,
) -> pd.DataFrame:
    out = table.copy()
    widths = np.array([hi - lo for lo, hi in cfg.bands_khz], dtype=float)
    centres = np.array([(lo + hi) / 2.0 for lo, hi in cfg.bands_khz], dtype=float)
    lf = np.asarray(cfg.lf_band_indices, dtype=int)
    hf = np.asarray(cfg.hf_band_indices, dtype=int)
    selected = np.r_[lf, hf]

    lf_density = np.nansum(energy[:, :, lf], axis=2) / widths[lf].sum()
    hf_density = np.nansum(energy[:, :, hf], axis=2) / widths[hf].sum()
    total = np.nansum(energy[:, :, selected], axis=2)
    centroid = np.nansum(
        energy[:, :, selected] * centres[selected][None, None, :], axis=2
    ) / (total + EPS)

    log_rms = np.log1p(np.maximum(rms, 0.0))
    log_power = np.log1p(np.maximum(total, 0.0))
    hf_logratio = np.log((hf_density + EPS) / (lf_density + EPS))

    for name, values in rowwise_sensor_statistics(log_rms, "log_rms").items():
        out[name] = values
    for name, values in rowwise_sensor_statistics(log_power, "log_power").items():
        out[name] = values
    for name, values in rowwise_sensor_statistics(hf_logratio, "hf_logratio").items():
        out[name] = values
    for name, values in rowwise_sensor_statistics(centroid, "spec_centroid_khz").items():
        out[name] = values

    shares = total / (np.nansum(total, axis=1, keepdims=True) + EPS)
    out["sensor_share_entropy"] = normalized_sensor_entropy(shares)
    out["sensor_share_dominance"] = np.nanmax(shares, axis=1)

    active_fraction = np.zeros(len(out), dtype=float)
    for pipe in out["pipe"].unique():
        idx = np.flatnonzero(out["pipe"].to_numpy() == pipe)
        d = out.iloc[idx]
        baseline = d["time_min"].to_numpy(float) <= (
            float(d["time_min"].min()) + cfg.baseline_minutes
        )
        if baseline.sum() < 5:
            baseline[: min(10, len(baseline))] = True
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            threshold = np.nanmedian(log_rms[idx][baseline], axis=0) + (
                6.0 * robust_mad(log_rms[idx][baseline], axis=0)
            )
        active = log_rms[idx] > threshold[None, :]
        valid = np.isfinite(log_rms[idx])
        active_fraction[idx] = np.sum(active & valid, axis=1) / np.maximum(valid.sum(axis=1), 1)
    out["active_sensor_fraction"] = active_fraction

    pooled_lf = np.nansum(energy[:, :, lf], axis=(1, 2)) / widths[lf].sum()
    pooled_hf = np.nansum(energy[:, :, hf], axis=(1, 2)) / widths[hf].sum()
    out["pooled_hf_logratio"] = np.log((pooled_hf + EPS) / (pooled_lf + EPS))
    return out


def stress_sensor_table(
    table: pd.DataFrame,
    scenario: str,
    repeat: int,
    cfg: HazardConfig = CFG,
) -> pd.DataFrame:
    scenario_seed = {
        "clean": 0,
        "drop_one_sensor": 101,
        "gain_drift": 211,
        "energy_floor": 307,
        "combined": 401,
    }[scenario]
    rng = np.random.default_rng(cfg.random_state + 1009 * repeat + scenario_seed)
    rms = np.column_stack(
        [table[f"rms_ch{ch}"].to_numpy(float) for ch in cfg.channels]
    )
    energy = np.empty((len(table), len(cfg.channels), len(cfg.bands_khz)), dtype=float)
    for j, ch in enumerate(cfg.channels):
        for b in range(len(cfg.bands_khz)):
            energy[:, j, b] = table[f"e_ch{ch}_b{b}"].to_numpy(float)

    if scenario in {"drop_one_sensor", "combined"}:
        j = repeat % len(cfg.channels)
        rms[:, j] = np.nan
        energy[:, j, :] = np.nan

    if scenario in {"gain_drift", "combined"}:
        gain = rng.lognormal(mean=0.0, sigma=cfg.gain_log_sigma, size=len(cfg.channels))
        rms *= gain[None, :]
        energy *= gain[None, :, None] ** 2

    if scenario in {"energy_floor", "combined"}:
        baseline_n = max(5, int(round(cfg.baseline_minutes / cfg.sample_minutes)))
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            floor = np.nanmedian(energy[:baseline_n], axis=0)
        random_floor = rng.uniform(0.5, 1.5, size=energy.shape)
        energy += cfg.noise_floor_fraction * floor[None, :, :] * random_floor
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            rms_floor = np.nanmedian(rms[:baseline_n], axis=0)
        rms = np.sqrt(
            np.maximum(rms, 0.0) ** 2
            + (cfg.noise_floor_fraction * rms_floor[None, :]) ** 2
        )

    return recompute_ae_aggregates(table, rms, energy, cfg)


def predict_with_artifact(
    artifact: Mapping[str, object],
    test_table: pd.DataFrame,
    cfg: HazardConfig = CFG,
) -> np.ndarray:
    method = artifact["method"]
    features = artifact["features"]
    if method == "PRISM-Hazard":
        residualizer = artifact["residualizer"]
        transformed = residualizer.transform(test_table)
        residual_features = [f"resid_{f}" for f in AE_FEATURES]
        ready, _ = add_causal_dynamics(transformed, residual_features, "", cfg)
    elif method == "Pressure + raw AE":
        ready, _ = add_causal_dynamics(test_table, AE_FEATURES, "raw_", cfg)
    else:
        ready = test_table.copy()
    score = artifact["model"].decision_function(ready[features])
    return np.clip(apply_calibrator(artifact["calibrator"], score), 1e-6, 1 - 1e-6)


STRESS_SCENARIOS = ["clean", "drop_one_sensor", "gain_drift", "energy_floor", "combined"]
STRESS_ROWS = []

for pipe in CFG.pipes:
    clean_test = PRIMARY[PRIMARY["pipe"] == pipe].copy()
    for method in ["Pressure + raw AE", "PRISM-Hazard"]:
        artifact = PRIMARY_ARTIFACTS[(CFG.primary_horizon_min, method, pipe)]
        repeats = {"clean": 1, **{s: CFG.stress_repeats for s in STRESS_SCENARIOS if s != "clean"}}
        for scenario, n_repeat in repeats.items():
            for repeat in range(n_repeat):
                stressed = (
                    clean_test.copy()
                    if scenario == "clean"
                    else stress_sensor_table(clean_test, scenario, repeat, CFG)
                )
                probability = predict_with_artifact(artifact, stressed, CFG)
                y = stressed[horizon_label_column(CFG.primary_horizon_min)].to_numpy(int)
                STRESS_ROWS.append(
                    {
                        "pipe": pipe,
                        "method": method,
                        "scenario": scenario,
                        "repeat": repeat,
                        "Brier": float(np.mean((probability - y) ** 2)),
                        "PR_AUC": safe_auc(y, probability, "pr"),
                        "mean_probability": float(np.mean(probability)),
                    }
                )

STRESS_RESULTS = pd.DataFrame(STRESS_ROWS)
clean_reference = STRESS_RESULTS[STRESS_RESULTS["scenario"] == "clean"][
    ["pipe", "method", "Brier"]
].rename(columns={"Brier": "clean_Brier"})
STRESS_RESULTS = STRESS_RESULTS.merge(clean_reference, on=["pipe", "method"], how="left")
STRESS_RESULTS["relative_Brier_degradation"] = (
    STRESS_RESULTS["Brier"] - STRESS_RESULTS["clean_Brier"]
) / (STRESS_RESULTS["clean_Brier"] + EPS)
STRESS_RESULTS.to_csv(CFG.output_dir / "11_sensor_stress_results.csv", index=False)

STRESS_SUMMARY = (
    STRESS_RESULTS.groupby(["method", "scenario"], as_index=False)
    .agg(
        Brier_mean=("Brier", "mean"),
        PR_AUC_mean=("PR_AUC", "mean"),
        relative_Brier_degradation_mean=("relative_Brier_degradation", "mean"),
        relative_Brier_degradation_max=("relative_Brier_degradation", "max"),
    )
)
STRESS_SUMMARY.to_csv(CFG.output_dir / "11_sensor_stress_summary.csv", index=False)
print(STRESS_SUMMARY.round(4).to_string(index=False))


def plot_sensor_stress(results: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    fig, ax = plt.subplots(figsize=(9.0, 4.8))
    scenarios = STRESS_SCENARIOS[1:]
    x = np.arange(len(scenarios))
    width = 0.36
    for j, method in enumerate(["Pressure + raw AE", "PRISM-Hazard"]):
        means, errors = [], []
        for scenario in scenarios:
            d = results[(results["method"] == method) & (results["scenario"] == scenario)]
            means.append(d["relative_Brier_degradation"].mean())
            errors.append(d["relative_Brier_degradation"].std())
        ax.bar(
            x + (j - 0.5) * width, means, width, yerr=errors,
            color=METHOD_COLORS[method], edgecolor="black", linewidth=0.6,
            capsize=3, label=method,
        )
    ax.axhline(0, color="black", lw=1)
    ax.set_xticks(x, [s.replace("_", "\n") for s in scenarios])
    ax.set_ylabel("Relative Brier degradation ↓")
    ax.set_title("Held-out-pipe sensor-stress robustness")
    ax.legend()
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "11_sensor_stress_robustness.png")
    plt.show()


plot_sensor_stress(STRESS_RESULTS, CFG)


# %% CELL 13 — Evidence-based Stage-2 decision gate

def stage2_decision_gate(cfg: HazardConfig = CFG) -> Dict[str, object]:
    h = cfg.primary_horizon_min
    primary = METRICS_BY_PIPE[METRICS_BY_PIPE["horizon_min"] == h]
    mean_brier = primary.groupby("method")["Brier"].mean()
    ae_candidates = ["Pressure + raw AE", "PRISM-Hazard"]
    selected_method = min(ae_candidates, key=lambda m: float(mean_brier.loc[m]))

    comparison = primary[primary["method"].isin(["Pressure only", selected_method])]
    wide = comparison.pivot(index="pipe", columns="method", values="Brier")
    pipe_skill = 1.0 - wide[selected_method] / (wide["Pressure only"] + EPS)
    mean_skill = float(pipe_skill.mean())
    wins = int((pipe_skill > 0).sum())

    ece = float(primary[primary["method"] == selected_method]["ECE"].mean())
    alarm = ALARM_POLICIES[ALARM_POLICIES["method"] == selected_method]
    detection_rate = float(alarm["detection_rate"].mean())
    false_alarm_rate = float(alarm["false_alarm_events_per_h"].mean())
    lead_time = float(alarm["mean_lead_time_min"].mean())

    sensitivity = ENDPOINT_SENSITIVITY[
        ENDPOINT_SENSITIVITY["method"] == selected_method
    ]
    scenario_skill = sensitivity.groupby("scenario")["Brier_skill_vs_pressure"].mean()
    worst_endpoint_skill = float(scenario_skill.min())

    stress = STRESS_RESULTS[STRESS_RESULTS["method"] == selected_method]
    combined = stress[stress["scenario"] == "combined"]
    combined_degradation = float(combined["relative_Brier_degradation"].mean())

    if (
        mean_skill >= 0.15
        and wins >= 3
        and ece <= 0.10
        and detection_rate >= 0.75
        and false_alarm_rate <= cfg.target_false_alarm_events_per_hour
        and worst_endpoint_skill > 0.0
    ):
        verdict = "GO — DYNAMIC HAZARD"
        next_step = (
            "Proceed to Bayesian/conformal uncertainty decomposition and decision-cost analysis."
        )
    elif (
        mean_skill > 0.0
        and wins >= 2
        and detection_rate >= 0.50
    ):
        verdict = "CONDITIONAL GO — HAZARD, NOT POINT RUL"
        next_step = (
            "Retain dynamic hazard as the paper target; improve calibration and endpoint/"
            "sensor robustness before adding a final uncertainty layer."
        )
    else:
        verdict = "PIVOT — CHANGE DETECTION / DATA LIMITATION STUDY"
        next_step = (
            "Do not claim transferable failure probability. Reframe as pressure-conditioned "
            "terminal-regime detection or a rigorous non-identifiability benchmark."
        )

    if not cfg.labels_are_physical or not cfg.pressure_alignment_is_external:
        verdict = "EXPLORATORY " + verdict
        next_step += (
            " External rupture timestamps and pressure alignment remain mandatory for a "
            "publication-grade reliability claim."
        )

    return {
        "verdict": verdict,
        "selected_AE_method": selected_method,
        "primary_horizon_min": h,
        "pressure_Brier_mean": float(mean_brier.loc["Pressure only"]),
        "selected_AE_Brier_mean": float(mean_brier.loc[selected_method]),
        "Brier_skill_vs_pressure_mean": mean_skill,
        "pipes_won_out_of_4": wins,
        "ECE_mean": ece,
        "alarm_detection_rate": detection_rate,
        "alarm_false_events_per_hour": false_alarm_rate,
        "alarm_mean_lead_time_min": lead_time,
        "worst_endpoint_scenario_skill": worst_endpoint_skill,
        "combined_sensor_stress_relative_Brier_degradation": combined_degradation,
        "labels_are_physical": cfg.labels_are_physical,
        "pressure_alignment_is_external": cfg.pressure_alignment_is_external,
        "next_step": next_step,
    }


DECISION = stage2_decision_gate(CFG)
print("\n" + json.dumps(DECISION, indent=2))


def json_safe(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, dict):
        return {str(k): json_safe(v) for k, v in value.items()}
    if isinstance(value, (tuple, list)):
        return [json_safe(v) for v in value]
    if isinstance(value, np.generic):
        return value.item()
    return value


manifest = {
    "project": "PRISM-Hazard Stage 2",
    "scientific_target": "calibrated failure-within-horizon probability",
    "primary_endpoint_scenario": PRIMARY_SCENARIO,
    "endpoint_scenarios": dict(CFG.endpoint_scenarios),
    "config": json_safe(asdict(CFG)),
    "stage1_manifest": STAGE1_MANIFEST,
    "decision": DECISION,
}
with open(CFG.output_dir / "12_stage2_reproducibility_manifest.json", "w", encoding="utf-8") as fh:
    json.dump(manifest, fh, indent=2)


# %% CELL 14 — Result handoff for the next research decision

print(
    """
PRISM-HAZARD STAGE 2 COMPLETE
-----------------------------
Send these files/results for interpretation:

1. 04_hazard_metrics_summary.csv
2. 04_incremental_hazard_value.csv
3. 05_brier_by_failure_horizon.png
4. 06_primary_horizon_calibration.png
5. 07_primary_horizon_incremental_value.png
6. 08_alarm_policy_lopo.csv
7. 09_lopo_dynamic_risk_trajectories.png
8. 10_endpoint_sensitivity_summary.csv
9. 10_endpoint_sensitivity_heatmap.png
10. 11_sensor_stress_summary.csv
11. 11_sensor_stress_robustness.png
12. The decision JSON printed by Cell 13.

Interpretation rule:
- GO: AE contributes calibrated, endpoint-robust safety information beyond pressure.
- CONDITIONAL GO: the dynamic-risk target is promising but needs stronger UQ/robustness.
- PIVOT: report non-identifiability or use terminal change detection; do not return to point RUL.
"""
)
```

```python
"""
PRISM-Hazard — Stage 2
======================

Pressure-conditioned, sensor-invariant, calibrated near-term failure-risk
forecasting for pressurized pipelines.

This is the scientifically motivated pivot after Stage 1 showed that point-RUL
regression was dominated by elapsed time/pressure, while acoustic emission (AE)
showed promising failure-within-10-minute Brier performance.

Run as a VS Code/Jupyter ``# %%`` notebook. The implementation is deliberately
strict about held-out-pipe separation and about the exploratory provenance of
AE-derived endpoint and pressure-alignment labels.
"""


# %% CELL 1 — Configuration and reproducible environment

from __future__ import annotations

import json
import math
import os
import warnings
from dataclasses import asdict, dataclass, field
from pathlib import Path
from typing import Dict, Iterable, List, Mapping, Optional, Sequence, Tuple

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.special import expit, logit
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    log_loss,
    roc_auc_score,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler


SEED = 20260829
EPS = 1e-12
np.random.seed(SEED)

mpl.rcParams.update(
    {
        "figure.dpi": 120,
        "savefig.dpi": 300,
        "font.family": "DejaVu Sans",
        "font.size": 10.5,
        "axes.labelsize": 11,
        "axes.titlesize": 12,
        "axes.grid": True,
        "grid.alpha": 0.22,
        "legend.frameon": False,
        "savefig.bbox": "tight",
        "savefig.facecolor": "white",
    }
)


@dataclass
class HazardConfig:
    # Input produced by PRISM-AE Stage 1.
    stage1_dir: Path = Path(
        r"F:\Pipeline Prressure Work\Code\results_prism_stage1"
    )
    output_dir: Path = Path(
        r"F:\Pipeline Prressure Work\Code\results_prism_stage2_hazard"
    )

    pipes: Tuple[str, ...] = ("B", "C", "D", "E")
    channels: Tuple[int, ...] = (5, 6, 7, 8)
    bands_khz: Tuple[Tuple[float, float], ...] = (
        (20.0, 60.0),
        (60.0, 100.0),
        (100.0, 140.0),
        (140.0, 190.0),
        (190.0, 250.0),
        (250.0, 400.0),
    )
    lf_band_indices: Tuple[int, ...] = (1, 2, 3)
    hf_band_indices: Tuple[int, ...] = (4, 5)

    # Dynamic risk horizons. The 10-minute horizon is the primary endpoint;
    # the others test whether conclusions depend on one arbitrary horizon.
    horizons_min: Tuple[float, ...] = (2.0, 5.0, 10.0, 15.0)
    primary_horizon_min: float = 10.0

    # Half-minute sampling limits serial pseudo-replication while preserving
    # sufficient positive samples for the shortest horizon.
    sample_minutes: float = 0.5
    analysis_start_min: float = 10.0
    healthy_margin_min: float = 20.0
    baseline_minutes: float = 10.0
    rolling_windows_min: Tuple[float, ...] = (1.0, 5.0, 10.0)

    # Simple models are intentional: there are only four independent pipes.
    logistic_c_grid: Tuple[float, ...] = tuple(np.logspace(-3, 3, 7))
    pressure_model_max_iter: int = 180

    # Sequential-alarm design. Policy selection uses cross-fitted predictions
    # from the training pipes only.
    alarm_threshold_grid: Tuple[float, ...] = tuple(np.linspace(0.05, 0.95, 37))
    alarm_persistence_grid: Tuple[int, ...] = (1, 2, 3, 4, 5)
    target_false_alarm_events_per_hour: float = 0.5

    # Endpoint sensitivity cases derived by Stage 1. These are diagnostic
    # proxies, not physical rupture times.
    endpoint_scenarios: Mapping[str, str] = field(
        default_factory=lambda: {
            "shutdown_onset": "existing_all_rms_gt3",
            "last_all_active": "last_4plus_active",
            "last_two_active": "last_2plus_active",
        }
    )

    # Feature-domain stress experiments; not substitutes for waveform-level tests.
    stress_repeats: int = 8
    gain_log_sigma: float = 0.45
    noise_floor_fraction: float = 0.20

    # These flags are written to every decision artifact.
    labels_are_physical: bool = False
    pressure_alignment_is_external: bool = False
    random_state: int = SEED


CFG = HazardConfig()
CFG.output_dir.mkdir(parents=True, exist_ok=True)

PIPE_COLORS = {"B": "#4477AA", "C": "#EE6677", "D": "#228833", "E": "#CCBB44"}
METHOD_COLORS = {
    "Age only": "#777777",
    "Pressure only": "#AA4499",
    "Pressure + raw AE": "#44AA99",
    "PRISM-Hazard": "#332288",
}
METHOD_ORDER = ["Age only", "Pressure only", "Pressure + raw AE", "PRISM-Hazard"]

print("Stage-1 directory:", CFG.stage1_dir)
print("Stage-2 directory:", CFG.output_dir)
print("Risk horizons (min):", CFG.horizons_min)
print("Primary scientific endpoint: failure within", CFG.primary_horizon_min, "min")


# %% CELL 2 — Load Stage-1 outputs and verify scientific provenance

def read_stage1_time_table(cfg: HazardConfig = CFG) -> pd.DataFrame:
    parquet = cfg.stage1_dir / "03_prism_time_bins.parquet"
    csv_gz = cfg.stage1_dir / "03_prism_time_bins.csv.gz"
    if parquet.exists():
        df = pd.read_parquet(parquet)
        source = parquet
    elif csv_gz.exists():
        df = pd.read_csv(csv_gz)
        source = csv_gz
    else:
        raise FileNotFoundError(
            "Could not find 03_prism_time_bins.parquet or .csv.gz in "
            f"{cfg.stage1_dir}. Run PRISM-AE Stage 1 first."
        )

    required = {
        "pipe", "time_min", "true_rul_min", "rupture_min", "pressure_pd",
        "time_in_hold_min", "dose_1", "dose_2", "dose_4",
        "log_rms_med", "log_power_med", "pooled_hf_logratio",
    }
    missing = sorted(required - set(df.columns))
    if missing:
        raise ValueError(f"Stage-1 table is missing columns: {missing}")

    df = df[df["pipe"].isin(cfg.pipes)].copy()
    df["pipe"] = df["pipe"].astype(str)
    df = df.sort_values(["pipe", "time_min"]).reset_index(drop=True)
    if df.empty:
        raise ValueError("Stage-1 time table contains no requested pipes")
    print(f"Loaded {len(df):,} rows from {source}")
    return df


def read_endpoint_candidates(cfg: HazardConfig = CFG) -> pd.DataFrame:
    path = cfg.stage1_dir / "02_endpoint_candidates_NOT_ground_truth.csv"
    if not path.exists():
        raise FileNotFoundError(f"Missing endpoint-candidate file: {path}")
    table = pd.read_csv(path)
    if "pipe" not in table:
        raise ValueError(f"{path} must contain a pipe column")
    table["pipe"] = table["pipe"].astype(str)
    return table


def read_stage1_manifest(cfg: HazardConfig = CFG) -> Dict[str, object]:
    path = cfg.stage1_dir / "12_reproducibility_manifest.json"
    if not path.exists():
        warnings.warn("Stage-1 manifest not found; provenance will be marked unresolved")
        return {}
    with open(path, "r", encoding="utf-8") as fh:
        return json.load(fh)


TIME_RAW = read_stage1_time_table(CFG)
ENDPOINT_CANDIDATES = read_endpoint_candidates(CFG)
STAGE1_MANIFEST = read_stage1_manifest(CFG)

CFG.labels_are_physical = bool(
    STAGE1_MANIFEST.get("decision_gate", {}).get("labels_are_physical", False)
)
CFG.pressure_alignment_is_external = bool(
    STAGE1_MANIFEST.get("decision_gate", {}).get("pressure_alignment_is_external", False)
)

schema_report = (
    TIME_RAW.groupby("pipe")
    .agg(
        rows=("time_min", "size"),
        start_min=("time_min", "min"),
        available_end_min=("time_min", "max"),
        stage1_endpoint_min=("rupture_min", "max"),
        max_pressure_pd=("pressure_pd", "max"),
    )
    .reset_index()
)
schema_report.to_csv(CFG.output_dir / "01_stage2_input_audit.csv", index=False)
print(schema_report.round(3).to_string(index=False))
print("\nLabel provenance:", "PHYSICAL" if CFG.labels_are_physical else "EXPLORATORY PROXY")
print(
    "Pressure alignment:",
    "EXTERNAL" if CFG.pressure_alignment_is_external else "EXPLORATORY/AE-ALIGNED",
)


# %% CELL 3 — Endpoint scenarios and common-support risk labels

def endpoint_map_for_scenario(
    scenario: str,
    candidates: pd.DataFrame = ENDPOINT_CANDIDATES,
    cfg: HazardConfig = CFG,
) -> Dict[str, float]:
    if scenario not in cfg.endpoint_scenarios:
        raise KeyError(f"Unknown endpoint scenario: {scenario}")
    column = cfg.endpoint_scenarios[scenario]
    if column not in candidates.columns:
        raise KeyError(f"Endpoint candidate column not found: {column}")
    indexed = candidates.set_index("pipe")[column]
    out = {p: float(indexed.loc[p]) for p in cfg.pipes}
    if not all(np.isfinite(list(out.values()))):
        raise ValueError(f"Non-finite endpoint in scenario {scenario}: {out}")
    return out


def attach_scenario_labels(
    table: pd.DataFrame,
    scenario: str,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    endpoints = endpoint_map_for_scenario(scenario, cfg=cfg)
    pieces, audit = [], []
    for pipe in cfg.pipes:
        d = table[table["pipe"] == pipe].copy()
        endpoint = endpoints[pipe]
        available_end = float(d["time_min"].max())
        # Rows after a proposed endpoint are invalid; later proposed endpoints may
        # extend past Stage-1 support and are explicitly audited below.
        d = d[d["time_min"] <= endpoint].copy()
        d["endpoint_scenario"] = scenario
        d["endpoint_min"] = endpoint
        d["true_rul_min"] = np.maximum(endpoint - d["time_min"], 0.0)
        for horizon in cfg.horizons_min:
            tag = str(horizon).replace(".", "p")
            d[f"fail_within_{tag}m"] = (d["true_rul_min"] <= horizon).astype(int)
        pieces.append(d)
        audit.append(
            {
                "scenario": scenario,
                "pipe": pipe,
                "endpoint_min": endpoint,
                "available_end_min": available_end,
                "unsupported_tail_min": max(endpoint - available_end, 0.0),
                "rows": len(d),
            }
        )
    return pd.concat(pieces, ignore_index=True), pd.DataFrame(audit)


def sample_timeline(table: pd.DataFrame, cfg: HazardConfig = CFG) -> pd.DataFrame:
    d = table[table["time_min"] >= cfg.analysis_start_min].copy()
    d["sample_bin"] = np.floor(d["time_min"] / cfg.sample_minutes).astype(int)
    d = (
        d.sort_values(["pipe", "time_min"])
        .groupby(["pipe", "sample_bin"], sort=False, as_index=False)
        .tail(1)
        .drop(columns="sample_bin")
        .reset_index(drop=True)
    )
    return d


PRIMARY_SCENARIO = "shutdown_onset"
PRIMARY_FULL, PRIMARY_SUPPORT_AUDIT = attach_scenario_labels(TIME_RAW, PRIMARY_SCENARIO, CFG)
PRIMARY = sample_timeline(PRIMARY_FULL, CFG)

support_tables = []
for scenario in CFG.endpoint_scenarios:
    _, a = attach_scenario_labels(TIME_RAW, scenario, CFG)
    support_tables.append(a)
SUPPORT_AUDIT = pd.concat(support_tables, ignore_index=True)
SUPPORT_AUDIT.to_csv(CFG.output_dir / "02_endpoint_scenario_support_audit.csv", index=False)

print(SUPPORT_AUDIT.round(3).to_string(index=False))
print("\nPrimary sampled rows per pipe:")
print(PRIMARY.groupby("pipe").size().to_string())


# %% CELL 4 — Causal AE dynamics and feature families

PRESSURE_FEATURES = [
    "pressure_pd",
    "delta_pressure_pd",
    "abs_delta_pressure_pd",
    "time_in_hold_min",
    "dose_1",
    "dose_2",
    "dose_4",
]

AE_FEATURES = [
    "log_rms_med",
    "log_rms_iqr",
    "log_power_med",
    "log_power_iqr",
    "hf_logratio_med",
    "hf_logratio_iqr",
    "spec_centroid_khz_med",
    "spec_centroid_khz_iqr",
    "sensor_share_entropy",
    "sensor_share_dominance",
    "active_sensor_fraction",
    "pooled_hf_logratio",
]


def robust_mad(x: np.ndarray, axis=None) -> np.ndarray:
    x = np.asarray(x, dtype=float)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", category=RuntimeWarning)
        med = np.nanmedian(x, axis=axis, keepdims=True)
        mad = np.nanmedian(np.abs(x - med), axis=axis)
    return 1.4826 * mad + EPS


def add_causal_dynamics(
    table: pd.DataFrame,
    features: Sequence[str],
    prefix: str,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, List[str]]:
    """Create only backward-looking changes, rolling dispersion, and baseline z scores."""
    pieces = []
    created: List[str] = []
    for pipe, src in table.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        dt = float(np.nanmedian(np.diff(d["time_min"]))) if len(d) > 1 else cfg.sample_minutes
        dt = max(dt, 1e-6)
        baseline_mask = d["time_min"] <= (
            float(d["time_min"].min()) + cfg.baseline_minutes
        )
        if baseline_mask.sum() < 5:
            baseline_mask.iloc[: min(10, len(d))] = True

        for feature in features:
            if feature not in d:
                continue
            x = d[feature].astype(float)
            base = x[baseline_mask].to_numpy(float)
            med = float(np.nanmedian(base))
            scale = float(robust_mad(base))

            current = f"{prefix}{feature}__current"
            basez = f"{prefix}{feature}__baseline_z"
            d[current] = x
            d[basez] = (x - med) / scale
            for name in (current, basez):
                if name not in created:
                    created.append(name)

            for window in cfg.rolling_windows_min:
                n = max(1, int(round(window / dt)))
                tag = str(window).replace(".", "p")
                delta = f"{prefix}{feature}__delta_{tag}m"
                spread = f"{prefix}{feature}__std_{tag}m"
                d[delta] = x - x.shift(n)
                d[spread] = x.rolling(n, min_periods=max(2, n // 3)).std()
                for name in (delta, spread):
                    if name not in created:
                        created.append(name)
        pieces.append(d)
    return pd.concat(pieces, ignore_index=True), created


def balanced_pipe_weights(df: pd.DataFrame) -> np.ndarray:
    counts = df["pipe"].value_counts()
    w = df["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    return w / np.mean(w)


def pressure_feature_family() -> List[str]:
    return ["time_min"] + PRESSURE_FEATURES


# %% CELL 5 — Training-pipe-only healthy-envelope residualization

class HealthyPressureResidualizer:
    """
    Learn the expected AE response to load from non-terminal training-pipe data.

    Unlike the Stage-1 exploratory residualizer, the healthy envelope excludes
    the final ``healthy_margin_min`` of each training pipe. This prevents the
    pressure model from explaining away the damage signal it is meant to expose.
    """

    def __init__(self, cfg: HazardConfig = CFG):
        self.cfg = cfg
        self.models: Dict[str, HistGradientBoostingRegressor] = {}
        self.pressure_fill: Optional[pd.Series] = None

    def fit(self, train: pd.DataFrame) -> "HealthyPressureResidualizer":
        healthy = train[train["true_rul_min"] > self.cfg.healthy_margin_min].copy()
        if healthy.groupby("pipe").size().min() < 10:
            warnings.warn(
                "Healthy-envelope filter left too few rows; using all training rows."
            )
            healthy = train.copy()

        X = healthy[PRESSURE_FEATURES].replace([np.inf, -np.inf], np.nan)
        self.pressure_fill = X.median(numeric_only=True)
        X = X.fillna(self.pressure_fill)
        weights = balanced_pipe_weights(healthy)

        self.models = {}
        for feature in AE_FEATURES:
            y = healthy[feature].to_numpy(float)
            finite = np.isfinite(y)
            model = HistGradientBoostingRegressor(
                loss="absolute_error",
                learning_rate=0.045,
                max_iter=self.cfg.pressure_model_max_iter,
                max_leaf_nodes=7,
                min_samples_leaf=max(10, int(0.02 * finite.sum())),
                l2_regularization=3.0,
                random_state=self.cfg.random_state,
            )
            model.fit(X.loc[finite], y[finite], sample_weight=weights[finite])
            self.models[feature] = model
        return self

    def transform(self, table: pd.DataFrame) -> pd.DataFrame:
        if self.pressure_fill is None or not self.models:
            raise RuntimeError("Residualizer has not been fitted")
        out = table.copy()
        X = out[PRESSURE_FEATURES].replace([np.inf, -np.inf], np.nan)
        X = X.fillna(self.pressure_fill)
        for feature, model in self.models.items():
            expected = model.predict(X)
            out[f"expected_{feature}"] = expected
            out[f"resid_{feature}"] = out[feature].to_numpy(float) - expected
        return out


def prepare_feature_pair(
    train: pd.DataFrame,
    other: pd.DataFrame,
    method: str,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame, List[str], Optional[HealthyPressureResidualizer]]:
    pressure = pressure_feature_family()

    if method == "Age only":
        return train.copy(), other.copy(), ["time_min"], None
    if method == "Pressure only":
        return train.copy(), other.copy(), pressure, None

    if method == "Pressure + raw AE":
        tr, ae_cols = add_causal_dynamics(train, AE_FEATURES, "raw_", cfg)
        ot, _ = add_causal_dynamics(other, AE_FEATURES, "raw_", cfg)
        return tr, ot, pressure + ae_cols, None

    if method == "PRISM-Hazard":
        residualizer = HealthyPressureResidualizer(cfg).fit(train)
        tr0 = residualizer.transform(train)
        ot0 = residualizer.transform(other)
        residual_features = [f"resid_{f}" for f in AE_FEATURES]
        tr, resid_cols = add_causal_dynamics(tr0, residual_features, "", cfg)
        ot, _ = add_causal_dynamics(ot0, residual_features, "", cfg)
        return tr, ot, pressure + resid_cols, residualizer

    raise KeyError(f"Unknown method: {method}")


# %% CELL 6 — Nested held-out-pipe probability models and cross-fitting

def horizon_label_column(horizon_min: float) -> str:
    return f"fail_within_{str(horizon_min).replace('.', 'p')}m"


def make_probability_model(c_value: float, seed: int = SEED) -> Pipeline:
    return Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
            ("scale", StandardScaler()),
            (
                "logistic",
                LogisticRegression(
                    C=float(c_value),
                    solver="lbfgs",
                    max_iter=4000,
                    random_state=seed,
                ),
            ),
        ]
    )


def fit_probability_model(
    model: Pipeline,
    X: pd.DataFrame,
    y: np.ndarray,
    groups: np.ndarray,
) -> Pipeline:
    frame = pd.DataFrame({"pipe": groups})
    counts = frame["pipe"].value_counts()
    weights = frame["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    weights /= weights.mean()
    model.fit(X, y, logistic__sample_weight=weights)
    return model


def group_balanced_brier(
    y: np.ndarray, probability: np.ndarray, groups: np.ndarray
) -> float:
    scores = []
    for pipe in np.unique(groups):
        m = groups == pipe
        scores.append(float(np.mean((probability[m] - y[m]) ** 2)))
    return float(np.mean(scores))


def fit_platt_calibrator(
    scores: np.ndarray,
    y: np.ndarray,
    groups: np.ndarray,
) -> Optional[LogisticRegression]:
    finite = np.isfinite(scores) & np.isfinite(y)
    scores, y, groups = scores[finite], y[finite].astype(int), groups[finite]
    if len(np.unique(y)) < 2:
        return None
    frame = pd.DataFrame({"pipe": groups})
    counts = frame["pipe"].value_counts()
    weights = frame["pipe"].map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    weights /= weights.mean()
    calibrator = LogisticRegression(
        C=100.0, solver="lbfgs", max_iter=3000, random_state=SEED
    )
    calibrator.fit(scores.reshape(-1, 1), y, sample_weight=weights)
    return calibrator


def apply_calibrator(
    calibrator: Optional[LogisticRegression], scores: np.ndarray
) -> np.ndarray:
    if calibrator is None:
        return expit(scores)
    return calibrator.predict_proba(scores.reshape(-1, 1))[:, 1]


def prepare_inner_splits(
    outer_train: pd.DataFrame,
    method: str,
    cfg: HazardConfig = CFG,
) -> List[Dict[str, object]]:
    prepared = []
    for held_pipe in sorted(outer_train["pipe"].unique()):
        tr0 = outer_train[outer_train["pipe"] != held_pipe].copy()
        va0 = outer_train[outer_train["pipe"] == held_pipe].copy()
        tr, va, features, _ = prepare_feature_pair(tr0, va0, method, cfg)
        prepared.append(
            {
                "held_pipe": held_pipe,
                "train": tr,
                "valid": va,
                "features": features,
            }
        )
    return prepared


def crossfit_for_c(
    prepared: Sequence[Mapping[str, object]],
    c_value: float,
    label_col: str,
) -> pd.DataFrame:
    rows = []
    for split in prepared:
        tr = split["train"]
        va = split["valid"]
        features = split["features"]
        model = make_probability_model(c_value)
        fit_probability_model(
            model,
            tr[features],
            tr[label_col].to_numpy(int),
            tr["pipe"].to_numpy(),
        )
        score = model.decision_function(va[features])
        block = va[
            ["pipe", "time_min", "true_rul_min", "endpoint_min", label_col]
        ].copy()
        block = block.rename(columns={label_col: "event"})
        block["raw_score"] = score
        block["raw_probability"] = expit(score)
        rows.append(block)
    return pd.concat(rows, ignore_index=True)


FEATURE_PREPARATION_CACHE: Dict[Tuple[object, ...], Dict[str, object]] = {}


def scenario_table_signature(table: pd.DataFrame) -> Tuple[object, ...]:
    endpoints = (
        table.groupby("pipe")["endpoint_min"].max().reindex(CFG.pipes).round(6).tolist()
    )
    return tuple(endpoints) + (len(table),)


def cross_calibrated_oof(oof: pd.DataFrame) -> Tuple[pd.DataFrame, Optional[LogisticRegression]]:
    """Cross-calibrate each training pipe without using that pipe in its calibrator."""
    out = oof.copy()
    out["probability"] = np.nan
    for held in out["pipe"].unique():
        tr = out[out["pipe"] != held]
        va = out[out["pipe"] == held]
        calibrator = fit_platt_calibrator(
            tr["raw_score"].to_numpy(float),
            tr["event"].to_numpy(int),
            tr["pipe"].to_numpy(),
        )
        out.loc[va.index, "probability"] = apply_calibrator(
            calibrator, va["raw_score"].to_numpy(float)
        )

    final_calibrator = fit_platt_calibrator(
        out["raw_score"].to_numpy(float),
        out["event"].to_numpy(int),
        out["pipe"].to_numpy(),
    )
    return out, final_calibrator


def fit_outer_fold(
    scenario_table: pd.DataFrame,
    test_pipe: str,
    method: str,
    horizon_min: float,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, pd.DataFrame, Dict[str, object], pd.DataFrame]:
    label_col = horizon_label_column(horizon_min)
    outer_train = scenario_table[scenario_table["pipe"] != test_pipe].copy()
    outer_test = scenario_table[scenario_table["pipe"] == test_pipe].copy()
    cache_key = (scenario_table_signature(scenario_table), test_pipe, method)
    if cache_key not in FEATURE_PREPARATION_CACHE:
        prepared = prepare_inner_splits(outer_train, method, cfg)
        train_ready, test_ready, features, residualizer = prepare_feature_pair(
            outer_train, outer_test, method, cfg
        )
        FEATURE_PREPARATION_CACHE[cache_key] = {
            "prepared": prepared,
            "train_ready": train_ready,
            "test_ready": test_ready,
            "features": features,
            "residualizer": residualizer,
        }
    cached = FEATURE_PREPARATION_CACHE[cache_key]
    prepared = cached["prepared"]

    tuning_rows = []
    oof_by_c: Dict[float, pd.DataFrame] = {}
    for c_value in cfg.logistic_c_grid:
        oof = crossfit_for_c(prepared, c_value, label_col)
        score = group_balanced_brier(
            oof["event"].to_numpy(int),
            oof["raw_probability"].to_numpy(float),
            oof["pipe"].to_numpy(),
        )
        tuning_rows.append({"C": c_value, "group_Brier": score})
        oof_by_c[float(c_value)] = oof

    tuning = pd.DataFrame(tuning_rows).sort_values(["group_Brier", "C"])
    best_c = float(tuning.iloc[0]["C"])
    oof, final_calibrator = cross_calibrated_oof(oof_by_c[best_c])

    train_ready = cached["train_ready"]
    test_ready = cached["test_ready"]
    features = cached["features"]
    residualizer = cached["residualizer"]
    model = make_probability_model(best_c)
    fit_probability_model(
        model,
        train_ready[features],
        train_ready[label_col].to_numpy(int),
        train_ready["pipe"].to_numpy(),
    )
    test_score = model.decision_function(test_ready[features])
    test_probability = apply_calibrator(final_calibrator, test_score)

    pred = test_ready[
        ["pipe", "time_min", "true_rul_min", "endpoint_min", label_col]
    ].copy()
    pred = pred.rename(columns={label_col: "event"})
    pred["probability"] = np.clip(test_probability, 1e-6, 1 - 1e-6)
    pred["raw_score"] = test_score
    pred["method"] = method
    pred["horizon_min"] = horizon_min
    pred["outer_test_pipe"] = test_pipe
    pred["selected_C"] = best_c

    oof["method"] = method
    oof["horizon_min"] = horizon_min
    oof["outer_test_pipe"] = test_pipe
    oof["selected_C"] = best_c

    artifact = {
        "method": method,
        "horizon_min": horizon_min,
        "test_pipe": test_pipe,
        "selected_C": best_c,
        "features": features,
        "model": model,
        "calibrator": final_calibrator,
        "residualizer": residualizer,
    }
    tuning["method"] = method
    tuning["horizon_min"] = horizon_min
    tuning["outer_test_pipe"] = test_pipe
    return pred, oof, artifact, tuning


# %% CELL 7 — Execute strict LOPO dynamic-risk evaluation

PRIMARY_PREDICTIONS: List[pd.DataFrame] = []
PRIMARY_OOF: List[pd.DataFrame] = []
PRIMARY_TUNING: List[pd.DataFrame] = []
PRIMARY_ARTIFACTS: Dict[Tuple[float, str, str], Dict[str, object]] = {}

for horizon in CFG.horizons_min:
    for test_pipe in CFG.pipes:
        print(f"\n=== Horizon {horizon:g} min | held-out pipe {test_pipe} ===")
        for method in METHOD_ORDER:
            pred, oof, artifact, tuning = fit_outer_fold(
                PRIMARY, test_pipe, method, horizon, CFG
            )
            PRIMARY_PREDICTIONS.append(pred)
            PRIMARY_OOF.append(oof)
            PRIMARY_TUNING.append(tuning)
            PRIMARY_ARTIFACTS[(float(horizon), method, test_pipe)] = artifact
            brier = float(np.mean((pred["probability"] - pred["event"]) ** 2))
            print(f"{method:24s} C={artifact['selected_C']:<8g} Brier={brier:.5f}")

PREDICTIONS = pd.concat(PRIMARY_PREDICTIONS, ignore_index=True)
OOF_PREDICTIONS = pd.concat(PRIMARY_OOF, ignore_index=True)
TUNING = pd.concat(PRIMARY_TUNING, ignore_index=True)

PREDICTIONS.to_csv(CFG.output_dir / "03_lopo_hazard_predictions.csv", index=False)
OOF_PREDICTIONS.to_csv(CFG.output_dir / "03_crossfitted_training_predictions.csv", index=False)
TUNING.to_csv(CFG.output_dir / "03_nested_group_tuning.csv", index=False)


# %% CELL 8 — Discrimination, calibration, and incremental-risk metrics

def expected_calibration_error(
    y: np.ndarray, probability: np.ndarray, n_bins: int = 10
) -> float:
    y = np.asarray(y, dtype=float)
    p = np.asarray(probability, dtype=float)
    edges = np.linspace(0.0, 1.0, n_bins + 1)
    value = 0.0
    for lo, hi in zip(edges[:-1], edges[1:]):
        mask = (p >= lo) & (p < hi if hi < 1.0 else p <= hi)
        if mask.any():
            value += mask.mean() * abs(float(y[mask].mean() - p[mask].mean()))
    return float(value)


def safe_auc(y: np.ndarray, p: np.ndarray, kind: str) -> float:
    if len(np.unique(y)) < 2:
        return np.nan
    if kind == "roc":
        return float(roc_auc_score(y, p))
    return float(average_precision_score(y, p))


def calibration_intercept_slope(y: np.ndarray, p: np.ndarray) -> Tuple[float, float]:
    if len(np.unique(y)) < 2:
        return np.nan, np.nan
    z = logit(np.clip(p, 1e-5, 1 - 1e-5)).reshape(-1, 1)
    model = LogisticRegression(C=1e4, solver="lbfgs", max_iter=3000)
    model.fit(z, y)
    return float(model.intercept_[0]), float(model.coef_[0, 0])


def metric_rows(predictions: pd.DataFrame) -> pd.DataFrame:
    rows = []
    keys = ["horizon_min", "method", "pipe"]
    for (horizon, method, pipe), d in predictions.groupby(keys, sort=False):
        y = d["event"].to_numpy(int)
        p = np.clip(d["probability"].to_numpy(float), 1e-6, 1 - 1e-6)
        intercept, slope = calibration_intercept_slope(y, p)
        rows.append(
            {
                "horizon_min": float(horizon),
                "method": method,
                "pipe": pipe,
                "n": len(d),
                "event_prevalence": float(y.mean()),
                "Brier": float(np.mean((p - y) ** 2)),
                "log_loss": float(log_loss(y, p, labels=[0, 1])),
                "ROC_AUC": safe_auc(y, p, "roc"),
                "PR_AUC": safe_auc(y, p, "pr"),
                "ECE": expected_calibration_error(y, p),
                "calibration_intercept": intercept,
                "calibration_slope": slope,
            }
        )
    return pd.DataFrame(rows)


METRICS_BY_PIPE = metric_rows(PREDICTIONS)
METRICS_SUMMARY = (
    METRICS_BY_PIPE.groupby(["horizon_min", "method"], as_index=False)
    .agg(
        Brier_mean=("Brier", "mean"),
        Brier_std=("Brier", "std"),
        log_loss_mean=("log_loss", "mean"),
        ROC_AUC_mean=("ROC_AUC", "mean"),
        PR_AUC_mean=("PR_AUC", "mean"),
        ECE_mean=("ECE", "mean"),
        calibration_slope_mean=("calibration_slope", "mean"),
    )
)

pressure_brier = METRICS_BY_PIPE[
    METRICS_BY_PIPE["method"] == "Pressure only"
][["horizon_min", "pipe", "Brier"]].rename(columns={"Brier": "pressure_Brier"})
INCREMENTAL = METRICS_BY_PIPE.merge(pressure_brier, on=["horizon_min", "pipe"])
INCREMENTAL["Brier_skill_vs_pressure"] = 1.0 - (
    INCREMENTAL["Brier"] / (INCREMENTAL["pressure_Brier"] + EPS)
)

METRICS_BY_PIPE.to_csv(CFG.output_dir / "04_hazard_metrics_by_pipe.csv", index=False)
METRICS_SUMMARY.to_csv(CFG.output_dir / "04_hazard_metrics_summary.csv", index=False)
INCREMENTAL.to_csv(CFG.output_dir / "04_incremental_hazard_value.csv", index=False)

print("\nGroup-balanced hazard summary:")
print(METRICS_SUMMARY.round(4).to_string(index=False))
print("\nPositive Brier skill means improvement over pressure-only:")
print(
    INCREMENTAL[
        ["horizon_min", "method", "pipe", "Brier_skill_vs_pressure"]
    ].round(4).to_string(index=False)
)


# %% CELL 9 — Publication figures for horizon value and calibration

def plot_brier_by_horizon(summary: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    fig, ax = plt.subplots(figsize=(8.2, 4.8))
    for method in METHOD_ORDER:
        d = summary[summary["method"] == method].sort_values("horizon_min")
        ax.plot(
            d["horizon_min"], d["Brier_mean"], marker="o", lw=1.8,
            color=METHOD_COLORS[method], label=method
        )
        ax.fill_between(
            d["horizon_min"].to_numpy(float),
            np.maximum(d["Brier_mean"] - d["Brier_std"], 0).to_numpy(float),
            (d["Brier_mean"] + d["Brier_std"]).to_numpy(float),
            color=METHOD_COLORS[method], alpha=0.10,
        )
    ax.set_xlabel("Failure horizon (min)")
    ax.set_ylabel("Group-balanced Brier score ↓")
    ax.set_title("Does AE improve dynamic risk beyond the pressure clock?")
    ax.legend(ncol=2)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "05_brier_by_failure_horizon.png")
    plt.show()


def plot_primary_calibration(predictions: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    d0 = predictions[predictions["horizon_min"] == cfg.primary_horizon_min]
    fig, ax = plt.subplots(figsize=(6.4, 5.5))
    ax.plot([0, 1], [0, 1], "--", color="black", lw=1.2, label="Ideal")
    edges = np.linspace(0, 1, 9)
    for method in METHOD_ORDER:
        d = d0[d0["method"] == method]
        points = []
        for lo, hi in zip(edges[:-1], edges[1:]):
            m = (d["probability"] >= lo) & (
                d["probability"] < hi if hi < 1 else d["probability"] <= hi
            )
            if m.sum() >= 4:
                # Equal pipe weight inside each plotted bin.
                by_pipe = d[m].groupby("pipe").agg(
                    predicted=("probability", "mean"), observed=("event", "mean")
                )
                points.append((by_pipe["predicted"].mean(), by_pipe["observed"].mean()))
        if points:
            x, y = np.asarray(points).T
            ax.plot(x, y, "o-", lw=1.5, color=METHOD_COLORS[method], label=method)
    ax.set_xlim(-0.02, 1.02)
    ax.set_ylim(-0.02, 1.02)
    ax.set_xlabel("Predicted failure probability")
    ax.set_ylabel("Observed event frequency")
    ax.set_title(f"Held-out-pipe calibration: failure within {cfg.primary_horizon_min:g} min")
    ax.legend(fontsize=8)
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "06_primary_horizon_calibration.png")
    plt.show()


def plot_pipe_brier_skill(incremental: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    d = incremental[
        (incremental["horizon_min"] == cfg.primary_horizon_min)
        & incremental["method"].isin(["Pressure + raw AE", "PRISM-Hazard"])
    ]
    fig, ax = plt.subplots(figsize=(7.8, 4.6))
    x = np.arange(len(cfg.pipes))
    width = 0.36
    for j, method in enumerate(["Pressure + raw AE", "PRISM-Hazard"]):
        values = d[d["method"] == method].set_index("pipe").reindex(cfg.pipes)
        ax.bar(
            x + (j - 0.5) * width,
            values["Brier_skill_vs_pressure"],
            width,
            color=METHOD_COLORS[method],
            edgecolor="black",
            linewidth=0.6,
            label=method,
        )
    ax.axhline(0, color="black", lw=1)
    ax.set_xticks(x, [f"Pipe {p}" for p in cfg.pipes])
    ax.set_ylabel("Brier skill versus pressure-only ↑")
    ax.set_title(f"Incremental AE value at {cfg.primary_horizon_min:g}-minute horizon")
    ax.legend()
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "07_primary_horizon_incremental_value.png")
    plt.show()


plot_brier_by_horizon(METRICS_SUMMARY, CFG)
plot_primary_calibration(PREDICTIONS, CFG)
plot_pipe_brier_skill(INCREMENTAL, CFG)


# %% CELL 10 — Training-only sequential alarm policy

def persistent_alarm(probability: np.ndarray, threshold: float, persistence: int) -> np.ndarray:
    above = np.asarray(probability, dtype=float) >= float(threshold)
    if persistence <= 1:
        return above
    run = pd.Series(above.astype(int)).rolling(persistence, min_periods=persistence).sum()
    return (run.to_numpy(float) >= persistence)


def evaluate_alarm_policy(
    frame: pd.DataFrame,
    threshold: float,
    persistence: int,
    horizon_min: float,
    cfg: HazardConfig = CFG,
) -> Tuple[pd.DataFrame, Dict[str, float]]:
    pipe_rows = []
    for pipe, src in frame.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        alarm = persistent_alarm(d["probability"].to_numpy(float), threshold, persistence)
        starts = alarm & ~np.r_[False, alarm[:-1]]
        preterminal = d["true_rul_min"].to_numpy(float) > horizon_min
        terminal = ~preterminal
        false_events = int(np.sum(starts & preterminal))
        exposure_h = max(float(np.sum(preterminal) * cfg.sample_minutes / 60.0), EPS)
        timely = alarm & terminal
        detected = bool(timely.any())
        lead = float(np.max(d.loc[timely, "true_rul_min"])) if detected else 0.0
        pipe_rows.append(
            {
                "pipe": pipe,
                "detected": int(detected),
                "lead_time_min": lead,
                "false_alarm_events": false_events,
                "preterminal_exposure_h": exposure_h,
                "false_alarm_events_per_h": false_events / exposure_h,
            }
        )
    detail = pd.DataFrame(pipe_rows)
    aggregate = {
        "detection_rate": float(detail["detected"].mean()),
        "mean_lead_time_min": float(detail.loc[detail["detected"] == 1, "lead_time_min"].mean())
        if detail["detected"].any() else 0.0,
        "false_alarm_events_per_h": float(
            detail["false_alarm_events"].sum()
            / max(detail["preterminal_exposure_h"].sum(), EPS)
        ),
    }
    return detail, aggregate


def select_alarm_policy(
    training_oof: pd.DataFrame,
    horizon_min: float,
    cfg: HazardConfig = CFG,
) -> Tuple[float, int, pd.DataFrame]:
    rows = []
    for threshold in cfg.alarm_threshold_grid:
        for persistence in cfg.alarm_persistence_grid:
            _, result = evaluate_alarm_policy(
                training_oof, threshold, persistence, horizon_min, cfg
            )
            rows.append(
                {
                    "threshold": threshold,
                    "persistence_bins": persistence,
                    **result,
                }
            )
    search = pd.DataFrame(rows)
    feasible = search[
        search["false_alarm_events_per_h"] <= cfg.target_false_alarm_events_per_hour
    ]
    if not feasible.empty:
        selected = feasible.sort_values(
            ["detection_rate", "mean_lead_time_min", "false_alarm_events_per_h", "threshold"],
            ascending=[False, False, True, False],
        ).iloc[0]
    else:
        selected = search.sort_values(
            ["false_alarm_events_per_h", "detection_rate", "mean_lead_time_min"],
            ascending=[True, False, False],
        ).iloc[0]
    return float(selected["threshold"]), int(selected["persistence_bins"]), search


ALARM_POLICIES = []
ALARM_BY_PIPE = []
ALARM_SEARCH = []
horizon = CFG.primary_horizon_min

for test_pipe in CFG.pipes:
    for method in METHOD_ORDER:
        train_oof = OOF_PREDICTIONS[
            (OOF_PREDICTIONS["outer_test_pipe"] == test_pipe)
            & (OOF_PREDICTIONS["horizon_min"] == horizon)
            & (OOF_PREDICTIONS["method"] == method)
        ].copy()
        test = PREDICTIONS[
            (PREDICTIONS["outer_test_pipe"] == test_pipe)
            & (PREDICTIONS["horizon_min"] == horizon)
            & (PREDICTIONS["method"] == method)
        ].copy()
        threshold, persistence, search = select_alarm_policy(train_oof, horizon, CFG)
        detail, result = evaluate_alarm_policy(test, threshold, persistence, horizon, CFG)
        ALARM_POLICIES.append(
            {
                "outer_test_pipe": test_pipe,
                "method": method,
                "horizon_min": horizon,
                "threshold": threshold,
                "persistence_bins": persistence,
                **result,
            }
        )
        detail["outer_test_pipe"] = test_pipe
        detail["method"] = method
        detail["horizon_min"] = horizon
        detail["threshold"] = threshold
        detail["persistence_bins"] = persistence
        ALARM_BY_PIPE.append(detail)
        search["outer_test_pipe"] = test_pipe
        search["method"] = method
        search["horizon_min"] = horizon
        ALARM_SEARCH.append(search)

ALARM_POLICIES = pd.DataFrame(ALARM_POLICIES)
ALARM_BY_PIPE = pd.concat(ALARM_BY_PIPE, ignore_index=True)
ALARM_SEARCH = pd.concat(ALARM_SEARCH, ignore_index=True)
ALARM_POLICIES.to_csv(CFG.output_dir / "08_alarm_policy_lopo.csv", index=False)
ALARM_BY_PIPE.to_csv(CFG.output_dir / "08_alarm_results_by_pipe.csv", index=False)
ALARM_SEARCH.to_csv(CFG.output_dir / "08_alarm_policy_training_search.csv", index=False)

print("\nHeld-out-pipe sequential alarm results:")
print(ALARM_POLICIES.round(4).to_string(index=False))


def plot_primary_risk_trajectories(
    predictions: pd.DataFrame,
    policies: pd.DataFrame,
    cfg: HazardConfig = CFG,
) -> None:
    fig, axes = plt.subplots(len(cfg.pipes), 1, figsize=(12.5, 10.5), sharex=False)
    for ax, pipe in zip(axes, cfg.pipes):
        for method in ["Pressure only", "Pressure + raw AE", "PRISM-Hazard"]:
            d = predictions[
                (predictions["pipe"] == pipe)
                & (predictions["horizon_min"] == cfg.primary_horizon_min)
                & (predictions["method"] == method)
            ].sort_values("time_min")
            ax.plot(
                d["time_min"], d["probability"], lw=1.25,
                color=METHOD_COLORS[method], label=method
            )
        policy = policies[
            (policies["outer_test_pipe"] == pipe)
            & (policies["method"] == "PRISM-Hazard")
        ].iloc[0]
        endpoint = float(d["endpoint_min"].iloc[0])
        ax.axvspan(
            endpoint - cfg.primary_horizon_min, endpoint,
            color="#CC3311", alpha=0.10, label="True failure horizon"
        )
        ax.axhline(
            policy["threshold"], color="#CC3311", ls="--", lw=1.0,
            label=f"PRISM alarm threshold ({int(policy['persistence_bins'])} bins)"
        )
        ax.set_ylim(-0.02, 1.02)
        ax.set_ylabel(f"Pipe {pipe}\nRisk")
        if pipe == cfg.pipes[0]:
            ax.legend(ncol=3, fontsize=8, loc="upper left")
    axes[-1].set_xlabel("Time from feature start (min)")
    fig.suptitle(
        f"Strict LOPO failure-within-{cfg.primary_horizon_min:g}-minute risk",
        y=1.005,
    )
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "09_lopo_dynamic_risk_trajectories.png")
    plt.show()


plot_primary_risk_trajectories(PREDICTIONS, ALARM_POLICIES, CFG)


# %% CELL 11 — Endpoint-label sensitivity under common observed support

SENSITIVITY_ROWS = []
SENSITIVITY_METHODS = ["Pressure only", "Pressure + raw AE", "PRISM-Hazard"]

for scenario in CFG.endpoint_scenarios:
    scenario_full, _ = attach_scenario_labels(TIME_RAW, scenario, CFG)
    scenario_table = sample_timeline(scenario_full, CFG)
    print(f"\n=== Endpoint sensitivity: {scenario} ===")
    for test_pipe in CFG.pipes:
        for method in SENSITIVITY_METHODS:
            pred, _, _, _ = fit_outer_fold(
                scenario_table,
                test_pipe,
                method,
                CFG.primary_horizon_min,
                CFG,
            )
            y = pred["event"].to_numpy(int)
            p = pred["probability"].to_numpy(float)
            SENSITIVITY_ROWS.append(
                {
                    "scenario": scenario,
                    "pipe": test_pipe,
                    "method": method,
                    "horizon_min": CFG.primary_horizon_min,
                    "Brier": float(np.mean((p - y) ** 2)),
                    "PR_AUC": safe_auc(y, p, "pr"),
                    "event_prevalence": float(y.mean()),
                }
            )

ENDPOINT_SENSITIVITY = pd.DataFrame(SENSITIVITY_ROWS)
pressure_reference = ENDPOINT_SENSITIVITY[
    ENDPOINT_SENSITIVITY["method"] == "Pressure only"
][["scenario", "pipe", "Brier"]].rename(columns={"Brier": "pressure_Brier"})
ENDPOINT_SENSITIVITY = ENDPOINT_SENSITIVITY.merge(
    pressure_reference, on=["scenario", "pipe"], how="left"
)
ENDPOINT_SENSITIVITY["Brier_skill_vs_pressure"] = 1.0 - (
    ENDPOINT_SENSITIVITY["Brier"] / (ENDPOINT_SENSITIVITY["pressure_Brier"] + EPS)
)
ENDPOINT_SENSITIVITY.to_csv(
    CFG.output_dir / "10_endpoint_sensitivity_metrics.csv", index=False
)

ENDPOINT_SENSITIVITY_SUMMARY = (
    ENDPOINT_SENSITIVITY.groupby(["scenario", "method"], as_index=False)
    .agg(
        Brier_mean=("Brier", "mean"),
        PR_AUC_mean=("PR_AUC", "mean"),
        Brier_skill_mean=("Brier_skill_vs_pressure", "mean"),
        pipes_with_positive_skill=("Brier_skill_vs_pressure", lambda x: int((x > 0).sum())),
    )
)
ENDPOINT_SENSITIVITY_SUMMARY.to_csv(
    CFG.output_dir / "10_endpoint_sensitivity_summary.csv", index=False
)
print(ENDPOINT_SENSITIVITY_SUMMARY.round(4).to_string(index=False))


def plot_endpoint_sensitivity(
    sensitivity: pd.DataFrame, cfg: HazardConfig = CFG
) -> None:
    methods = ["Pressure + raw AE", "PRISM-Hazard"]
    scenarios = list(cfg.endpoint_scenarios)
    matrix = np.full((len(scenarios), len(methods)), np.nan)
    for i, scenario in enumerate(scenarios):
        for j, method in enumerate(methods):
            d = sensitivity[
                (sensitivity["scenario"] == scenario)
                & (sensitivity["method"] == method)
            ]
            matrix[i, j] = d["Brier_skill_vs_pressure"].mean()

    bound = max(0.25, float(np.nanmax(np.abs(matrix))))
    fig, ax = plt.subplots(figsize=(7.4, 4.2))
    im = ax.imshow(matrix, cmap="RdBu_r", vmin=-bound, vmax=bound, aspect="auto")
    ax.set_xticks(range(len(methods)), methods)
    ax.set_yticks(range(len(scenarios)), scenarios)
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            ax.text(j, i, f"{matrix[i, j]:+.2f}", ha="center", va="center")
    ax.set_title("Endpoint-robust Brier skill versus pressure-only")
    fig.colorbar(im, ax=ax, label="Brier skill ↑")
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "10_endpoint_sensitivity_heatmap.png")
    plt.show()


plot_endpoint_sensitivity(ENDPOINT_SENSITIVITY, CFG)


# %% CELL 12 — Sensor dropout, gain drift, and energy-floor stress experiments

def rowwise_sensor_statistics(values: np.ndarray, prefix: str) -> Dict[str, np.ndarray]:
    return {
        f"{prefix}_med": np.nanmedian(values, axis=1),
        f"{prefix}_mean": np.nanmean(values, axis=1),
        f"{prefix}_std": np.nanstd(values, axis=1),
        f"{prefix}_iqr": np.nanpercentile(values, 75, axis=1)
        - np.nanpercentile(values, 25, axis=1),
        f"{prefix}_max": np.nanmax(values, axis=1),
    }


def normalized_sensor_entropy(shares: np.ndarray) -> np.ndarray:
    p = np.asarray(shares, dtype=float)
    valid = np.isfinite(p) & (p > 0)
    entropy = -np.nansum(np.where(valid, p * np.log(p + EPS), 0.0), axis=1)
    count = valid.sum(axis=1)
    denom = np.log(np.maximum(count, 2))
    return entropy / denom


def recompute_ae_aggregates(
    table: pd.DataFrame,
    rms: np.ndarray,
    energy: np.ndarray,
    cfg: HazardConfig = CFG,
) -> pd.DataFrame:
    out = table.copy()
    widths = np.array([hi - lo for lo, hi in cfg.bands_khz], dtype=float)
    centres = np.array([(lo + hi) / 2.0 for lo, hi in cfg.bands_khz], dtype=float)
    lf = np.asarray(cfg.lf_band_indices, dtype=int)
    hf = np.asarray(cfg.hf_band_indices, dtype=int)
    selected = np.r_[lf, hf]

    lf_density = np.nansum(energy[:, :, lf], axis=2) / widths[lf].sum()
    hf_density = np.nansum(energy[:, :, hf], axis=2) / widths[hf].sum()
    total = np.nansum(energy[:, :, selected], axis=2)
    centroid = np.nansum(
        energy[:, :, selected] * centres[selected][None, None, :], axis=2
    ) / (total + EPS)

    log_rms = np.log1p(np.maximum(rms, 0.0))
    log_power = np.log1p(np.maximum(total, 0.0))
    hf_logratio = np.log((hf_density + EPS) / (lf_density + EPS))

    for name, values in rowwise_sensor_statistics(log_rms, "log_rms").items():
        out[name] = values
    for name, values in rowwise_sensor_statistics(log_power, "log_power").items():
        out[name] = values
    for name, values in rowwise_sensor_statistics(hf_logratio, "hf_logratio").items():
        out[name] = values
    for name, values in rowwise_sensor_statistics(centroid, "spec_centroid_khz").items():
        out[name] = values

    shares = total / (np.nansum(total, axis=1, keepdims=True) + EPS)
    out["sensor_share_entropy"] = normalized_sensor_entropy(shares)
    out["sensor_share_dominance"] = np.nanmax(shares, axis=1)

    active_fraction = np.zeros(len(out), dtype=float)
    for pipe in out["pipe"].unique():
        idx = np.flatnonzero(out["pipe"].to_numpy() == pipe)
        d = out.iloc[idx]
        baseline = d["time_min"].to_numpy(float) <= (
            float(d["time_min"].min()) + cfg.baseline_minutes
        )
        if baseline.sum() < 5:
            baseline[: min(10, len(baseline))] = True
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            threshold = np.nanmedian(log_rms[idx][baseline], axis=0) + (
                6.0 * robust_mad(log_rms[idx][baseline], axis=0)
            )
        active = log_rms[idx] > threshold[None, :]
        valid = np.isfinite(log_rms[idx])
        active_fraction[idx] = np.sum(active & valid, axis=1) / np.maximum(valid.sum(axis=1), 1)
    out["active_sensor_fraction"] = active_fraction

    pooled_lf = np.nansum(energy[:, :, lf], axis=(1, 2)) / widths[lf].sum()
    pooled_hf = np.nansum(energy[:, :, hf], axis=(1, 2)) / widths[hf].sum()
    out["pooled_hf_logratio"] = np.log((pooled_hf + EPS) / (pooled_lf + EPS))
    return out


def stress_sensor_table(
    table: pd.DataFrame,
    scenario: str,
    repeat: int,
    cfg: HazardConfig = CFG,
) -> pd.DataFrame:
    scenario_seed = {
        "clean": 0,
        "drop_one_sensor": 101,
        "gain_drift": 211,
        "energy_floor": 307,
        "combined": 401,
    }[scenario]
    rng = np.random.default_rng(cfg.random_state + 1009 * repeat + scenario_seed)
    rms = np.column_stack(
        [table[f"rms_ch{ch}"].to_numpy(float) for ch in cfg.channels]
    )
    energy = np.empty((len(table), len(cfg.channels), len(cfg.bands_khz)), dtype=float)
    for j, ch in enumerate(cfg.channels):
        for b in range(len(cfg.bands_khz)):
            energy[:, j, b] = table[f"e_ch{ch}_b{b}"].to_numpy(float)

    if scenario in {"drop_one_sensor", "combined"}:
        j = repeat % len(cfg.channels)
        rms[:, j] = np.nan
        energy[:, j, :] = np.nan

    if scenario in {"gain_drift", "combined"}:
        gain = rng.lognormal(mean=0.0, sigma=cfg.gain_log_sigma, size=len(cfg.channels))
        rms *= gain[None, :]
        energy *= gain[None, :, None] ** 2

    if scenario in {"energy_floor", "combined"}:
        baseline_n = max(5, int(round(cfg.baseline_minutes / cfg.sample_minutes)))
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            floor = np.nanmedian(energy[:baseline_n], axis=0)
        random_floor = rng.uniform(0.5, 1.5, size=energy.shape)
        energy += cfg.noise_floor_fraction * floor[None, :, :] * random_floor
        with warnings.catch_warnings():
            warnings.simplefilter("ignore", category=RuntimeWarning)
            rms_floor = np.nanmedian(rms[:baseline_n], axis=0)
        rms = np.sqrt(
            np.maximum(rms, 0.0) ** 2
            + (cfg.noise_floor_fraction * rms_floor[None, :]) ** 2
        )

    return recompute_ae_aggregates(table, rms, energy, cfg)


def predict_with_artifact(
    artifact: Mapping[str, object],
    test_table: pd.DataFrame,
    cfg: HazardConfig = CFG,
) -> np.ndarray:
    method = artifact["method"]
    features = artifact["features"]
    if method == "PRISM-Hazard":
        residualizer = artifact["residualizer"]
        transformed = residualizer.transform(test_table)
        residual_features = [f"resid_{f}" for f in AE_FEATURES]
        ready, _ = add_causal_dynamics(transformed, residual_features, "", cfg)
    elif method == "Pressure + raw AE":
        ready, _ = add_causal_dynamics(test_table, AE_FEATURES, "raw_", cfg)
    else:
        ready = test_table.copy()
    score = artifact["model"].decision_function(ready[features])
    return np.clip(apply_calibrator(artifact["calibrator"], score), 1e-6, 1 - 1e-6)


STRESS_SCENARIOS = ["clean", "drop_one_sensor", "gain_drift", "energy_floor", "combined"]
STRESS_ROWS = []

for pipe in CFG.pipes:
    clean_test = PRIMARY[PRIMARY["pipe"] == pipe].copy()
    for method in ["Pressure + raw AE", "PRISM-Hazard"]:
        artifact = PRIMARY_ARTIFACTS[(CFG.primary_horizon_min, method, pipe)]
        repeats = {"clean": 1, **{s: CFG.stress_repeats for s in STRESS_SCENARIOS if s != "clean"}}
        for scenario, n_repeat in repeats.items():
            for repeat in range(n_repeat):
                stressed = (
                    clean_test.copy()
                    if scenario == "clean"
                    else stress_sensor_table(clean_test, scenario, repeat, CFG)
                )
                probability = predict_with_artifact(artifact, stressed, CFG)
                y = stressed[horizon_label_column(CFG.primary_horizon_min)].to_numpy(int)
                STRESS_ROWS.append(
                    {
                        "pipe": pipe,
                        "method": method,
                        "scenario": scenario,
                        "repeat": repeat,
                        "Brier": float(np.mean((probability - y) ** 2)),
                        "PR_AUC": safe_auc(y, probability, "pr"),
                        "mean_probability": float(np.mean(probability)),
                    }
                )

STRESS_RESULTS = pd.DataFrame(STRESS_ROWS)
clean_reference = STRESS_RESULTS[STRESS_RESULTS["scenario"] == "clean"][
    ["pipe", "method", "Brier"]
].rename(columns={"Brier": "clean_Brier"})
STRESS_RESULTS = STRESS_RESULTS.merge(clean_reference, on=["pipe", "method"], how="left")
STRESS_RESULTS["relative_Brier_degradation"] = (
    STRESS_RESULTS["Brier"] - STRESS_RESULTS["clean_Brier"]
) / (STRESS_RESULTS["clean_Brier"] + EPS)
STRESS_RESULTS.to_csv(CFG.output_dir / "11_sensor_stress_results.csv", index=False)

STRESS_SUMMARY = (
    STRESS_RESULTS.groupby(["method", "scenario"], as_index=False)
    .agg(
        Brier_mean=("Brier", "mean"),
        PR_AUC_mean=("PR_AUC", "mean"),
        relative_Brier_degradation_mean=("relative_Brier_degradation", "mean"),
        relative_Brier_degradation_max=("relative_Brier_degradation", "max"),
    )
)
STRESS_SUMMARY.to_csv(CFG.output_dir / "11_sensor_stress_summary.csv", index=False)
print(STRESS_SUMMARY.round(4).to_string(index=False))


def plot_sensor_stress(results: pd.DataFrame, cfg: HazardConfig = CFG) -> None:
    fig, ax = plt.subplots(figsize=(9.0, 4.8))
    scenarios = STRESS_SCENARIOS[1:]
    x = np.arange(len(scenarios))
    width = 0.36
    for j, method in enumerate(["Pressure + raw AE", "PRISM-Hazard"]):
        means, errors = [], []
        for scenario in scenarios:
            d = results[(results["method"] == method) & (results["scenario"] == scenario)]
            means.append(d["relative_Brier_degradation"].mean())
            errors.append(d["relative_Brier_degradation"].std())
        ax.bar(
            x + (j - 0.5) * width, means, width, yerr=errors,
            color=METHOD_COLORS[method], edgecolor="black", linewidth=0.6,
            capsize=3, label=method,
        )
    ax.axhline(0, color="black", lw=1)
    ax.set_xticks(x, [s.replace("_", "\n") for s in scenarios])
    ax.set_ylabel("Relative Brier degradation ↓")
    ax.set_title("Held-out-pipe sensor-stress robustness")
    ax.legend()
    fig.tight_layout()
    fig.savefig(cfg.output_dir / "11_sensor_stress_robustness.png")
    plt.show()


plot_sensor_stress(STRESS_RESULTS, CFG)


# %% CELL 13 — Evidence-based Stage-2 decision gate

def stage2_decision_gate(cfg: HazardConfig = CFG) -> Dict[str, object]:
    h = cfg.primary_horizon_min
    primary = METRICS_BY_PIPE[METRICS_BY_PIPE["horizon_min"] == h]
    mean_brier = primary.groupby("method")["Brier"].mean()
    ae_candidates = ["Pressure + raw AE", "PRISM-Hazard"]
    selected_method = min(ae_candidates, key=lambda m: float(mean_brier.loc[m]))

    comparison = primary[primary["method"].isin(["Pressure only", selected_method])]
    wide = comparison.pivot(index="pipe", columns="method", values="Brier")
    pipe_skill = 1.0 - wide[selected_method] / (wide["Pressure only"] + EPS)
    mean_skill = float(pipe_skill.mean())
    wins = int((pipe_skill > 0).sum())

    ece = float(primary[primary["method"] == selected_method]["ECE"].mean())
    alarm = ALARM_POLICIES[ALARM_POLICIES["method"] == selected_method]
    detection_rate = float(alarm["detection_rate"].mean())
    false_alarm_rate = float(alarm["false_alarm_events_per_h"].mean())
    lead_time = float(alarm["mean_lead_time_min"].mean())

    sensitivity = ENDPOINT_SENSITIVITY[
        ENDPOINT_SENSITIVITY["method"] == selected_method
    ]
    scenario_skill = sensitivity.groupby("scenario")["Brier_skill_vs_pressure"].mean()
    worst_endpoint_skill = float(scenario_skill.min())

    stress = STRESS_RESULTS[STRESS_RESULTS["method"] == selected_method]
    combined = stress[stress["scenario"] == "combined"]
    combined_degradation = float(combined["relative_Brier_degradation"].mean())

    if (
        mean_skill >= 0.15
        and wins >= 3
        and ece <= 0.10
        and detection_rate >= 0.75
        and false_alarm_rate <= cfg.target_false_alarm_events_per_hour
        and worst_endpoint_skill > 0.0
    ):
        verdict = "GO — DYNAMIC HAZARD"
        next_step = (
            "Proceed to Bayesian/conformal uncertainty decomposition and decision-cost analysis."
        )
    elif (
        mean_skill > 0.0
        and wins >= 2
        and detection_rate >= 0.50
    ):
        verdict = "CONDITIONAL GO — HAZARD, NOT POINT RUL"
        next_step = (
            "Retain dynamic hazard as the paper target; improve calibration and endpoint/"
            "sensor robustness before adding a final uncertainty layer."
        )
    else:
        verdict = "PIVOT — CHANGE DETECTION / DATA LIMITATION STUDY"
        next_step = (
            "Do not claim transferable failure probability. Reframe as pressure-conditioned "
            "terminal-regime detection or a rigorous non-identifiability benchmark."
        )

    if not cfg.labels_are_physical or not cfg.pressure_alignment_is_external:
        verdict = "EXPLORATORY " + verdict
        next_step += (
            " External rupture timestamps and pressure alignment remain mandatory for a "
            "publication-grade reliability claim."
        )

    return {
        "verdict": verdict,
        "selected_AE_method": selected_method,
        "primary_horizon_min": h,
        "pressure_Brier_mean": float(mean_brier.loc["Pressure only"]),
        "selected_AE_Brier_mean": float(mean_brier.loc[selected_method]),
        "Brier_skill_vs_pressure_mean": mean_skill,
        "pipes_won_out_of_4": wins,
        "ECE_mean": ece,
        "alarm_detection_rate": detection_rate,
        "alarm_false_events_per_hour": false_alarm_rate,
        "alarm_mean_lead_time_min": lead_time,
        "worst_endpoint_scenario_skill": worst_endpoint_skill,
        "combined_sensor_stress_relative_Brier_degradation": combined_degradation,
        "labels_are_physical": cfg.labels_are_physical,
        "pressure_alignment_is_external": cfg.pressure_alignment_is_external,
        "next_step": next_step,
    }


DECISION = stage2_decision_gate(CFG)
print("\n" + json.dumps(DECISION, indent=2))


def json_safe(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, dict):
        return {str(k): json_safe(v) for k, v in value.items()}
    if isinstance(value, (tuple, list)):
        return [json_safe(v) for v in value]
    if isinstance(value, np.generic):
        return value.item()
    return value


manifest = {
    "project": "PRISM-Hazard Stage 2",
    "scientific_target": "calibrated failure-within-horizon probability",
    "primary_endpoint_scenario": PRIMARY_SCENARIO,
    "endpoint_scenarios": dict(CFG.endpoint_scenarios),
    "config": json_safe(asdict(CFG)),
    "stage1_manifest": STAGE1_MANIFEST,
    "decision": DECISION,
}
with open(CFG.output_dir / "12_stage2_reproducibility_manifest.json", "w", encoding="utf-8") as fh:
    json.dump(manifest, fh, indent=2)


# %% CELL 14 — Result handoff for the next research decision

print(
    """
PRISM-HAZARD STAGE 2 COMPLETE
-----------------------------
Send these files/results for interpretation:

1. 04_hazard_metrics_summary.csv
2. 04_incremental_hazard_value.csv
3. 05_brier_by_failure_horizon.png
4. 06_primary_horizon_calibration.png
5. 07_primary_horizon_incremental_value.png
6. 08_alarm_policy_lopo.csv
7. 09_lopo_dynamic_risk_trajectories.png
8. 10_endpoint_sensitivity_summary.csv
9. 10_endpoint_sensitivity_heatmap.png
10. 11_sensor_stress_summary.csv
11. 11_sensor_stress_robustness.png
12. The decision JSON printed by Cell 13.

Interpretation rule:
- GO: AE contributes calibrated, endpoint-robust safety information beyond pressure.
- CONDITIONAL GO: the dynamic-risk target is promising but needs stronger UQ/robustness.
- PIVOT: report non-identifiability or use terminal change detection; do not return to point RUL.
"""
)
```

```python
"""
CLEAR-AE — Stage 3 identifiability and terminal-regime audit
============================================================

Confounding, Leakage, Endpoint-Ambiguity, and Robustness audit for acoustic
emission (AE) pipeline prognostics.

This notebook-style script is the scientifically necessary pivot after strict
leave-one-pipe-out Stage 1/2 experiments showed that:

* elapsed time / nominal pressure already predicts the exploratory endpoint;
* AE does not improve 10-minute risk or point RUL across held-out pipes;
* raw AE has a small, intriguing 2-minute terminal-regime signal on 3/4 pipes;
* the endpoint and pressure alignment are not external physical measurements.

The goal is not to rescue a positive RUL result.  The goal is to determine what
the data can and cannot identify.  CLEAR-AE implements five falsification gates:

1. clock-dominance and exact pipe-level inference;
2. coherent endpoint-interval (partial-identification) bounds;
3. sensor/pipe fingerprint quantification;
4. pressure-preserving AE negative controls;
5. held-out-pipe, label-free terminal change detection.

It also includes simulations showing how the same audit behaves in clock-only,
transferable-damage, and pipe-specific-damage worlds.  Run each ``# %% CELL``
in order in VS Code or Jupyter.  Results are deliberately labelled exploratory
unless physical rupture timestamps and external pressure alignment are supplied.
"""


# %% CELL 1 — Reproducible configuration and publication style

from __future__ import annotations

import hashlib
import itertools
import json
import math
import os
import platform
import sys
import warnings
from dataclasses import asdict, dataclass, field
from pathlib import Path
from typing import Dict, Iterable, List, Mapping, Optional, Sequence, Tuple

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.special import expit
from sklearn.base import clone
from sklearn.compose import TransformedTargetRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    brier_score_loss,
    confusion_matrix,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler


SEED = 20260830
EPS = 1e-12
RNG = np.random.default_rng(SEED)

mpl.rcParams.update(
    {
        "figure.dpi": 125,
        "savefig.dpi": 320,
        "font.family": "DejaVu Sans",
        "font.size": 10.5,
        "axes.labelsize": 11,
        "axes.titlesize": 12,
        "axes.grid": True,
        "grid.alpha": 0.20,
        "legend.frameon": False,
        "savefig.bbox": "tight",
        "savefig.facecolor": "white",
    }
)
sns.set_palette("colorblind")


@dataclass
class ClearConfig:
    # Inputs created by PRISM-AE Stage 1 and PRISM-Hazard Stage 2.
    stage1_dir: Path = Path(r"F:\Pipeline Prressure Work\Code\results_prism_stage1")
    stage2_dir: Path = Path(
        r"F:\Pipeline Prressure Work\Code\results_prism_stage2_hazard"
    )
    output_dir: Path = Path(
        r"F:\Pipeline Prressure Work\Code\results_clear_ae_stage3"
    )

    pipes: Tuple[str, ...] = ("B", "C", "D", "E")
    horizons_min: Tuple[float, ...] = (2.0, 5.0, 10.0, 15.0)
    terminal_horizon_min: float = 2.0
    operational_horizon_min: float = 10.0
    sample_minutes: float = 0.5
    analysis_start_min: float = 10.0

    # The endpoint interval is [shutdown onset, last two sensors active].
    # Both bounds remain AE-derived inspection candidates, not rupture truth.
    endpoint_lower_col: str = "existing_all_rms_gt3"
    endpoint_upper_col: str = "last_2plus_active"

    # Transparent feature families.  Missing optional columns are skipped.
    pressure_features: Tuple[str, ...] = (
        "time_min",
        "pressure_pd",
        "delta_pressure_pd",
        "abs_delta_pressure_pd",
        "time_in_hold_min",
        "dose_1",
        "dose_2",
        "dose_4",
    )
    ae_features: Tuple[str, ...] = (
        "log_rms_med",
        "log_rms_iqr",
        "log_power_med",
        "log_power_iqr",
        "hf_logratio_med",
        "hf_logratio_iqr",
        "spec_centroid_khz_med",
        "spec_centroid_khz_iqr",
        "sensor_share_entropy",
        "sensor_share_dominance",
        "active_sensor_fraction",
        "pooled_hf_logratio",
    )
    dynamic_lags_min: Tuple[float, ...] = (0.5, 2.0, 5.0)

    # Negative controls preserve pressure/pipe structure but break AE timing.
    negative_control_repeats: int = 100
    negative_controls: Tuple[str, ...] = (
        "within_pressure_permutation",
        "within_pressure_circular_shift",
        "pipe_swap_by_life_fraction",
    )
    logistic_c: float = 1.0

    # Pipe-fingerprint audit.
    fingerprint_blocks: int = 6

    # Label-free terminal change detection.
    healthy_guard_min: float = 15.0
    causal_reference_min: float = 10.0
    change_persistence_bins: Tuple[int, ...] = (1, 2, 3, 4)
    change_quantiles: Tuple[float, ...] = (0.97, 0.98, 0.99, 0.995, 0.9975)
    target_false_events_per_h: float = 0.5

    # Synthetic framework validation.
    simulation_repeats: int = 40
    simulation_pipes: int = 24
    simulation_points_per_pipe: int = 181

    # Provenance gates.  Change these only after documentary verification.
    labels_are_physical: bool = False
    pressure_alignment_is_external: bool = False
    random_state: int = SEED


CFG = ClearConfig()
CFG.output_dir.mkdir(parents=True, exist_ok=True)

PIPE_COLORS = {"B": "#4477AA", "C": "#EE6677", "D": "#228833", "E": "#CCBB44"}
METHOD_COLORS = {
    "Age only": "#888888",
    "Pressure only": "#AA4499",
    "Pressure + raw AE": "#44AA99",
    "PRISM-Hazard": "#332288",
}

print("CLEAR-AE Stage 3")
print("Stage-1 directory:", CFG.stage1_dir)
print("Stage-2 directory:", CFG.stage2_dir)
print("Output directory :", CFG.output_dir)
print("Independent experimental units:", CFG.pipes, "(n = 4 pipes)")
print("Terminal hypothesis horizon:", CFG.terminal_horizon_min, "min")


# %% CELL 2 — Schema-safe loading and provenance audit

def read_table(path: Path) -> pd.DataFrame:
    if not path.exists():
        raise FileNotFoundError(path)
    if path.suffix == ".parquet":
        return pd.read_parquet(path)
    return pd.read_csv(path)


def read_stage1_time_table(cfg: ClearConfig = CFG) -> Tuple[pd.DataFrame, Path]:
    candidates = [
        cfg.stage1_dir / "03_prism_time_bins.parquet",
        cfg.stage1_dir / "03_prism_time_bins.csv.gz",
        cfg.stage1_dir / "03_prism_time_bins.csv",
    ]
    path = next((p for p in candidates if p.exists()), None)
    if path is None:
        raise FileNotFoundError(
            "Run PRISM-AE Stage 1 first; 03_prism_time_bins was not found in "
            f"{cfg.stage1_dir}"
        )
    table = read_table(path)
    required = {"pipe", "time_min", "pressure_pd"}
    missing = sorted(required - set(table.columns))
    if missing:
        raise ValueError(f"Stage-1 table is missing {missing}")
    table = table[table["pipe"].astype(str).isin(cfg.pipes)].copy()
    table["pipe"] = table["pipe"].astype(str)
    table = table.sort_values(["pipe", "time_min"]).reset_index(drop=True)
    return table, path


def require_stage2_file(name: str, cfg: ClearConfig = CFG) -> Path:
    path = cfg.stage2_dir / name
    if not path.exists():
        raise FileNotFoundError(
            f"Missing Stage-2 artifact {path}. Run PRISM-Hazard Stage 2 first."
        )
    return path


TIME_RAW, TIME_SOURCE = read_stage1_time_table(CFG)
ENDPOINTS = read_table(
    CFG.stage1_dir / "02_endpoint_candidates_NOT_ground_truth.csv"
)
PRED = read_table(require_stage2_file("03_lopo_hazard_predictions.csv"))
METRICS = read_table(require_stage2_file("04_hazard_metrics_by_pipe.csv"))
METRICS_SUMMARY = read_table(require_stage2_file("04_hazard_metrics_summary.csv"))
ENDPOINT_SENS = read_table(require_stage2_file("10_endpoint_sensitivity_summary.csv"))
STRESS = read_table(require_stage2_file("11_sensor_stress_summary.csv"))

manifest_path = require_stage2_file("12_stage2_reproducibility_manifest.json")
with open(manifest_path, "r", encoding="utf-8") as fh:
    STAGE2_MANIFEST = json.load(fh)

CFG.labels_are_physical = bool(
    STAGE2_MANIFEST.get("decision", STAGE2_MANIFEST.get("decision_gate", {})).get(
        "labels_are_physical", False
    )
)
CFG.pressure_alignment_is_external = bool(
    STAGE2_MANIFEST.get("decision", STAGE2_MANIFEST.get("decision_gate", {})).get(
        "pressure_alignment_is_external", False
    )
)

required_endpoint_cols = {
    "pipe",
    CFG.endpoint_lower_col,
    CFG.endpoint_upper_col,
}
missing = sorted(required_endpoint_cols - set(ENDPOINTS.columns))
if missing:
    raise ValueError(f"Endpoint candidate table is missing {missing}")

ENDPOINTS["pipe"] = ENDPOINTS["pipe"].astype(str)
ENDPOINTS = ENDPOINTS[ENDPOINTS["pipe"].isin(CFG.pipes)].copy()
ENDPOINTS["endpoint_lower_min"] = ENDPOINTS[CFG.endpoint_lower_col].astype(float)
ENDPOINTS["endpoint_upper_min"] = ENDPOINTS[CFG.endpoint_upper_col].astype(float)
ENDPOINTS["endpoint_width_min"] = (
    ENDPOINTS["endpoint_upper_min"] - ENDPOINTS["endpoint_lower_min"]
)
if (ENDPOINTS["endpoint_width_min"] < -1e-9).any():
    bad = ENDPOINTS.loc[
        ENDPOINTS["endpoint_width_min"] < 0,
        ["pipe", "endpoint_lower_min", "endpoint_upper_min"],
    ]
    raise ValueError(f"Endpoint interval is reversed:\n{bad}")

last_observed = TIME_RAW.groupby("pipe")["time_min"].max().rename("last_observed_min")
INPUT_AUDIT = ENDPOINTS[
    ["pipe", "endpoint_lower_min", "endpoint_upper_min", "endpoint_width_min"]
].merge(last_observed, on="pipe", how="left")
INPUT_AUDIT["unobserved_to_upper_min"] = np.maximum(
    0.0, INPUT_AUDIT["endpoint_upper_min"] - INPUT_AUDIT["last_observed_min"]
)
INPUT_AUDIT["labels_are_physical"] = CFG.labels_are_physical
INPUT_AUDIT["pressure_alignment_is_external"] = CFG.pressure_alignment_is_external
INPUT_AUDIT.to_csv(CFG.output_dir / "01_input_and_endpoint_provenance.csv", index=False)

print(f"Loaded {len(TIME_RAW):,} Stage-1 rows from {TIME_SOURCE}")
print(INPUT_AUDIT.round(3).to_string(index=False))
if not CFG.labels_are_physical:
    warnings.warn(
        "Endpoints are AE-derived inspection candidates. Failure probability and RUL "
        "are not physically identified."
    )
if not CFG.pressure_alignment_is_external:
    warnings.warn(
        "Pressure alignment is inferred rather than externally measured; clock/pressure "
        "effects must be treated as confounding."
    )


# %% CELL 3 — Common timeline, causal AE dynamics, and interval labels

def available_columns(table: pd.DataFrame, names: Sequence[str]) -> List[str]:
    return [c for c in names if c in table.columns]


PRESSURE_COLS = available_columns(TIME_RAW, CFG.pressure_features)
AE_COLS = available_columns(TIME_RAW, CFG.ae_features)
if len(AE_COLS) < 4:
    raise ValueError(
        f"Only {len(AE_COLS)} AE features found. Expected Stage-1 features such as "
        "log_rms_med and pooled_hf_logratio."
    )


def downsample_by_time(table: pd.DataFrame, step_min: float) -> pd.DataFrame:
    pieces = []
    for pipe, src in table.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        if d.empty:
            continue
        target = np.arange(d["time_min"].min(), d["time_min"].max() + 1e-9, step_min)
        idx = np.searchsorted(d["time_min"].to_numpy(float), target, side="left")
        idx = np.clip(idx, 0, len(d) - 1)
        idx = np.unique(idx)
        pieces.append(d.iloc[idx].copy())
    return pd.concat(pieces, ignore_index=True)


def add_causal_dynamics(
    table: pd.DataFrame,
    base_features: Sequence[str],
    lags_min: Sequence[float] = CFG.dynamic_lags_min,
) -> Tuple[pd.DataFrame, List[str]]:
    """Current values and strictly backward-looking deltas; never future windows."""
    pieces: List[pd.DataFrame] = []
    created = list(base_features)
    for pipe, src in table.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        dt = float(np.nanmedian(np.diff(d["time_min"]))) if len(d) > 1 else CFG.sample_minutes
        dt = max(dt, 1e-6)
        for col in base_features:
            x = d[col].astype(float)
            for lag in lags_min:
                n = max(1, int(round(lag / dt)))
                tag = str(lag).replace(".", "p")
                name = f"{col}__delta_{tag}m"
                d[name] = x - x.shift(n)
                if name not in created:
                    created.append(name)
        pieces.append(d)
    return pd.concat(pieces, ignore_index=True), created


TIME = downsample_by_time(
    TIME_RAW[TIME_RAW["time_min"] >= CFG.analysis_start_min].copy(),
    CFG.sample_minutes,
)
TIME = TIME.merge(
    ENDPOINTS[
        ["pipe", "endpoint_lower_min", "endpoint_upper_min", "endpoint_width_min"]
    ],
    on="pipe",
    how="left",
    validate="many_to_one",
)
TIME["life_fraction_lower"] = TIME["time_min"] / TIME["endpoint_lower_min"]


def attach_partial_label(table: pd.DataFrame, horizon_min: float) -> pd.DataFrame:
    """
    Coherent interval label for T in [L, U].

    definite_event: even the latest candidate U occurs within the horizon;
    definite_safe : even the earliest candidate L is beyond the horizon;
    ambiguous     : status depends on the unresolved endpoint location.
    """
    out = table.copy()
    earliest_rul = out["endpoint_lower_min"] - out["time_min"]
    latest_rul = out["endpoint_upper_min"] - out["time_min"]
    out["partial_label"] = "ambiguous"
    out.loc[earliest_rul > horizon_min, "partial_label"] = "definite_safe"
    out.loc[latest_rul <= horizon_min, "partial_label"] = "definite_event"
    out["event_lower_endpoint"] = (earliest_rul <= horizon_min).astype(int)
    out["event_upper_endpoint"] = (latest_rul <= horizon_min).astype(int)
    return out


partial_support = []
for horizon in CFG.horizons_min:
    d = attach_partial_label(TIME, horizon)
    a = (
        d.groupby(["pipe", "partial_label"])
        .size()
        .unstack(fill_value=0)
        .reindex(columns=["definite_safe", "ambiguous", "definite_event"], fill_value=0)
        .reset_index()
    )
    a["horizon_min"] = horizon
    partial_support.append(a)
PARTIAL_SUPPORT = pd.concat(partial_support, ignore_index=True)
PARTIAL_SUPPORT.to_csv(CFG.output_dir / "02_partial_label_support.csv", index=False)

print("Pressure features:", PRESSURE_COLS)
print("AE features      :", AE_COLS)
print("\nEndpoint-interval label support:")
print(PARTIAL_SUPPORT.to_string(index=False))


# %% CELL 4 — Clock dominance, short-horizon exception, and exact n=4 inference

def exact_sign_flip_test(differences: Sequence[float]) -> Dict[str, float]:
    """Exact randomization test over all 2^n sign assignments."""
    d = np.asarray(differences, dtype=float)
    observed = float(np.mean(d))
    null = []
    for signs in itertools.product((-1.0, 1.0), repeat=len(d)):
        null.append(float(np.mean(d * np.asarray(signs))))
    null = np.asarray(null)
    return {
        "n_pipes": int(len(d)),
        "mean_improvement": observed,
        "one_sided_p": float(np.mean(null >= observed - 1e-15)),
        "two_sided_p": float(np.mean(np.abs(null) >= abs(observed) - 1e-15)),
        "smallest_possible_one_sided_p": float(1.0 / (2 ** len(d))),
    }


def stage2_horizon_evidence(metrics: pd.DataFrame) -> Tuple[pd.DataFrame, pd.DataFrame]:
    pressure = metrics[metrics["method"] == "Pressure only"][
        ["pipe", "horizon_min", "Brier"]
    ].rename(columns={"Brier": "pressure_Brier"})
    ae = metrics[metrics["method"] == "Pressure + raw AE"][
        ["pipe", "horizon_min", "Brier"]
    ].rename(columns={"Brier": "ae_Brier"})
    pair = pressure.merge(ae, on=["pipe", "horizon_min"], validate="one_to_one")
    pair["Brier_improvement"] = pair["pressure_Brier"] - pair["ae_Brier"]
    pair["Brier_skill"] = pair["Brier_improvement"] / pair["pressure_Brier"]

    rows = []
    for horizon, d in pair.groupby("horizon_min"):
        test = exact_sign_flip_test(d["Brier_improvement"])
        rows.append(
            {
                "horizon_min": horizon,
                "pressure_Brier_mean": d["pressure_Brier"].mean(),
                "raw_AE_Brier_mean": d["ae_Brier"].mean(),
                "mean_Brier_improvement": d["Brier_improvement"].mean(),
                "relative_skill_from_group_means": 1.0
                - d["ae_Brier"].mean() / d["pressure_Brier"].mean(),
                "pipes_won": int((d["Brier_improvement"] > 0).sum()),
                **test,
            }
        )
    return pair, pd.DataFrame(rows)


PIPE_HORIZON_EVIDENCE, HORIZON_EVIDENCE = stage2_horizon_evidence(METRICS)
PIPE_HORIZON_EVIDENCE.to_csv(
    CFG.output_dir / "03_short_horizon_evidence_by_pipe.csv", index=False
)
HORIZON_EVIDENCE.to_csv(
    CFG.output_dir / "03_short_horizon_exact_inference.csv", index=False
)


def probability_collapse_audit(pred: pd.DataFrame) -> pd.DataFrame:
    rows = []
    for keys, d in pred.groupby(["outer_test_pipe", "method", "horizon_min"]):
        pipe, method, horizon = keys
        p = d["probability"].to_numpy(float)
        rows.append(
            {
                "pipe": pipe,
                "method": method,
                "horizon_min": horizon,
                "n": len(p),
                "probability_min": np.min(p),
                "probability_max": np.max(p),
                "probability_std": np.std(p),
                "near_constant_probability": bool(np.std(p) < 1e-4),
            }
        )
    return pd.DataFrame(rows)


COLLAPSE = probability_collapse_audit(PRED)
COLLAPSE.to_csv(CFG.output_dir / "03_probability_collapse_audit.csv", index=False)

print("Horizon-level incremental value (positive improvement favours AE):")
print(HORIZON_EVIDENCE.round(5).to_string(index=False))
print("\nNear-constant held-out probability trajectories:")
print(COLLAPSE[COLLAPSE["near_constant_probability"]].to_string(index=False))

fig, ax = plt.subplots(figsize=(8.3, 4.6))
for pipe, d in PIPE_HORIZON_EVIDENCE.groupby("pipe"):
    ax.plot(
        d["horizon_min"], d["Brier_skill"], "o-", lw=1.2,
        color=PIPE_COLORS.get(pipe), alpha=0.78, label=f"Pipe {pipe}",
    )
mean = PIPE_HORIZON_EVIDENCE.groupby("horizon_min")["Brier_skill"].mean()
ax.plot(mean.index, mean.values, "ko-", lw=2.6, label="Pipe-balanced mean")
ax.axhline(0, color="black", lw=1)
ax.set(
    xlabel="Failure horizon (min)",
    ylabel="Brier skill of raw AE vs pressure",
    title="A short-horizon AE exception that does not persist",
)
ax.legend(ncol=3, fontsize=8)
fig.tight_layout()
fig.savefig(CFG.output_dir / "03_short_horizon_exception.png")
plt.show()


# %% CELL 5 — Coherent endpoint-interval Brier and skill bounds

def coherent_endpoint_loss_curve(
    pred: pd.DataFrame,
    endpoint_lower: float,
    endpoint_upper: float,
    grid_size: int = 401,
) -> pd.DataFrame:
    """Evaluate one fixed prediction trajectory for coherent T across [L, U]."""
    horizon = float(pred["horizon_min"].iloc[0])
    tau_grid = np.linspace(endpoint_lower, endpoint_upper, grid_size)
    time = pred["time_min"].to_numpy(float)
    probability = pred["probability"].to_numpy(float)
    rows = []
    for tau in tau_grid:
        event = ((tau - time) <= horizon).astype(float)
        rows.append(
            {
                "endpoint_min": tau,
                "Brier": np.mean((probability - event) ** 2),
            }
        )
    return pd.DataFrame(rows)


ENDPOINT_LOOKUP = ENDPOINTS.set_index("pipe")[
    ["endpoint_lower_min", "endpoint_upper_min"]
].to_dict("index")

bound_rows = []
curve_rows = []
for keys, d in PRED.groupby(["outer_test_pipe", "method", "horizon_min"]):
    pipe, method, horizon = keys
    if pipe not in ENDPOINT_LOOKUP:
        continue
    bounds = ENDPOINT_LOOKUP[pipe]
    curve = coherent_endpoint_loss_curve(
        d,
        float(bounds["endpoint_lower_min"]),
        float(bounds["endpoint_upper_min"]),
    )
    curve["pipe"] = pipe
    curve["method"] = method
    curve["horizon_min"] = horizon
    curve_rows.append(curve)
    bound_rows.append(
        {
            "pipe": pipe,
            "method": method,
            "horizon_min": horizon,
            "Brier_best_case": curve["Brier"].min(),
            "Brier_worst_case": curve["Brier"].max(),
            "Brier_at_lower_endpoint": curve.iloc[0]["Brier"],
            "Brier_at_upper_endpoint": curve.iloc[-1]["Brier"],
            "endpoint_width_min": float(bounds["endpoint_upper_min"])
            - float(bounds["endpoint_lower_min"]),
        }
    )

ENDPOINT_CURVES = pd.concat(curve_rows, ignore_index=True)
ENDPOINT_BOUNDS = pd.DataFrame(bound_rows)
ENDPOINT_BOUNDS.to_csv(
    CFG.output_dir / "04_endpoint_interval_brier_bounds.csv", index=False
)


def paired_skill_bounds(curves: pd.DataFrame) -> pd.DataFrame:
    rows = []
    for (pipe, horizon), d in curves.groupby(["pipe", "horizon_min"]):
        p = d[d["method"] == "Pressure only"][
            ["endpoint_min", "Brier"]
        ].rename(columns={"Brier": "pressure_Brier"})
        a = d[d["method"] == "Pressure + raw AE"][
            ["endpoint_min", "Brier"]
        ].rename(columns={"Brier": "ae_Brier"})
        x = p.merge(a, on="endpoint_min", how="inner")
        if x.empty:
            continue
        x["improvement"] = x["pressure_Brier"] - x["ae_Brier"]
        rows.append(
            {
                "pipe": pipe,
                "horizon_min": horizon,
                "improvement_worst_case": x["improvement"].min(),
                "improvement_best_case": x["improvement"].max(),
                "AE_better_for_every_endpoint": bool((x["improvement"] > 0).all()),
                "AE_worse_for_every_endpoint": bool((x["improvement"] < 0).all()),
            }
        )
    pipe_bounds = pd.DataFrame(rows)
    group = (
        pipe_bounds.groupby("horizon_min")
        .agg(
            group_improvement_lower=("improvement_worst_case", "mean"),
            group_improvement_upper=("improvement_best_case", "mean"),
            pipes_robustly_better=("AE_better_for_every_endpoint", "sum"),
            pipes_robustly_worse=("AE_worse_for_every_endpoint", "sum"),
        )
        .reset_index()
    )
    return pipe_bounds.merge(group, on="horizon_min", how="left")


ENDPOINT_SKILL_BOUNDS = paired_skill_bounds(ENDPOINT_CURVES)
ENDPOINT_SKILL_BOUNDS.to_csv(
    CFG.output_dir / "04_endpoint_interval_incremental_value.csv", index=False
)

print("Endpoint-interval incremental-value bounds:")
print(ENDPOINT_SKILL_BOUNDS.round(6).to_string(index=False))

plot_horizons = [CFG.terminal_horizon_min, CFG.operational_horizon_min]
plot_methods = ["Pressure only", "Pressure + raw AE", "PRISM-Hazard"]
fig, axes = plt.subplots(1, len(plot_horizons), figsize=(12.0, 4.6), sharey=True)
for ax, horizon in zip(np.atleast_1d(axes), plot_horizons):
    d = ENDPOINT_BOUNDS[
        (ENDPOINT_BOUNDS["horizon_min"] == horizon)
        & ENDPOINT_BOUNDS["method"].isin(plot_methods)
    ].copy()
    d["x"] = d["pipe"].map({p: i for i, p in enumerate(CFG.pipes)})
    offsets = {m: (i - 1) * 0.18 for i, m in enumerate(plot_methods)}
    for method in plot_methods:
        q = d[d["method"] == method]
        x = q["x"].to_numpy(float) + offsets[method]
        mid = 0.5 * (q["Brier_best_case"] + q["Brier_worst_case"])
        err = 0.5 * (q["Brier_worst_case"] - q["Brier_best_case"])
        ax.errorbar(
            x, mid, yerr=err, fmt="o", capsize=3,
            color=METHOD_COLORS[method], label=method,
        )
    ax.set_xticks(range(len(CFG.pipes)), CFG.pipes)
    ax.set_xlabel("Held-out pipe")
    ax.set_title(f"{horizon:g}-minute horizon")
    ax.axhline(0, color="black", lw=0.7)
axes[0].set_ylabel("Coherent endpoint-interval Brier range")
axes[0].legend(fontsize=8)
fig.suptitle("How much can endpoint ambiguity change the conclusion?", y=1.02)
fig.tight_layout()
fig.savefig(CFG.output_dir / "04_endpoint_partial_identification.png")
plt.show()


# %% CELL 6 — Is AE a persistent pipe/sensor fingerprint?

def robust_residualize_columns(
    train: pd.DataFrame,
    test: pd.DataFrame,
    target_cols: Sequence[str],
    nuisance_cols: Sequence[str],
) -> Tuple[np.ndarray, np.ndarray]:
    """Fit nuisance response only on train; return train/test residual matrices."""
    nuisance_cols = available_columns(train, nuisance_cols)
    Xtr = train[nuisance_cols].replace([np.inf, -np.inf], np.nan)
    Xte = test[nuisance_cols].replace([np.inf, -np.inf], np.nan)
    fill = Xtr.median(numeric_only=True)
    Xtr = Xtr.fillna(fill)
    Xte = Xte.fillna(fill)
    rtr, rte = [], []
    for col in target_cols:
        y = train[col].to_numpy(float)
        finite = np.isfinite(y)
        model = Pipeline(
            [
                ("impute", SimpleImputer(strategy="median")),
                ("scale", StandardScaler()),
                ("ridge", Ridge(alpha=10.0)),
            ]
        )
        model.fit(Xtr.loc[finite], y[finite])
        rtr.append(y - model.predict(Xtr))
        rte.append(test[col].to_numpy(float) - model.predict(Xte))
    return np.column_stack(rtr), np.column_stack(rte)


def make_fingerprint_blocks(table: pd.DataFrame, n_blocks: int) -> pd.DataFrame:
    out = table.copy()
    out["fingerprint_block"] = -1
    for pipe, idx in out.groupby("pipe").groups.items():
        d = out.loc[idx].sort_values("life_fraction_lower")
        ranks = np.arange(len(d))
        block = np.minimum(n_blocks - 1, (ranks * n_blocks // max(1, len(d))))
        out.loc[d.index, "fingerprint_block"] = block
    return out


FINGERPRINT_TABLE = make_fingerprint_blocks(TIME.copy(), CFG.fingerprint_blocks)
fp_rows, fp_truth, fp_pred = [], [], []
for held_block in range(CFG.fingerprint_blocks):
    train = FINGERPRINT_TABLE[FINGERPRINT_TABLE["fingerprint_block"] != held_block].copy()
    test = FINGERPRINT_TABLE[FINGERPRINT_TABLE["fingerprint_block"] == held_block].copy()
    # Age/pressure nuisance response is learned without held temporal block.
    nuisance = [
        "life_fraction_lower", "pressure_pd", "time_in_hold_min",
        "delta_pressure_pd", "abs_delta_pressure_pd",
    ]
    Xtr, Xte = robust_residualize_columns(train, test, AE_COLS, nuisance)
    model = Pipeline(
        [
            ("impute", SimpleImputer(strategy="median", add_indicator=True)),
            ("scale", StandardScaler()),
            (
                "classifier",
                LogisticRegression(
                    C=1.0,
                    solver="lbfgs",
                    max_iter=4000,
                    class_weight="balanced",
                    random_state=SEED,
                ),
            ),
        ]
    )
    model.fit(Xtr, train["pipe"])
    yhat = model.predict(Xte)
    fp_truth.extend(test["pipe"].tolist())
    fp_pred.extend(yhat.tolist())
    fp_rows.append(
        {
            "held_temporal_block": held_block,
            "n_test": len(test),
            "accuracy": accuracy_score(test["pipe"], yhat),
            "balanced_accuracy": balanced_accuracy_score(test["pipe"], yhat),
        }
    )

FINGERPRINT_FOLDS = pd.DataFrame(fp_rows)
FINGERPRINT_FOLDS.to_csv(
    CFG.output_dir / "05_pipe_fingerprint_blocked_cv.csv", index=False
)
FP_CM = confusion_matrix(fp_truth, fp_pred, labels=list(CFG.pipes), normalize="true")
fingerprint_summary = {
    "n_pipes": len(CFG.pipes),
    "chance_accuracy": 1.0 / len(CFG.pipes),
    "blocked_accuracy": float(accuracy_score(fp_truth, fp_pred)),
    "blocked_balanced_accuracy": float(balanced_accuracy_score(fp_truth, fp_pred)),
    "interpretation": (
        "High accuracy means AE retains pipe/sensor identity after linear age/pressure "
        "adjustment; it is not evidence of transferable damage."
    ),
}
with open(CFG.output_dir / "05_pipe_fingerprint_summary.json", "w", encoding="utf-8") as fh:
    json.dump(fingerprint_summary, fh, indent=2)

print(json.dumps(fingerprint_summary, indent=2))
fig, ax = plt.subplots(figsize=(5.3, 4.6))
sns.heatmap(
    FP_CM,
    annot=True,
    fmt=".2f",
    cmap="mako",
    vmin=0,
    vmax=1,
    xticklabels=CFG.pipes,
    yticklabels=CFG.pipes,
    ax=ax,
)
ax.set(
    xlabel="Predicted pipe identity",
    ylabel="True pipe identity",
    title="Pressure/age-adjusted AE still identifies the pipe",
)
fig.tight_layout()
fig.savefig(CFG.output_dir / "05_pipe_fingerprint_confusion.png")
plt.show()


# %% CELL 7 — Pressure-preserving AE negative-control engine

def pressure_blocks(d: pd.DataFrame) -> np.ndarray:
    """Contiguous load regimes, allowing small ramp increments within a regime."""
    p = d["pressure_pd"].to_numpy(float)
    scale = max(float(np.nanmax(p) - np.nanmin(p)), EPS)
    jump_tolerance = max(1e-6, 0.03 * scale)
    change = np.r_[True, np.abs(np.diff(p)) > jump_tolerance]
    return np.cumsum(change) - 1


def make_negative_control(
    table: pd.DataFrame,
    control: str,
    seed: int,
    ae_cols: Sequence[str] = AE_COLS,
) -> pd.DataFrame:
    rng = np.random.default_rng(seed)
    out = table.copy()
    if control == "observed":
        return out

    if control in {"within_pressure_permutation", "within_pressure_circular_shift"}:
        pieces = []
        for pipe, src in out.groupby("pipe", sort=False):
            d = src.sort_values("time_min").copy()
            d["_pressure_block"] = pressure_blocks(d)
            if control == "within_pressure_permutation":
                # Approximate pressure strata remain usable for continuous ramps.
                ranks = d["pressure_pd"].rank(method="first", pct=True)
                d["_control_group"] = np.minimum(11, np.floor(12 * ranks)).astype(int)
            else:
                d["_control_group"] = d["_pressure_block"]
            for _, idx in d.groupby("_control_group").groups.items():
                idx = np.asarray(list(idx))
                values = d.loc[idx, ae_cols].to_numpy(copy=True)
                n = len(values)
                if n < 3:
                    continue
                if control == "within_pressure_permutation":
                    order = rng.permutation(n)
                else:
                    lo = max(1, int(math.ceil(0.20 * n)))
                    hi = max(lo + 1, int(math.floor(0.80 * n)) + 1)
                    shift = int(rng.integers(lo, hi)) if hi > lo else 1
                    order = np.roll(np.arange(n), shift)
                d.loc[idx, ae_cols] = values[order]
            pieces.append(d.drop(columns=["_pressure_block", "_control_group"]))
        return pd.concat(pieces, ignore_index=True)

    if control == "pipe_swap_by_life_fraction":
        pipes = list(sorted(out["pipe"].unique()))
        offset = int(rng.integers(1, len(pipes)))
        donor = {p: pipes[(i + offset) % len(pipes)] for i, p in enumerate(pipes)}
        pieces = []
        for pipe, src in out.groupby("pipe", sort=False):
            d = src.sort_values("life_fraction_lower").copy()
            q = out[out["pipe"] == donor[pipe]].sort_values("life_fraction_lower")
            target_x = d["life_fraction_lower"].to_numpy(float)
            donor_x = q["life_fraction_lower"].to_numpy(float)
            for col in ae_cols:
                d[col] = np.interp(
                    target_x,
                    donor_x,
                    q[col].to_numpy(float),
                    left=q[col].iloc[0],
                    right=q[col].iloc[-1],
                )
            pieces.append(d)
        return pd.concat(pieces, ignore_index=True)

    raise KeyError(control)


def pipe_balanced_weights(groups: Sequence[str]) -> np.ndarray:
    s = pd.Series(np.asarray(groups, dtype=str))
    counts = s.value_counts()
    w = s.map({p: 1.0 / n for p, n in counts.items()}).to_numpy(float)
    return w / np.mean(w)


def make_logistic() -> Pipeline:
    return Pipeline(
        [
            ("impute", SimpleImputer(strategy="median", add_indicator=True)),
            ("scale", StandardScaler()),
            (
                "model",
                LogisticRegression(
                    C=CFG.logistic_c,
                    solver="lbfgs",
                    max_iter=4000,
                    random_state=SEED,
                ),
            ),
        ]
    )


def evaluate_control_lopo(
    table: pd.DataFrame,
    horizon_min: float,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    d = table.copy()
    d["event"] = (
        d["endpoint_lower_min"] - d["time_min"] <= horizon_min
    ).astype(int)
    d, ae_dynamic = add_causal_dynamics(d, AE_COLS)
    feature_sets = {
        "Pressure only": PRESSURE_COLS,
        "Pressure + raw AE": list(dict.fromkeys(PRESSURE_COLS + ae_dynamic)),
    }
    pred_rows = []
    for held in CFG.pipes:
        train = d[d["pipe"] != held].copy()
        test = d[d["pipe"] == held].copy()
        for method, features in feature_sets.items():
            model = make_logistic()
            model.fit(
                train[features],
                train["event"],
                model__sample_weight=pipe_balanced_weights(train["pipe"]),
            )
            p = model.predict_proba(test[features])[:, 1]
            block = test[["pipe", "time_min", "event"]].copy()
            block["probability"] = np.clip(p, 1e-6, 1 - 1e-6)
            block["method"] = method
            block["outer_test_pipe"] = held
            pred_rows.append(block)
    pred = pd.concat(pred_rows, ignore_index=True)
    metrics = (
        pred.groupby(["outer_test_pipe", "method"])
        .apply(
            lambda x: pd.Series(
                {
                    "n": len(x),
                    "Brier": np.mean((x["probability"] - x["event"]) ** 2),
                }
            ),
            include_groups=False,
        )
        .reset_index()
    )
    return pred, metrics


def improvement_from_control_metrics(metrics: pd.DataFrame) -> Tuple[float, int]:
    p = metrics[metrics["method"] == "Pressure only"].set_index("outer_test_pipe")[
        "Brier"
    ]
    a = metrics[
        metrics["method"] == "Pressure + raw AE"
    ].set_index("outer_test_pipe")["Brier"]
    improvement = p - a
    return float(improvement.mean()), int((improvement > 0).sum())


OBSERVED_CONTROL_PRED, OBSERVED_CONTROL_METRICS = evaluate_control_lopo(
    make_negative_control(TIME, "observed", SEED), CFG.terminal_horizon_min
)
observed_improvement, observed_wins = improvement_from_control_metrics(
    OBSERVED_CONTROL_METRICS
)

control_rows = [
    {
        "control": "observed",
        "repeat": 0,
        "mean_Brier_improvement": observed_improvement,
        "pipes_won": observed_wins,
    }
]
for control in CFG.negative_controls:
    print("Running negative control:", control)
    for repeat in range(CFG.negative_control_repeats):
        altered = make_negative_control(
            TIME,
            control,
            seed=SEED + 10000 * (1 + CFG.negative_controls.index(control)) + repeat,
        )
        _, metrics = evaluate_control_lopo(altered, CFG.terminal_horizon_min)
        improvement, wins = improvement_from_control_metrics(metrics)
        control_rows.append(
            {
                "control": control,
                "repeat": repeat,
                "mean_Brier_improvement": improvement,
                "pipes_won": wins,
            }
        )

NEGATIVE_CONTROLS = pd.DataFrame(control_rows)
NEGATIVE_CONTROLS.to_csv(
    CFG.output_dir / "06_pressure_preserving_negative_controls.csv", index=False
)

negative_control_summary = []
for control in CFG.negative_controls:
    null = NEGATIVE_CONTROLS.loc[
        NEGATIVE_CONTROLS["control"] == control, "mean_Brier_improvement"
    ].to_numpy(float)
    negative_control_summary.append(
        {
            "control": control,
            "observed_improvement": observed_improvement,
            "null_mean": np.mean(null),
            "null_q025": np.quantile(null, 0.025),
            "null_q975": np.quantile(null, 0.975),
            "empirical_one_sided_p": (1 + np.sum(null >= observed_improvement))
            / (1 + len(null)),
        }
    )
NEGATIVE_CONTROL_SUMMARY = pd.DataFrame(negative_control_summary)
NEGATIVE_CONTROL_SUMMARY.to_csv(
    CFG.output_dir / "06_negative_control_summary.csv", index=False
)
print(NEGATIVE_CONTROL_SUMMARY.round(6).to_string(index=False))

fig, ax = plt.subplots(figsize=(9.0, 4.8))
order = list(CFG.negative_controls)
sns.violinplot(
    data=NEGATIVE_CONTROLS[NEGATIVE_CONTROLS["control"] != "observed"],
    x="control",
    y="mean_Brier_improvement",
    order=order,
    inner="quartile",
    cut=0,
    color="#9ecae1",
    ax=ax,
)
ax.axhline(observed_improvement, color="#D55E00", lw=2.2, label="Observed AE")
ax.axhline(0, color="black", lw=1)
ax.set_xticks(range(len(order)))
ax.set_xticklabels(["Within-pressure\npermutation", "Within-pressure\ncircular shift", "Pipe swap at\nmatched life fraction"])
ax.set(
    ylabel="Mean held-out-pipe Brier improvement",
    xlabel="Pressure-preserving negative control",
    title="Does the 2-minute AE signal beat schedule-preserving nulls?",
)
ax.legend()
fig.tight_layout()
fig.savefig(CFG.output_dir / "06_negative_control_falsification.png")
plt.show()


# %% CELL 8 — Label-free, held-out-pipe terminal change detector

def robust_mad(x: np.ndarray, axis: Optional[int] = None) -> np.ndarray:
    x = np.asarray(x, dtype=float)
    med = np.nanmedian(x, axis=axis, keepdims=True)
    return 1.4826 * np.nanmedian(np.abs(x - med), axis=axis) + EPS


class TrainPipePressureEnvelope:
    """Expected AE under load, fitted only to non-terminal training-pipe rows."""

    def __init__(self, pressure_cols: Sequence[str], ae_cols: Sequence[str]):
        self.pressure_cols = list(pressure_cols)
        self.ae_cols = list(ae_cols)
        self.fill: Optional[pd.Series] = None
        self.models: Dict[str, HistGradientBoostingRegressor] = {}

    def fit(self, table: pd.DataFrame) -> "TrainPipePressureEnvelope":
        healthy = table[
            table["endpoint_lower_min"] - table["time_min"] > CFG.healthy_guard_min
        ].copy()
        X = healthy[self.pressure_cols].replace([np.inf, -np.inf], np.nan)
        self.fill = X.median(numeric_only=True)
        X = X.fillna(self.fill)
        weights = pipe_balanced_weights(healthy["pipe"])
        for col in self.ae_cols:
            y = healthy[col].to_numpy(float)
            finite = np.isfinite(y)
            model = HistGradientBoostingRegressor(
                loss="absolute_error",
                learning_rate=0.05,
                max_iter=140,
                max_leaf_nodes=7,
                min_samples_leaf=max(10, int(0.03 * finite.sum())),
                l2_regularization=4.0,
                random_state=SEED,
            )
            model.fit(X.loc[finite], y[finite], sample_weight=weights[finite])
            self.models[col] = model
        return self

    def transform(self, table: pd.DataFrame) -> pd.DataFrame:
        if self.fill is None:
            raise RuntimeError("Pressure envelope is not fitted")
        out = table.copy()
        X = out[self.pressure_cols].replace([np.inf, -np.inf], np.nan).fillna(self.fill)
        for col, model in self.models.items():
            out[f"innovation_{col}"] = out[col].to_numpy(float) - model.predict(X)
        return out


def causal_change_score(
    table: pd.DataFrame,
    innovation_cols: Sequence[str],
    reference_min: float,
) -> pd.DataFrame:
    pieces = []
    for pipe, src in table.groupby("pipe", sort=False):
        d = src.sort_values("time_min").copy()
        dt = float(np.nanmedian(np.diff(d["time_min"]))) if len(d) > 1 else CFG.sample_minutes
        n = max(5, int(round(reference_min / max(dt, 1e-6))))
        z_cols = []
        for col in innovation_cols:
            x = d[col].astype(float)
            # Shift one bin so the current observation never enters its reference.
            ref_med = x.shift(1).rolling(n, min_periods=max(5, n // 3)).median()
            abs_dev = (x.shift(1) - ref_med).abs()
            ref_mad = 1.4826 * abs_dev.rolling(
                n, min_periods=max(5, n // 3)
            ).median()
            z = (x - ref_med).abs() / (ref_mad + 0.05)
            name = f"z_{col}"
            d[name] = z.clip(0, 50)
            z_cols.append(name)
        # Median is robust to one bad sensor-derived summary; 90th percentile
        # remains sensitive to a coordinated spectral/RMS shift.
        d["change_score"] = np.nanpercentile(d[z_cols], 75, axis=1)
        pieces.append(d)
    return pd.concat(pieces, ignore_index=True)


def event_starts(mask: np.ndarray) -> np.ndarray:
    mask = np.asarray(mask, dtype=bool)
    return np.flatnonzero(mask & ~np.r_[False, mask[:-1]])


def persistent_mask(score: np.ndarray, threshold: float, persistence: int) -> np.ndarray:
    above = np.asarray(score, dtype=float) >= threshold
    if persistence <= 1:
        return above
    run = np.convolve(above.astype(int), np.ones(persistence, dtype=int), mode="full")[: len(above)]
    return run >= persistence


def false_event_rate(
    table: pd.DataFrame,
    threshold: float,
    persistence: int,
    guard_min: float,
) -> float:
    events, hours = 0, 0.0
    for pipe, d in table.groupby("pipe"):
        q = d[d["endpoint_lower_min"] - d["time_min"] > guard_min].sort_values("time_min")
        if len(q) < 2:
            continue
        alarm = persistent_mask(q["change_score"].to_numpy(float), threshold, persistence)
        events += len(event_starts(alarm))
        hours += (q["time_min"].iloc[-1] - q["time_min"].iloc[0]) / 60.0
    return float(events / max(hours, EPS))


CHANGE_ROWS = []
CHANGE_TRAJECTORIES = []
for held in CFG.pipes:
    train = TIME[TIME["pipe"] != held].copy()
    test = TIME[TIME["pipe"] == held].copy()
    envelope = TrainPipePressureEnvelope(PRESSURE_COLS, AE_COLS).fit(train)
    train_r = envelope.transform(train)
    test_r = envelope.transform(test)
    innovation_cols = [f"innovation_{c}" for c in AE_COLS]
    train_s = causal_change_score(train_r, innovation_cols, CFG.causal_reference_min)
    test_s = causal_change_score(test_r, innovation_cols, CFG.causal_reference_min)

    healthy_scores = train_s.loc[
        train_s["endpoint_lower_min"] - train_s["time_min"] > CFG.healthy_guard_min,
        "change_score",
    ].dropna()
    candidates = []
    for quantile in CFG.change_quantiles:
        threshold = float(healthy_scores.quantile(quantile))
        for persistence in CFG.change_persistence_bins:
            far = false_event_rate(
                train_s, threshold, persistence, CFG.healthy_guard_min
            )
            candidates.append(
                {
                    "quantile": quantile,
                    "threshold": threshold,
                    "persistence_bins": persistence,
                    "training_false_events_per_h": far,
                    "feasible": far <= CFG.target_false_events_per_h,
                }
            )
    cand = pd.DataFrame(candidates)
    feasible = cand[cand["feasible"]]
    if feasible.empty:
        choice = cand.sort_values(
            ["training_false_events_per_h", "threshold", "persistence_bins"]
        ).iloc[0]
    else:
        # Most sensitive threshold among policies satisfying the false-event limit.
        choice = feasible.sort_values(
            ["threshold", "persistence_bins"], ascending=[True, True]
        ).iloc[0]

    test_s = test_s.sort_values("time_min").copy()
    alarm = persistent_mask(
        test_s["change_score"].to_numpy(float),
        float(choice["threshold"]),
        int(choice["persistence_bins"]),
    )
    starts = event_starts(alarm)
    lower = float(test_s["endpoint_lower_min"].iloc[0])
    upper = float(test_s["endpoint_upper_min"].iloc[0])
    terminal_start = lower - CFG.terminal_horizon_min
    terminal_candidates = [i for i in starts if test_s["time_min"].iloc[i] >= terminal_start]
    detection_i = terminal_candidates[0] if terminal_candidates else None
    detection_time = (
        float(test_s["time_min"].iloc[detection_i]) if detection_i is not None else np.nan
    )
    early_events = sum(test_s["time_min"].iloc[i] < terminal_start for i in starts)
    observed_hours = max(
        EPS,
        (terminal_start - float(test_s["time_min"].min())) / 60.0,
    )
    CHANGE_ROWS.append(
        {
            "outer_test_pipe": held,
            "threshold": float(choice["threshold"]),
            "persistence_bins": int(choice["persistence_bins"]),
            "training_false_events_per_h": float(choice["training_false_events_per_h"]),
            "detected_terminal_window": detection_i is not None,
            "detection_time_min": detection_time,
            "lead_to_endpoint_lower_min": lower - detection_time if detection_i is not None else np.nan,
            "lead_to_endpoint_upper_min": upper - detection_time if detection_i is not None else np.nan,
            "preterminal_false_events": int(early_events),
            "preterminal_false_events_per_h": early_events / observed_hours,
        }
    )
    test_s["outer_test_pipe"] = held
    test_s["threshold"] = float(choice["threshold"])
    test_s["alarm"] = alarm
    CHANGE_TRAJECTORIES.append(test_s)

CHANGE_RESULTS = pd.DataFrame(CHANGE_ROWS)
CHANGE_PRED = pd.concat(CHANGE_TRAJECTORIES, ignore_index=True)
CHANGE_RESULTS.to_csv(
    CFG.output_dir / "07_label_free_change_detection_by_pipe.csv", index=False
)
CHANGE_PRED.to_csv(
    CFG.output_dir / "07_label_free_change_trajectories.csv.gz",
    index=False,
    compression="gzip",
)
print(CHANGE_RESULTS.round(4).to_string(index=False))

fig, axes = plt.subplots(len(CFG.pipes), 1, figsize=(12.0, 9.5), sharex=False)
for ax, pipe in zip(axes, CFG.pipes):
    d = CHANGE_PRED[CHANGE_PRED["pipe"] == pipe].sort_values("time_min")
    ax.plot(d["time_min"], d["change_score"], color=PIPE_COLORS[pipe], lw=1)
    ax.axhline(d["threshold"].iloc[0], color="#D55E00", ls="--", lw=1.1)
    lower = d["endpoint_lower_min"].iloc[0]
    upper = d["endpoint_upper_min"].iloc[0]
    ax.axvspan(
        lower - CFG.terminal_horizon_min,
        upper,
        color="#CC6677",
        alpha=0.13,
        label="Endpoint-ambiguous terminal region" if pipe == CFG.pipes[0] else None,
    )
    starts = event_starts(d["alarm"].to_numpy(bool))
    if len(starts):
        ax.scatter(
            d["time_min"].iloc[starts],
            d["change_score"].iloc[starts],
            marker="v", s=25, color="black", zorder=4,
        )
    ax.set_ylabel(f"Pipe {pipe}\nscore")
axes[-1].set_xlabel("Time from feature start (min)")
axes[0].legend(fontsize=8)
fig.suptitle(
    "Held-out-pipe pressure-conditioned change detection (no test endpoint used)",
    y=0.995,
)
fig.tight_layout()
fig.savefig(CFG.output_dir / "07_label_free_terminal_change_detection.png")
plt.show()


# %% CELL 9 — Reinterpret the existing sequential alarms and sensor stress

ALARM_PATH = CFG.stage2_dir / "08_alarm_results_by_pipe.csv"
if not ALARM_PATH.exists():
    ALARM_PATH = CFG.stage2_dir / "08_alarm_policy_lopo.csv"
ALARMS = read_table(ALARM_PATH)
# Stage 2 stores row-level names; the console handoff used summary names.
alarm_detect_col = "detection_rate" if "detection_rate" in ALARMS else "detected"
alarm_lead_col = (
    "mean_lead_time_min" if "mean_lead_time_min" in ALARMS else "lead_time_min"
)

alarm_summary = (
    ALARMS[ALARMS["horizon_min"] == CFG.operational_horizon_min]
    .groupby("method")
    .agg(
        detection_rate=(alarm_detect_col, "mean"),
        mean_lead_time_min=(alarm_lead_col, "mean"),
        false_alarm_events_per_h=("false_alarm_events_per_h", "mean"),
    )
    .reset_index()
)
if "Pressure only" in set(alarm_summary["method"]):
    base = alarm_summary.set_index("method").loc["Pressure only"]
    alarm_summary["lead_gain_vs_pressure_min"] = (
        alarm_summary["mean_lead_time_min"] - base["mean_lead_time_min"]
    )
    alarm_summary["false_event_gain_vs_pressure_per_h"] = (
        base["false_alarm_events_per_h"] - alarm_summary["false_alarm_events_per_h"]
    )
alarm_summary.to_csv(CFG.output_dir / "08_alarm_incremental_value.csv", index=False)

stress_clean = STRESS[STRESS["scenario"] == "clean"][
    ["method", "Brier_mean"]
].rename(columns={"Brier_mean": "clean_Brier"})
stress_audit = STRESS.merge(stress_clean, on="method", how="left")
stress_audit["recomputed_relative_degradation"] = (
    stress_audit["Brier_mean"] / stress_audit["clean_Brier"] - 1.0
)
stress_audit.to_csv(CFG.output_dir / "08_sensor_stress_audit.csv", index=False)

print("Operational alarm value beyond pressure:")
print(alarm_summary.round(4).to_string(index=False))
print("\nSensor stress:")
print(stress_audit.round(4).to_string(index=False))


# %% CELL 10 — Synthetic identifiability experiment

def simulate_world(
    world: str,
    seed: int,
    n_pipes: int = CFG.simulation_pipes,
    n_points: int = CFG.simulation_points_per_pipe,
) -> pd.DataFrame:
    """
    Three worlds with identical clock/load structure.

    clock_only: AE has pressure response and pipe identity but no damage signal.
    transferable_damage: a common terminal damage coordinate enters AE.
    pipe_specific_damage: damage enters with pipe-specific sign/gain, defeating
    transfer even though each individual pipe shows a terminal change.
    """
    rng = np.random.default_rng(seed)
    rows = []
    for i in range(n_pipes):
        pipe = f"S{i:02d}"
        endpoint = float(rng.uniform(88, 116))
        time = np.linspace(0, endpoint, n_points)
        onset = float(rng.uniform(8, 18))
        rate = float(rng.uniform(0.70, 1.15))
        pressure = np.maximum(0, time - onset) * rate
        pressure += 4.0 * np.floor(np.maximum(0, time - onset) / 18.0)
        pressure /= max(np.max(pressure), EPS)
        damage = np.clip((time - (endpoint - 12.0)) / 12.0, 0, 1) ** 2
        pipe_offset = rng.normal(0, 1.2, size=4)
        pressure_gain = rng.normal([1.4, 0.9, -0.6, 0.4], 0.12, size=4)
        if world == "clock_only":
            damage_gain = np.zeros(4)
        elif world == "transferable_damage":
            damage_gain = np.array([4.0, 3.2, -3.0, 3.6])
        elif world == "pipe_specific_damage":
            damage_gain = rng.normal(0, 2.2, size=4)
        else:
            raise KeyError(world)
        ae = (
            pipe_offset[None, :]
            + pressure[:, None] * pressure_gain[None, :]
            + damage[:, None] * damage_gain[None, :]
            + rng.normal(0, 0.30, size=(n_points, 4))
        )
        for j in range(n_points):
            rows.append(
                {
                    "pipe": pipe,
                    "time_min": time[j],
                    "endpoint_lower_min": endpoint,
                    "pressure_pd": pressure[j],
                    "ae_1": ae[j, 0],
                    "ae_2": ae[j, 1],
                    "ae_3": ae[j, 2],
                    "ae_4": ae[j, 3],
                }
            )
    return pd.DataFrame(rows)


def simulation_lopo(table: pd.DataFrame, horizon_min: float = 5.0) -> Dict[str, float]:
    pressure_cols = ["time_min", "pressure_pd"]
    ae_cols = ["ae_1", "ae_2", "ae_3", "ae_4"]
    d = table.copy()
    # Online installation calibration: each unseen pipe contributes only its own
    # early reference segment, never its event label.  This removes arbitrary
    # sensor offsets while retaining terminal changes.
    normalized_cols = []
    for col in ae_cols:
        name = f"{col}_from_early_baseline"
        d[name] = np.nan
        for pipe, idx in d.groupby("pipe").groups.items():
            q = d.loc[idx].sort_values("time_min")
            cutoff = q["time_min"].min() + 0.20 * (
                q["time_min"].max() - q["time_min"].min()
            )
            baseline = q.loc[q["time_min"] <= cutoff, col].to_numpy(float)
            med = float(np.median(baseline))
            scale = float(robust_mad(baseline))
            d.loc[q.index, name] = (q[col].to_numpy(float) - med) / scale
        normalized_cols.append(name)
    d["event"] = (d["endpoint_lower_min"] - d["time_min"] <= horizon_min).astype(int)
    pipes = sorted(d["pipe"].unique())
    # Six held-out groups reduce runtime while preserving pipe-disjoint evaluation.
    folds = [pipes[i::6] for i in range(6)]
    losses = {"Pressure only": [], "Pressure + AE": []}
    for held in folds:
        train = d[~d["pipe"].isin(held)]
        test = d[d["pipe"].isin(held)]
        for method, cols in {
            "Pressure only": pressure_cols,
            "Pressure + AE": pressure_cols + normalized_cols,
        }.items():
            model = make_logistic()
            model.fit(
                train[cols],
                train["event"],
                model__sample_weight=pipe_balanced_weights(train["pipe"]),
            )
            probability = model.predict_proba(test[cols])[:, 1]
            for pipe, idx in test.groupby("pipe").groups.items():
                y = test.loc[idx, "event"].to_numpy(int)
                p = probability[test.index.get_indexer(idx)]
                losses[method].append(np.mean((p - y) ** 2))
    pressure = float(np.mean(losses["Pressure only"]))
    ae = float(np.mean(losses["Pressure + AE"]))
    return {
        "pressure_Brier": pressure,
        "AE_Brier": ae,
        "Brier_improvement": pressure - ae,
        "Brier_skill": 1.0 - ae / pressure,
    }


SIMULATION_ROWS = []
for world in ["clock_only", "transferable_damage", "pipe_specific_damage"]:
    print("Simulating:", world)
    for repeat in range(CFG.simulation_repeats):
        sim = simulate_world(world, SEED + 50000 + 1000 * repeat + len(world))
        result = simulation_lopo(sim)
        SIMULATION_ROWS.append({"world": world, "repeat": repeat, **result})

SIMULATIONS = pd.DataFrame(SIMULATION_ROWS)
SIMULATIONS.to_csv(CFG.output_dir / "09_synthetic_identifiability.csv", index=False)
simulation_summary = (
    SIMULATIONS.groupby("world")
    .agg(
        skill_mean=("Brier_skill", "mean"),
        skill_q025=("Brier_skill", lambda x: np.quantile(x, 0.025)),
        skill_q975=("Brier_skill", lambda x: np.quantile(x, 0.975)),
        probability_positive_skill=("Brier_skill", lambda x: np.mean(x > 0)),
    )
    .reset_index()
)
simulation_summary.to_csv(
    CFG.output_dir / "09_synthetic_identifiability_summary.csv", index=False
)
print(simulation_summary.round(4).to_string(index=False))

fig, ax = plt.subplots(figsize=(8.4, 4.7))
sns.violinplot(
    data=SIMULATIONS,
    x="world",
    y="Brier_skill",
    order=["clock_only", "transferable_damage", "pipe_specific_damage"],
    inner="quartile",
    cut=0,
    palette=["#999999", "#44AA99", "#CC6677"],
    ax=ax,
)
ax.axhline(0, color="black", lw=1)
ax.set_xticks(range(3))
ax.set_xticklabels(["Clock only", "Transferable\ndamage", "Pipe-specific\ndamage"])
ax.set(
    xlabel="Simulated data-generating world",
    ylabel="Held-out-pipe Brier skill of AE",
    title="CLEAR-AE distinguishes terminal signal from transferable prognostic value",
)
fig.tight_layout()
fig.savefig(CFG.output_dir / "09_synthetic_identifiability.png")
plt.show()


# %% CELL 11 — CLEAR-AE evidence certificate and publication decision

terminal_row = HORIZON_EVIDENCE.loc[
    np.isclose(HORIZON_EVIDENCE["horizon_min"], CFG.terminal_horizon_min)
].iloc[0]
operational_row = HORIZON_EVIDENCE.loc[
    np.isclose(HORIZON_EVIDENCE["horizon_min"], CFG.operational_horizon_min)
].iloc[0]
terminal_bounds = ENDPOINT_SKILL_BOUNDS.loc[
    np.isclose(ENDPOINT_SKILL_BOUNDS["horizon_min"], CFG.terminal_horizon_min)
]
stress_selected = stress_audit[
    (stress_audit["method"] == "Pressure + raw AE")
    & (stress_audit["scenario"] == "combined")
]
combined_stress = (
    float(stress_selected["relative_Brier_degradation_mean"].iloc[0])
    if len(stress_selected)
    else np.nan
)

negative_pass = bool(
    (NEGATIVE_CONTROL_SUMMARY["empirical_one_sided_p"] <= 0.05).all()
)
endpoint_robust = bool(
    len(terminal_bounds)
    and terminal_bounds["AE_better_for_every_endpoint"].all()
)
fingerprint_pass = bool(
    fingerprint_summary["blocked_balanced_accuracy"]
    <= fingerprint_summary["chance_accuracy"] + 0.15
)
stress_pass = bool(np.isfinite(combined_stress) and combined_stress <= 0.25)

CERTIFICATE = pd.DataFrame(
    [
        {
            "gate": "Physical endpoint provenance",
            "passed": CFG.labels_are_physical,
            "value": str(CFG.labels_are_physical),
            "requirement": "Externally verified rupture or inspection timestamp",
        },
        {
            "gate": "External pressure alignment",
            "passed": CFG.pressure_alignment_is_external,
            "value": str(CFG.pressure_alignment_is_external),
            "requirement": "Measured synchronized pressure log",
        },
        {
            "gate": "Operational incremental value",
            "passed": operational_row["mean_Brier_improvement"] > 0,
            "value": f"10-min improvement={operational_row['mean_Brier_improvement']:.5f}",
            "requirement": "AE Brier lower than pressure under strict LOPO",
        },
        {
            "gate": "Short-horizon pipe-level evidence",
            "passed": terminal_row["one_sided_p"] <= 0.05,
            "value": (
                f"wins={int(terminal_row['pipes_won'])}/4, exact one-sided "
                f"p={terminal_row['one_sided_p']:.3f}"
            ),
            "requirement": "Exact pipe-level p <= 0.05",
        },
        {
            "gate": "Endpoint partial-identification robustness",
            "passed": endpoint_robust,
            "value": (
                f"group improvement interval="
                f"[{terminal_bounds['group_improvement_lower'].iloc[0]:.5f}, "
                f"{terminal_bounds['group_improvement_upper'].iloc[0]:.5f}]"
                if len(terminal_bounds) else "unavailable"
            ),
            "requirement": "AE improvement positive for all coherent endpoints",
        },
        {
            "gate": "Pressure-preserving negative controls",
            "passed": negative_pass,
            "value": "all empirical p <= 0.05" if negative_pass else "one or more nulls not rejected",
            "requirement": "Observed AE beats every schedule-preserving null",
        },
        {
            "gate": "Pipe-fingerprint acceptability",
            "passed": fingerprint_pass,
            "value": f"balanced accuracy={fingerprint_summary['blocked_balanced_accuracy']:.3f}",
            "requirement": "Pipe identity near chance after pressure/age adjustment",
        },
        {
            "gate": "Combined sensor-stress robustness",
            "passed": stress_pass,
            "value": f"relative Brier degradation={combined_stress:.3f}",
            "requirement": "Relative degradation <= 0.25",
        },
    ]
)
CERTIFICATE.to_csv(CFG.output_dir / "10_CLEAR_AE_evidence_certificate.csv", index=False)

terminal_hypothesis = bool(
    terminal_row["mean_Brier_improvement"] > 0
    and int(terminal_row["pipes_won"]) >= 3
)
failure_identified = bool(CERTIFICATE["passed"].all())

if failure_identified:
    verdict = "IDENTIFIED — EXTERNAL VALIDATION STILL REQUIRED"
elif terminal_hypothesis:
    verdict = "TERMINAL CHANGE HYPOTHESIS ONLY — FAILURE/RUL NON-IDENTIFIABLE"
else:
    verdict = "NON-IDENTIFIABLE — REPORT DATA LIMITATION BENCHMARK"

DECISION = {
    "framework": "CLEAR-AE Stage 3",
    "verdict": verdict,
    "failure_or_RUL_claim_allowed": failure_identified,
    "terminal_change_hypothesis_supported_descriptively": terminal_hypothesis,
    "terminal_horizon_min": CFG.terminal_horizon_min,
    "terminal_raw_AE_relative_skill": float(
        terminal_row["relative_skill_from_group_means"]
    ),
    "terminal_pipes_won_out_of_4": int(terminal_row["pipes_won"]),
    "terminal_exact_one_sided_p": float(terminal_row["one_sided_p"]),
    "operational_10min_raw_AE_relative_skill": float(
        operational_row["relative_skill_from_group_means"]
    ),
    "probability_collapse_pipes_PRISM_10min": COLLAPSE.loc[
        (COLLAPSE["method"] == "PRISM-Hazard")
        & np.isclose(COLLAPSE["horizon_min"], CFG.operational_horizon_min)
        & COLLAPSE["near_constant_probability"],
        "pipe",
    ].tolist(),
    "pipe_fingerprint_balanced_accuracy": float(
        fingerprint_summary["blocked_balanced_accuracy"]
    ),
    "combined_sensor_stress_relative_degradation": combined_stress,
    "labels_are_physical": CFG.labels_are_physical,
    "pressure_alignment_is_external": CFG.pressure_alignment_is_external,
    "recommended_paper_claim": (
        "A leakage-, confounding-, and endpoint-aware audit demonstrates that apparent "
        "pipeline prognostic accuracy is largely a pressure/age clock. AE contains a "
        "descriptive near-terminal regime signature, but transferable failure risk and "
        "RUL are not identified by these four tests."
    ),
    "mandatory_new_data": [
        "synchronized measured pressure for every test",
        "independently logged physical rupture or inspection endpoint",
        "additional pipes spanning pressure schedules and sensor installations",
        "pre-registered held-out-pipe evaluation",
    ],
}
with open(CFG.output_dir / "10_CLEAR_AE_decision.json", "w", encoding="utf-8") as fh:
    json.dump(DECISION, fh, indent=2)

print(CERTIFICATE.to_string(index=False))
print("\n" + json.dumps(DECISION, indent=2))

fig, ax = plt.subplots(figsize=(10.5, 5.2))
plot = CERTIFICATE.iloc[::-1].copy()
colors = plot["passed"].map({True: "#228833", False: "#CC3311"})
ax.barh(np.arange(len(plot)), 1, color=colors, alpha=0.86)
ax.set_yticks(np.arange(len(plot)), plot["gate"])
ax.set_xlim(0, 1)
ax.set_xticks([])
for y, (_, row) in enumerate(plot.iterrows()):
    ax.text(0.02, y, "PASS" if row["passed"] else "FAIL", va="center", color="white", weight="bold")
    ax.text(0.17, y, str(row["value"]), va="center", color="white", fontsize=8.5)
ax.set_title("CLEAR-AE prognostic identifiability certificate")
for spine in ax.spines.values():
    spine.set_visible(False)
fig.tight_layout()
fig.savefig(CFG.output_dir / "10_CLEAR_AE_evidence_certificate.png")
plt.show()


# %% CELL 12 — Reproducibility manifest and exact result handoff

def sha256(path: Path, chunk_size: int = 2**20) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as fh:
        while True:
            chunk = fh.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)
    return h.hexdigest()


input_files = [
    TIME_SOURCE,
    CFG.stage1_dir / "02_endpoint_candidates_NOT_ground_truth.csv",
    CFG.stage2_dir / "03_lopo_hazard_predictions.csv",
    CFG.stage2_dir / "04_hazard_metrics_by_pipe.csv",
    CFG.stage2_dir / "10_endpoint_sensitivity_summary.csv",
    CFG.stage2_dir / "11_sensor_stress_summary.csv",
    ALARM_PATH,
]
manifest = {
    "project": "CLEAR-AE Stage 3",
    "seed": SEED,
    "config": {**asdict(CFG), **{
        "stage1_dir": str(CFG.stage1_dir),
        "stage2_dir": str(CFG.stage2_dir),
        "output_dir": str(CFG.output_dir),
    }},
    "python": sys.version,
    "platform": platform.platform(),
    "packages": {
        "numpy": np.__version__,
        "pandas": pd.__version__,
        "matplotlib": mpl.__version__,
        "seaborn": sns.__version__,
    },
    "independent_unit": "pipe",
    "n_independent_units": len(CFG.pipes),
    "input_sha256": {str(p): sha256(p) for p in input_files if p.exists()},
    "decision": DECISION,
}
with open(CFG.output_dir / "11_reproducibility_manifest.json", "w", encoding="utf-8") as fh:
    json.dump(manifest, fh, indent=2, default=str)

print(
    """
CLEAR-AE STAGE 3 COMPLETE
-------------------------

Primary manuscript tables and figures:
  03_short_horizon_exact_inference.csv
  03_short_horizon_exception.png
  04_endpoint_interval_incremental_value.csv
  04_endpoint_partial_identification.png
  05_pipe_fingerprint_confusion.png
  06_negative_control_falsification.png
  07_label_free_terminal_change_detection.png
  08_alarm_incremental_value.csv
  09_synthetic_identifiability.png
  10_CLEAR_AE_evidence_certificate.csv
  10_CLEAR_AE_evidence_certificate.png
  10_CLEAR_AE_decision.json
  11_reproducibility_manifest.json

Interpretation discipline:
  * A positive 2-minute result is a terminal-change hypothesis, not RUL.
  * No failure-probability claim is allowed while labels_are_physical=False.
  * Rows are not independent replicates; pipe-level effects and exact n=4
    inference are primary.
  * A negative result is scientifically useful when framed as a general,
    reproducible identifiability audit rather than another failed model.
"""
)
```
